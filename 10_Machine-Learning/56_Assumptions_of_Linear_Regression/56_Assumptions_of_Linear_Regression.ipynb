##  ****1- Linear Relationship****



* Sabse pehle LR perform karne ke liye jo aapko sabse pehla assumption chahiye wo yeh hai ki aapke data me:


* Aapke data me Linear Relationship hona chahiye, input aur output (target) me linear relationship hona chahiye.


* Jaise pehle graph me positive linear relationship dekh rahe ho.


* Second me negative linear relationship hai.


* Third me not a linear relationship (curved/non-linear) hai.

* Maan lo aapke paas multiple input columns hain, tab bhi aapke har ek individual input feature ka output (target) ke sath linear relationship hona hi chahiye.

___

## ******2. No Multicollinearity:******


* Yahan pe hum dekhte hain ki aapke system me koi bhi multicollinearity nahi honi chahiye, aur sub col ek dusre  me nahi honi chahiye.


* **Multicollinearity ka simple matlab:** Yahan pe jitne bhi input columns honge, unke aapas ke beech mein koi strong relation nahi hona chahiye—bas yahi simple concept hai.

* Lekin Input columns ka Output ($Y$) ke sath accha relation hona chahiye (Linear Relationship).
