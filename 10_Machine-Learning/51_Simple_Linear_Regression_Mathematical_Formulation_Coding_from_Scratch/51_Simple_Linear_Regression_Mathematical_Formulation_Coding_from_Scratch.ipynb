## ****How to Find $m$ and $b$ (or $c$) in a Linear Equation?****

Linear equation ($y = mx + c$) mein slope ($m$) aur intercept ($c$) calculate karne ke do primary approaches/solutions hote hain:

### ****1. Closed-Form Solution (Analytical Approach)****

* **What is it?** Closed-form solution ek direct mathematical formula hota hai jisme hum direct values feed karke exact optimal parameters ($m$ aur $c$) calculate kar lete hain. Isme kisi iterative optimization loop ki zaroorat nahi padti, direct formula solve karo aur answer aa jata hai.

* **Key Point:** Yahan hume real-time code execution ke dauran bar-bar differentiation ya integration chalane ki zaroorat nahi hoti, derived direct formula use hota hai (jise OLS - Ordinary Least Squares formula bolte hain).

* **Important Note:** Scikit-Learn ka `LinearRegression` class internally yahi closed-form solution (OLS technique) use karta hai parameters find karne ke liye.

---

### *****2. Non-Closed Form Solution (Iterative Optimization Approach)*****

* **What is it?** Jab dataset bahut bada hota hai ya parameters directly solve nahi ho sakte, tab hum iterative techniques use karte hain.
* **Key Technique:** Yahan hum parameters ko approximate aur optimize karne ke liye **Gradient Descent** algorithm ka use karte hain.

* ***Jab Data Bahut bada higher Dimension Me Chala Jata Hai Waha pe hame Ye Approch lagni padti hai Esliye haam Tab Fomule Se yaha pe Shift Hote hai***

* ***Agar aapko `Gradient Descent` ke through linear regression perform karna hai, toh aap Scikit-Learn ka `SGDRegressor (Stochastic Gradient Descent Regressor)` use kar sakte hain.***

### ****Now Fot this Note Book We Focus On OLS To Find M and C Which Can Use Direct Fomula ⬇️⬇️⬇️****

___




******Ordinary Least Squares (OLS) closed-form solution mein hum direct in formulas ka use karke slope ($m$) aur intercept ($c$) calculate karte hain:******

#### **1. Formula for Slope ($m$):**

$$m = \frac{\sum_{i=1}^{n} (x_i - \bar{x})(y_i - \bar{y})}{\sum_{i=1}^{n} (x_i - \bar{x})^2}$$

*Jahan:*

* $x_i$ aur $y_i$ = Individual data points hain.
* $\bar{x}$ = Feature $x$ ka mean (average) hai.(jaise Tumne CGPa Nikala tha uska mean hai yaha pe honga)
* $\bar{y}$ = Target $y$ ka mean (average) hai.(Yaha pe Package ka)
* $n$ = Total number of observations / rows hain.




 #### **2. Formula for Intercept ($c$ or $b$):**

$$C = \bar{y} - M\bar{x}$$

***Jahan:***

* $\bar{y}$ = ***Target values ka mean hai.***
* $\bar{x}$ = ***Feature values ka mean hai.***
* $M$ = ***Calculate kiya hua slope hai.***

  <h3 style="color: #a6e3a1;">🔍 Total Error ($E$) Kaise Nikalein Aur Kyun?</h3>
  <p>
    Graph me har point ke liye vertical distance hai: 
    <code style="background:#313244; color:#f9e2af; padding: 2px 6px; border-radius: 4px;">dᵢ = (yᵢ - ŷᵢ)</code>, jahan $y_i$ actual value hai aur $\hat{y}_i$ predicted line ki value hai. Ab total error calculate karne ke 3 approaches ho sakte hain:
  </p>
  <div style="margin-left: 10px;">
    <!-- Option 1 -->
    <div style="margin-bottom: 15px; padding: 12px; background: #24273a; border-left: 4px solid #f38ba8; border-radius: 6px;">
      <b style="color: #f38ba8;">1. Simple Sum: <code>E = d₁ + d₂ + d₃ + ... + dₙ</code>  ❌ (Rejected)</b><br>
      Line ke upar wale points ke liye distance positive (+) hoga aur line ke neeche wale points ke liye negative (-) hoga. Agar hum simple sum karenge, toh positive aur negative values ek dusre ko <b>cancel out</b> kar dengi aur total error zero ke barabar aa sakta hai, jabki actually line bilkul kharab fit ho sakti hai.
    </div>
    <!-- Option 2 -->
    <div style="margin-bottom: 15px; padding: 12px; background: #24273a; border-left: 4px solid #f9e2af; border-radius: 6px;">
      <b style="color: #f9e2af;">2. Absolute Sum: <code>E = |d₁| + |d₂| + |d₃| + ... + |dₙ|</code> (MAE) ⚠️</b><br>
      Cancellation problem solve karne ke liye modulus $|d_i|$ le sakte hain. Lekin calculus me modulus function zero point ($x=0$) par <b>non-differentiable</b> hota hai (wahan sharp corner banta hai). Is wajah se derivative nikaal kar slope aur intercept derive karna mathematically complex ho jata hai.
    </div>
    <!-- Option 3 -->
    <div style="margin-bottom: 15px; padding: 12px; background: #24273a; border-left: 4px solid #a6e3a1; border-radius: 6px;">
      <b style="color: #a6e3a1;">3. Sum of Squared Errors: <code>E = d₁² + d₂² + d₃² + ... + dₙ²</code>  (Selected - OLS)</b><br>
      Yeh hamara best approach hai jise <b>Ordinary Least Squares (OLS)</b> kehte hain:
      <ul style="margin: 8px 0 0 18px; padding: 0;">
        <li>Square karne se sabhi negative distances positive ban jate hain, toh cancellation nahi hota.</li>
        <li>Bade errors ko yeh heavily penalize karta hai (jaise error 4 hai toh square hoke 16 ho jayega).</li>
        <li>Yeh function <b>smooth aur continuous</b> hota hai, jiska partial differentiation ($\frac{\partial E}{\partial m}$ aur $\frac{\partial E}{\partial b}$) nikaal kar $0$ ke barabar equate karna aur direct $m$ aur $b$ ka formula derive karna bohot easy ho jata hai.</li>
      </ul>
    </div>
