## ****How to Find $m$ and $b$ (or $c$) in a Linear Equation?****

Linear equation ($y = mx + c$) mein slope ($m$) aur intercept ($c$) calculate karne ke do primary approaches/solutions hote hain:

### ****1. Closed-Form Solution (Analytical Approach)****

* **What is it?** Closed-form solution ek direct mathematical formula hota hai jisme hum direct values feed karke exact optimal parameters ($m$ aur $c$) calculate kar lete hain. Isme kisi iterative optimization loop ki zaroorat nahi padti, direct formula solve karo aur answer aa jata hai.

* **Key Point:** Yahan hume real-time code execution ke dauran bar-bar differentiation ya integration chalane ki zaroorat nahi hoti, derived direct formula use hota hai (jise OLS - Ordinary Least Squares formula bolte hain).

* **Important Note:** Scikit-Learn ka `LinearRegression` class internally yahi closed-form solution (OLS technique) use karta hai parameters find karne ke liye.

---

### *****2. Non-Closed Form Solution (Iterative Optimization Approach)*****

* **What is it?** Jab dataset bahut bada hota hai ya parameters directly solve nahi ho sakte, tab hum iterative techniques use karte hain.
* **Key Technique:** Yahan hum parameters ko approximate aur optimize karne ke liye **Gradient Descent** algorithm ka use karte hain.
