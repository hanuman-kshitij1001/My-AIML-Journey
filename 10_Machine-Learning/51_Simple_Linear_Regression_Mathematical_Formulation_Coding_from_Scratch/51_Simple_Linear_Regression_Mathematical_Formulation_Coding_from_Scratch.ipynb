## ****How to Find $m$ and $b$ (or $c$) in a Linear Equation?****

Linear equation ($y = mx + c$) mein slope ($m$) aur intercept ($c$) calculate karne ke do primary approaches/solutions hote hain:

### ****1. Closed-Form Solution (Analytical Approach)****

* **What is it?** Closed-form solution ek direct mathematical formula hota hai jisme hum direct values feed karke exact optimal parameters ($m$ aur $c$) calculate kar lete hain. Isme kisi iterative optimization loop ki zaroorat nahi padti, direct formula solve karo aur answer aa jata hai.

* **Key Point:** Yahan hume real-time code execution ke dauran bar-bar differentiation ya integration chalane ki zaroorat nahi hoti, derived direct formula use hota hai (jise OLS - Ordinary Least Squares formula bolte hain).

* **Important Note:** Scikit-Learn ka `LinearRegression` class internally yahi closed-form solution (OLS technique) use karta hai parameters find karne ke liye.

---

### *****2. Non-Closed Form Solution (Iterative Optimization Approach)*****

* **What is it?** Jab dataset bahut bada hota hai ya parameters directly solve nahi ho sakte, tab hum iterative techniques use karte hain.
* **Key Technique:** Yahan hum parameters ko approximate aur optimize karne ke liye **Gradient Descent** algorithm ka use karte hain.

* ***Jab Data Bahut bada higher Dimension Me Chala Jata Hai Waha pe hame Ye Approch lagni padti hai Esliye haam Tab Fomule Se yaha pe Shift Hote hai***

* ***Agar aapko `Gradient Descent` ke through linear regression perform karna hai, toh aap Scikit-Learn ka `SGDRegressor (Stochastic Gradient Descent Regressor)` use kar sakte hain.***

******Ordinary Least Squares (OLS) closed-form solution mein hum direct in formulas ka use karke slope ($m$) aur intercept ($c$) calculate karte hain:******

#### **1. Formula for Slope ($m$):**

$$m = \frac{\sum_{i=1}^{n} (x_i - \bar{x})(y_i - \bar{y})}{\sum_{i=1}^{n} (x_i - \bar{x})^2}$$

*Jahan:*

* $x_i$ aur $y_i$ = Individual data points hain.
* $\bar{x}$ = Feature $x$ ka mean (average) hai.(jaise Tumne CGPa Nikala tha uska mean hai yaha pe honga)
* $\bar{y}$ = Target $y$ ka mean (average) hai.(Yaha pe Package ka)
* $n$ = Total number of observations / rows hain.




 #### **2. Formula for Intercept ($c$ or $b$):**

$$C = \bar{y} - M\bar{x}$$

***Jahan:***

* $\bar{y}$ = ***Target values ka mean hai.***
* $\bar{x}$ = ***Feature values ka mean hai.***
* $M$ = ***Calculate kiya hua slope hai.***

## ****Humne Modulus ($\vert{}d\vert{}$) ke Bajaye Square ($d^2$) Kyun Liya?****

* **Reason 1:** Mai outliers ko, jo bahut door wale points hain, unko heavily penalize karna chahta hoon.
* **Reason 2:** Hame aage chal kar ek jagah pe jaake pure $E$ expression ka differentiation karna hai with respect to $m$ aur $b$ ($\frac{\partial E}{\partial m} = 0$ aur $\frac{\partial E}{\partial b} = 0$).
* **Reason 3:** Modulus ($\vert{}x\vert{}$) ka jo graph hota hai wo continuous toh hota hai, but origin par differentiable nahi hota hai. Jabki jab aap square lete ho, toh square ko aap kahin bhi kisi bhi point pe differentiate kar sakte ho, that's why humne square liya.

***Par boss, yahan pe $m$ aur $b$ toh dikh hi nahi raha hai? Arey chupa hua hai, dhyan se dekho:***

$$\hat{y}_i = mx_i + b$$

**Distance $d_i$ hota hai actual value minus predicted value:**

$$d_i = (y_i - \hat{y}_i)$$

**Toh total error ka formula ban gaya:**

$$E = \sum_{i=1}^{n} (y_i - \hat{y}_i)^2$$

**Ab agar hum $\hat{y}_i$ ki jagah $(mx_i + b)$ replace kar dein:**

$$E(m, b) = \sum_{i=1}^{n} (y_i - (mx_i + b))^2$$

$$E(m, b) = \sum_{i=1}^{n} (y_i - mx_i - b)^2$$

**Ab saaf-saaf dikh raha hai ki error $E$ actually $m$ aur $b$ ka function hai!**

## ****Minima Find Karna: Partial Differentiation****

Calculus ke according, jab bhi hume kisi single variable function $E(x)$ ka maxima ya minima nikalna hota hai, toh hum uska derivative nikal kar zero ke barabar equate karte hain:

$$\frac{dE}{dx} = 0$$

Lekin hamare case mein Error function $E$ do variables par depend karta hai: slope ($m$) aur intercept ($b$). Yani yeh ek multi-variable function hai:

$$f(m, b)$$

Isliye hum yahan **Partial Differentiation** use karenge.

Hume $E(m, b)$ ka minimum point (minima) find karne ke liye dono variables ke respect mein partially differentiate karke $0$ se equate karna padega:

$$\frac{\partial E}{\partial m} = 0 \quad \text{aur} \quad \frac{\partial E}{\partial b} = 0$$

In dono equations ko solve karne se hume optimal $(m, b)$ ki exact values mil jayengi jahan par total error sabse minimum hoga.

# Simple Linear Regression: Step-by-Step Derivation & Formulations

In Simple Linear Regression, we try to fit a straight line $y = mx + b$ that best fits our data points $(x_i, y_i)$. 

Our primary goal is to find the values of **slope ($m$)** and **y-intercept ($b$)** such that the total error between actual and predicted values is **minimized**.

---

### 1. Defining the Loss Function (Error Function)
To calculate the total error, we use the **Sum of Squared Errors (SSE)** / **Cost Function** $E$:

$$E(m, b) = \sum_{i=1}^{n} (y_i - \hat{y}_i)^2$$

Where the predicted value $\hat{y}_i = m x_i + b$. Substituting this gives:

$$E(m, b) = \sum_{i=1}^{n} \left( y_i - (m x_i + b) \right)^2$$

> **Why minimize $E(m, b)$?**  
> We square the residual errors $(y_i - \hat{y}_i)$ to make all distances positive (preventing positive and negative errors from canceling each other out) and to penalize larger outliers more heavily.

---

### 2. Deriving Intercept ($b$)
To minimize the Loss Function $E$ with respect to $b$, we take the partial derivative $\frac{\partial E}{\partial b}$ and set it to `0`:

$$\frac{\partial E}{\partial b} = \frac{\partial}{\partial b} \sum_{i=1}^{n} (y_i - m x_i - b)^2 = 0$$

Applying the Chain Rule:

$$-2 \sum_{i=1}^{n} (y_i - m x_i - b) = 0$$

Dividing both sides by $-2$:

$$\sum_{i=1}^{n} (y_i - m x_i - b) = 0$$

Splitting the summation across terms:

$$\sum_{i=1}^{n} y_i - m \sum_{i=1}^{n} x_i - \sum_{i=1}^{n} b = 0$$

Since $\sum_{i=1}^{n} b = n \cdot b$, we divide the entire equation by $n$:

$$\frac{1}{n}\sum_{i=1}^{n} y_i - m \left(\frac{1}{n}\sum_{i=1}^{n} x_i\right) - b = 0$$

Since $\bar{y} = \frac{1}{n}\sum y_i$ and $\bar{x} = \frac{1}{n}\sum x_i$:

$$\bar{y} - m\bar{x} - b = 0$$

$$\mathbf{b = \bar{y} - m\bar{x}}$$

---

### 3. Deriving Slope ($m$)
To minimize $E$ with respect to $m$, we set the partial derivative $\frac{\partial E}{\partial m}$ to `0`:

$$\frac{\partial E}{\partial m} = \frac{\partial}{\partial m} \sum_{i=1}^{n} \left( y_i - m x_i - b \right)^2 = 0$$

Substituting $b = \bar{y} - m\bar{x}$ into the equation:

$$\frac{\partial E}{\partial m} \sum_{i=1}^{n} \left( y_i - m x_i - (\bar{y} - m\bar{x}) \right)^2 = 0$$

$$\frac{\partial E}{\partial m} \sum_{i=1}^{n} \left( (y_i - \bar{y}) - m(x_i - \bar{x}) \right)^2 = 0$$

Applying the Chain Rule:

$$-2 \sum_{i=1}^{n} \left[ (y_i - \bar{y}) - m(x_i - \bar{x}) \right] (x_i - \bar{x}) = 0$$

Dividing by $-2$ and distributing $(x_i - \bar{x})$:

$$\sum_{i=1}^{n} (y_i - \bar{y})(x_i - \bar{x}) - m \sum_{i=1}^{n} (x_i - \bar{x})^2 = 0$$

Solving directly for $m$:

$$\mathbf{m = \frac{\sum_{i=1}^{n} (x_i - \bar{x})(y_i - \bar{y})}{\sum_{i=1}^{n} (x_i - \bar{x})^2}}$$

---

| **Intercept ($b$)** | $b = \bar{y} - m\bar{x}$ | Point where line crosses Y-axis |
