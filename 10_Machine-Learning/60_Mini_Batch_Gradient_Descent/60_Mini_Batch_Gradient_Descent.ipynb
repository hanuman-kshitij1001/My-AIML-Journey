# ****Mini-Batch Gradient Descent****

Mini-Batch Gradient Descent is a combination of **Batch Gradient Descent** and **Stochastic Gradient Descent (SGD)**.

### 🔹 ****How it works?****

Dataset ko small groups (**batches**) me divide karte hain, aur **har batch ke baad weights update** karte hain.

**Example:**

- Total rows = `1000`
- Batch Size = `100`

$$
\text{Number of Batches} = \frac{1000}{100} = 10
$$

So,

**1 Epoch = 10 Weight Updates**

### ****Comparison****

| Method | Rows per Update | Updates / Epoch |
|---|---:|---:|
| **Batch GD** | All rows | 1 |
| **SGD** | 1 row | 1000 |
| **Mini-Batch GD** | 100 rows | 10 |

###  ****Intuition****

**Batch GD** → पूरा data → update  
**SGD** → 1 row → update  
**Mini-Batch GD** → small group → update

> Mini-Batch gives a good balance between **speed, memory and stability**.# 🚀 Mini-Batch Gradient Descent
