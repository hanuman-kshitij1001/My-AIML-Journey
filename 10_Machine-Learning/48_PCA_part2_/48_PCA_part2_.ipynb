### ***4. Problem Formulation: PCA Mathematical Foundation***

* **Problem Statement:** 
  * Maan lo hamare paas 2D me dataset hai aur hume use reduce karke **1D me convert** karna hai.
  * Data 2D space me spread out hai. Humara objective yeh hai ki hume wo **naya Axis (Direction)** find karna hai jiske upar data ko project karne par utna hi accha result aur distinction mile jitna original 2D space me mil raha tha.
* **The Direction / Unit Vector:**
  * Kisi bhi direction ko define karne ke liye hum ek **Unit Vector ($\vec{u}$)** lete hain, jiska magnitude $1$ hota hai ($|\vec{u}| = 1$).
* **Projection of a Point on Unit Vector:**
  * Har data point $\vec{x} = (x, y)$ ko is unit vector $\vec{u}$ par perpendicular (orthogonal) drop karke project karte hain.
  * Dot product formula se projection length nikalte hain:
    $$\text{Projection} = \frac{\vec{u} \cdot \vec{x}}{|\vec{u}|}$$
  * Since $\vec{u}$ ek unit vector hai ($|\vec{u}| = 1$):
    $$\text{Projection} = \vec{u} \cdot \vec{x} = u^T x$$
  * Matrix multiplication form:
    $$\begin{bmatrix} x_1 & y_1 \end{bmatrix} \begin{bmatrix} x_2 \\ y_2 \end{bmatrix} = x_1 x_2 + y_1 y_2 \quad (\text{Scalar value / 1D coordinate})$$

### ***📐 Finding the Variance on the Projected Axis***

* **Objective:**
  * Unit vector $\vec{u}$ koi bhi direction ho sakti hai (multiple possible directions/arrows).
  * Hum saare data points $x_1, x_2, \dots, x_n$ ko is vector $\vec{u}$ par project karenge.
  * Har projected point ki 1D value banegi: **$u^T x_i$**.
  * Aur in projected points ka mean banega: **$u^T \bar{x}$**.
* **Variance Formula Transformation:**
  * Normal 1D space me variance ka formula hota hai:
    $$\text{Variance} = \frac{1}{n} \sum_{i=1}^{n} (x_i - \bar{x})^2$$
  * Jab hum data points ko unit vector $\vec{u}$ par project karte hain:
    * Purana point $x_i \longrightarrow$ Naya projected point **$u^T x_i$**
    * Purana mean $\bar{x} \longrightarrow$ Naya projected mean **$u^T \bar{x}$**
  * Toh projected axis par variance ka mathematical formula ban jata hai:
    $$\mathbf{\text{Variance}} = \frac{\sum_{i=1}^{n} (u^T x_i - u^T \bar{x})^2}{n}$$
* **The Optimization Goal of PCA:**
  * Hum us unit vector $\vec{u}$ ko choose karenge jiske liye is formula ka **Variance sabse maximum** nikal kar aaye!

### ***Covariance and Covariance Matrix — Core Foundations***

* **Mean (Central Tendency):** Mean basically central tendency ka measure hota hai jo yeh batata hai ki data ka center kahan par hai. Lekin sirf Mean ko use karke data kitna spread out hai, yeh nahi bata sakte.
* **Variance (The Limitation):** Is problem ko solve karne ke liye **Variance** aaya. Par variance ki bhi ek badi limit hai: variance hamesha ek single variable/axis par nikalta hai.
* **Missing Link:** Variance yeh kabhi nahi batata ki $X$ aur $Y$ ke beech me kaisa relation hai (matlab do alag-alag variables aapas me ek saath kaise change ho rahe hain). Is relation ko samajhne ke liye **Covariance** ki zaroorat padti hai.

* **Explanation (Figures 3 & 4 — The Variance Dilemma):**
  * **Dataset 3:** Points hain $\{(-1, 1), (0, 0), (1, -1)\}$
    * $X$ values: $\{-1, 0, 1\} \longrightarrow \text{Mean} = 0, \quad \text{Var}(X) = \frac{(-1)^2 + 0^2 + 1^2}{3} = \frac{2}{3} \approx 0.67$
    * $Y$ values: $\{1, 0, -1\} \longrightarrow \text{Mean} = 0, \quad \text{Var}(Y) = \frac{1^2 + 0^2 + (-1)^2}{3} = \frac{2}{3} \approx 0.67$
  * **Dataset 4:** Points hain $\{(-1, -1), (0, 0), (1, 1)\}$
    * $X$ values: $\{-1, 0, 1\} \longrightarrow \text{Mean} = 0, \quad \text{Var}(X) = \frac{2}{3} \approx 0.67$
    * $Y$ values: $\{-1, 0, 1\} \longrightarrow \text{Mean} = 0, \quad \text{Var}(Y) = \frac{2}{3} \approx 0.67$
  * **The Conflict:**
    Dono datasets me $\text{Var}(X)$ same hai aur $\text{Var}(Y)$ bhi same hai!
    Lekin Figure 3 me $X$ badhne par $Y$ ghat raha hai (**Negative Relationship**), aur Figure 4 me $X$ badhne par $Y$ badh raha hai (**Positive Relationship**).
  * **Conclusion:** Sirf individual variance dekh kar aap dono dataset me differentiate nahi kar sakte. Yahi par **Covariance** picture me aata hai jo do variables ka joint behavior batata hai.

___

Covariance do variables ke beech ka **direction of relationship** batata hai—yaani jab ek variable badhta hai, toh dusra variable kaisa behave karta hai.

---

**Covariance Ka Formula**

$$\text{Cov}(X, Y) = \frac{\sum_{i=1}^{n} (x_i - \bar{x})(y_i - \bar{y})}{n}$$

* $(x_i - \bar{x})$: $X$ ka apne mean se distance
* $(y_i - \bar{y})$: $Y$ ka apne mean se distance
* Dono distances ka product aapas me multiply hota hai

---

**Example 1: Positive Relationship (Figure 4 Wala Data)**

Data points: $\{(-1, -1), (0, 0), (1, 1)\}$

* Mean of $X$ ($\bar{x}$) = $0$
* Mean of $Y$ ($\bar{y}$) = $0$

Har point ka product calculate karte hain:

* Point $(-1, -1)$: $(-1 - 0) \times (-1 - 0) = (-1) \times (-1) = \mathbf{+1}$
* Point $(0, 0)$: $(0 - 0) \times (0 - 0) = 0 \times 0 = \mathbf{0}$
* Point $(1, 1)$: $(1 - 0) \times (1 - 0) = 1 \times 1 = \mathbf{+1}$

$$\text{Cov}(X, Y) = \frac{1 + 0 + 1}{3} = \frac{2}{3} \approx \mathbf{+0.67}$$

* **Result:** Covariance **Positive ($>0$)** aaya.
* **Meaning:** $X$ aur $Y$ ek hi direction me move kar rahe hain ($X$ badhega toh $Y$ bhi badhega).

---

**Example 2: Negative Relationship (Figure 3 Wala Data)**

Data points: $\{(-1, 1), (0, 0), (1, -1)\}$

* Mean of $X$ ($\bar{x}$) = $0$
* Mean of $Y$ ($\bar{y}$) = $0$

Har point ka product calculate karte hain:

* Point $(-1, 1)$: $(-1 - 0) \times (1 - 0) = (-1) \times (1) = \mathbf{-1}$
* Point $(0, 0)$: $(0 - 0) \times (0 - 0) = 0 \times 0 = \mathbf{0}$
* Point $(1, -1)$: $(1 - 0) \times (-1 - 0) = (1) \times (-1) = \mathbf{-1}$

$$\text{Cov}(X, Y) = \frac{-1 + 0 + (-1)}{3} = \frac{-2}{3} \approx \mathbf{-0.67}$$

* **Result:** Covariance **Negative ($<0$)** aaya.
* **Meaning:** $X$ aur $Y$ opposite directions me move kar rahe hain ($X$ badhega toh $Y$ ghatega).

---

**Quick Verdict**

* **$\text{Cov}(X, Y) > 0$:** Direct/Positive relationship (dono sath me badhte/ghat-te hain).
* **$\text{Cov}(X, Y) < 0$:** Inverse/Negative relationship (ek badhega toh dusra ghatega).
* **$\text{Cov}(X, Y) \approx 0$:** No linear relationship (dono ek dusre se independent hain).

Jaha variance dono examples me same ($0.67$) tha aur relation nahi pakad pa raha tha, wahi covariance ke sign ($+0.67$ vs $-0.67$) ne turant clear kar diya ki dono datasets ka behavior bilkul alag hai.

### *****Covariance Matrix (Concept & Structure)*****

Agar aapke paas **2-Dimensional data** hai jisme do columns hain: $x_1$ aur $x_2$:

* **Matrix Dimensions:**
  * Original features = $2$
  * Covariance Matrix ka size hamesha **$2 \times 2$** hoga:
    $$\text{Dimension} = n \times n \quad (n = \text{number of features})$$

---

### 🧱 *******Covariance Matrix Form*******

Initial formula structure:
$$\Sigma = \begin{bmatrix} \text{cov}(x_1, x_1) & \text{cov}(x_2, x_1) \\ \text{cov}(x_1, x_2) & \text{cov}(x_2, x_2) \end{bmatrix}$$

Is matrix me do mathematical properties lagti hain:

* **1. Khud ka khud ke saath Covariance = Variance:**
  * $\text{cov}(x_1, x_1) = \text{var}(x_1)$
  * $\text{cov}(x_2, x_2) = \text{var}(x_2)$
  * *(Diagonal elements hamesha individual features ke variance hote hain).*

* **2. Symmetry Property:**
  * $\text{cov}(a, b) = \text{cov}(b, a)$
  * $x_1$ aur $x_2$ ka relation wahi hai jo $x_2$ aur $x_1$ ka hai. Isliye:
    $$\text{cov}(x_1, x_2) = \text{cov}(x_2, x_1)$$

Final standard structure:
$$\Sigma = \begin{bmatrix} \mathbf{\text{var}(x_1)} & \text{cov}(x_1, x_2) \\ \text{cov}(x_1, x_2) & \mathbf{\text{var}(x_2)} \end{bmatrix}$$

---

### 🔍 *****Covariance Matrix Ki 2 Sabse Badi Khasiyat*****

* **1. Square Matrix:** Rows aur Columns hamesha barabar hote hain ($n \times n$).
* **2. Symmetric Matrix:** Diagonal ke across mirror image hoti hai ($A = A^T$), kyunki off-diagonal elements $\text{cov}(x_1, x_2)$ aur $\text{cov}(x_2, x_1)$ identical hote hain.

### ***3×3 Covariance Matrix (Detailed Breakdown)***

Agar aapke paas **3-Dimensional data** hai jisme 3 features hain: $x, y, z$:

$$\text{Matrix Size} = 3 \times 3 \quad (n \times n \text{ where } n = 3)$$

$$\Sigma = \begin{bmatrix} 
\mathbf{V_x} & C_{x,y} & C_{x,z} \\ 
C_{y,x} & \mathbf{V_y} & C_{y,z} \\ 
C_{z,x} & C_{z,y} & \mathbf{V_z} 
\end{bmatrix}$$

---

### 🔍 ****Matrix Components****

* **1. Principal Diagonal (Red Circles — $V_x, V_y, V_z$):**
  * Diagonal elements hamesha individual features ke **Variances** hote hain.
  * $V_x = \text{cov}(x, x) = \text{var}(x)$
  * $V_y = \text{cov}(y, y) = \text{var}(y)$
  * $V_z = \text{cov}(z, z) = \text{var}(z)$
  * Yeh batate hain ki har individual feature apne mean ke along kitna spread out hai.

* **2. Off-Diagonal Elements (Red Groups — Covariances):**
  * Diagonal ke alawa baaki saare elements do alag variables ke beech ka **Covariance** show karte hain.
  * Upper Triangle: $\{C_{x,y}, C_{x,z}, C_{y,z}\}$
  * Lower Triangle: $\{C_{y,x}, C_{z,x}, C_{z,y}\}$

* **3. Symmetric Nature ($\text{cov}(a, b) = \text{cov}(b, a)$):**
  * $C_{x,y} = C_{y,x}$
  * $C_{x,z} = C_{z,x}$
  * $C_{y,z} = C_{z,y}$
  * Matrix ka transpose lene par original matrix hi banta hai ($\Sigma = \Sigma^T$). Isliye 3D data ke liye hume sirf 3 variances aur 3 unique covariances calculate karne padte hain.

### **Kaise Ghuma Raha Hai Yeh Matrix? (Geometric Intuition)**

* **1. Basis Vectors ($\hat{i}$ aur $\hat{j}$) Ghumte Hain:**
* Har 2D point do basic vectors se bana hota hai:
* X-axis ka direction: $\hat{i} = \begin{bmatrix} 1 \\ 0 \end{bmatrix}$
* Y-axis ka direction: $\hat{j} = \begin{bmatrix} 0 \\ 1 \end{bmatrix}$


* Jab hum kisi matrix $A = \begin{bmatrix} a & b \\ c & d \end{bmatrix}$ se multiply karte hain:

$$A \hat{i} = \begin{bmatrix} a & b \\ c & d \end{bmatrix} \begin{bmatrix} 1 \\ 0 \end{bmatrix} = \begin{bmatrix} a \\ c \end{bmatrix}$$


$$A \hat{j} = \begin{bmatrix} a & b \\ c & d \end{bmatrix} \begin{bmatrix} 0 \\ 1 \end{bmatrix} = \begin{bmatrix} b \\ d \end{bmatrix}$$


* Matrix ke pehle column $\begin{bmatrix} a \\ c \end{bmatrix}$ ne $\hat{i}$ ko ghumakar nayi jagah bhej diya, aur dusre column $\begin{bmatrix} b \\ d \end{bmatrix}$ ne $\hat{j}$ ko ghumakar nayi jagah bhej diya.


* **2. Saara Data Sath Me Kyu Ghoom Gaya?**
* Kyunki grid ka har point $\vec{v} = x\hat{i} + y\hat{j}$ hota hai.
* Jab $\hat{i}$ aur $\hat{j}$ rotate aur stretch hote hain, toh unke upar tika hua pura space, saare data points aur grid lines ek saath tilt aur stretch ho jaate hain ($A\vec{v} = x(A\hat{i}) + y(A\hat{j})$).


* **3. Next Step (Eigenvectors Ki Entry):**
* Is ghoomte hue space me lagbhag saare vectors apni original direction se hat (rotate ho) jaate hain.
* Lekin **kuch special vectors aise hote hain jo space ke ghoomne ke baad bhi apni original line/direction par hi tike rehte hain**, sirf stretch ya shrink hote hain—unhi special vectors ko hum **Eigenvectors** kehte hain.

## ****Eigen Vector****

## ******🔷 Eigenvectors: Core Concept******

> **Definition:** Eigenvectors are special vectors jinke upar linear transformation lagane par unka **direction change nahi hota**, sirf magnitude (scale) badalta hai.

---

### *****Key Points:*****
* **Special Vectors:** Eigenvectors aise vectors hote hain jo transformation ke baad bhi apni original line (span) par hi rehte hain.
* **Direction vs. Magnitude:** Inka direction bilkul same rehta hai (ya exactly opposite flip ho sakta hai), bas scale badal jata hai—yaani vector lamba (stretch) ya chhota (shrink) ho sakta hai.
* **2-D Transformations:** Normal cases me, jab hum 2D space me koi linear transformation apply karte hain, toh hume standard roop se **do eigenvectors** milte hain.
* **Core Formula:** 
  $$A v = \lambda v$$
  *(Jaha $A$ transformation matrix hai, $v$ apka eigenvector hai, aur $\lambda$ wo factor hai jisse vector scale hota hai).*

## ******🔶 Eigenvalues: Core Concept & Intuition******

> **Definition:** Jab linear transformation lagta hai, toh eigenvector kitna **stretch** (lamba) ho raha hai ya kitna **shrink** (chhota) ho raha hai, us scaling factor ko **Eigenvalue ($\lambda$)** kehte hain.

---

### ***Key Intuition:***
* **Coordinate Transformation:** Jab hum matrix ko as a linear transformation apply karte hain, toh pura coordinate system transform ho jata hai aur mostly saare vectors rotate ya shift ho jate hain.
* **Special Invariant Vectors:** Lekin kuch special vectors apni original line (direction) nahi chhodte:
  * **2D Space:** Maximum **2** eigenvectors
  * **3D Space:** Maximum **3** eigenvectors
  * **n-D Space:** Maximum **$n$** eigenvectors
* **Eigenvector vs Eigenvalue:**
  * Jo vectors direction change nahi karte = **Eigenvectors**
  * Vector jis factor se scale (expand/shrink/flip) hota hai = **Eigenvalues ($\lambda$)**

---

### ***💡 Simple Example:***
Maan lo hamare paas ek 2D transformation matrix $A$ aur ek vector $v$ hai:

$$A = \begin{bmatrix} 3 & 0 \\ 0 & 2 \end{bmatrix}, \quad v = \begin{bmatrix} 1 \\ 0 \end{bmatrix}$$

Ab transformation apply karte hain ($A \cdot v$):

$$A v = \begin{bmatrix} 3 & 0 \\ 0 & 2 \end{bmatrix} \begin{bmatrix} 1 \\ 0 \end{bmatrix} = \begin{bmatrix} 3 \\ 0 \end{bmatrix} = 3 \cdot \begin{bmatrix} 1 \\ 0 \end{bmatrix}$$

* **Direction:** Vector $\begin{bmatrix} 1 \\ 0 \end{bmatrix}$ pehle bhi x-axis par tha, abhi bhi x-axis par hi hai (Direction same raha $\rightarrow$ **Eigenvector**).
* **Magnitude:** Vector pehle se **3 guna** lamba ho gaya, toh scaling factor **$\lambda = 3$** hua ($\rightarrow$ **Eigenvalue**).

$$\mathbf{A v = \lambda v}$$

## ****🔷 PCA (Principal Component Analysis) Me Eigenvalues & Eigenvectors Ka Role****

> **Core Idea:** PCA ka main purpose data ki dimensionality reduce karna hota hai bina important information (variance) ko khoye. Yaha **Covariance Matrix** ke eigenvalues aur eigenvectors hamare guide bante hain.

---

### ****1. Direction & Spread (Variance):****
* **Largest Eigenvector ($PC_1$):** Covariance matrix ka sabse bada eigenvector us direction me point karta hai jaha data ka **sabse jyada variance (spread)** hota hai. Isko hum **First Principal Component** bolte hain.
* **Largest Eigenvalue ($\lambda_1$):** Is vector ke corresponding jo eigenvalue hoti hai, wo batati hai ki us direction me kitna spread/variance hai—jitni badi value, utna jyada information.
* **Second Largest Eigenvector ($PC_2$):** Ye hamesha first eigenvector ke **orthogonal (90° perpendicular)** hota hai aur data ke second highest variance ki direction ko capture karta hai.

---

### ****2. Ye PCA Me Kaam Kaise Aate Hain?****
* **Information Retention:** Badi eigenvalues batati hain ki kaunse components me maximum data spread (signal) hai aur chhoti values noise represent karti hain.
* **Dimensionality Reduction:** Agar aapke paas 100 features hain, toh 100 eigenvectors milenge. Aap unko unke eigenvalues ke hisaab se sort karke top 2 ya 3 select kar sakte ho jisme 90-95% variance cover ho jaye.
* **No Redundancy (Uncorrelated):** Saare eigenvectors aapas me orthogonal hote hain, jiska matlab naye features ek dusre se bilkul independent hote hain aur multicollinearity khatam ho jati hai.

---

### ****📌 Summary Formula:****
$$\text{Covariance Matrix } (\Sigma) \xrightarrow{\text{Eigen-decomposition}} v \text{ (Directions / Components)}, \; \lambda \text{ (Variance Explained)}$$

___

> ### ****📌 Crucial Takeaway: Covariance Matrix Ka Eigenvector Kyu Nikalte Hain?****
> 
> Jab hum **Covariance Matrix** ka Eigen-decomposition karte hain:
> * **Highest Eigenvalue:** Sabse badi eigenvalue ke corresponding jo eigenvector hota hai, wahi actually data ke **Maximum Variance (largest spread)** ki direction me point karta hai.
> * **Primary Goal:** Hum covariance matrix ka eigenvector isliye nikaalte hain taaki hume wo exact direction mil sake jisme data ki **maximum information** store hai—aur wahi hamara **First Principal Component ($PC_1$)** banta hai!

# 🔷 Step-by-Step PCA (Principal Component Analysis)

> **Problem Statement:** Hamare paas ek **3-D dataset** hai jisme 3 features hain: **$f_1, f_2, f_3$**. PCA laga kar hum is data ki dimensionality reduce karenge (e.g., 3-D se 1-D ya 2-D me).

---

### Step 1: Mean Centering (Data ko Center me Lana)
* Apka jitna bhi data hai ap Use sabse pahele Centre me Leke ate ho ( Its not Compulsury) but karo Aur likho esse ( Mean in Evey Direction) esse hi mean centric bola jata hai 

* Sabse pehla step hota hai apne pure data ko origin $(0, 0, 0)$ par shift karna.
* **Is it compulsory?** Math point of view se zaroori hai taaki covariance simple matrix multiplication se nikal sake aur origin se spread measure ho sake.
* Har feature ka individual mean $(\bar{X})$ calculate karke data points se subtract karte hain:
  $$f_1' = f_1 - \bar{f_1}, \quad f_2' = f_2 - \bar{f_2}, \quad f_3' = f_3 - \bar{f_3}$$

```text
       [ Original Data ]                         [ Mean Centered Data ]
           y                                           y
           |    * *                                    |    *
           |   * x *   <-- Mean (x̄)                    |  * x *   <-- Origin (0,0)
           |    * *                               -----+----- x
      -----+--------- x                                |    *
           |                                           |

```



---

### Step 2: Covariance Matrix Calculate Karna

* Mean centering ke baad hum pata karte hain ki features aapas me kaise vary kar rahe hain.
* 3 features ($f_1, f_2, f_3$) ke liye hamare paas ek **$3 \times 3$ Covariance Matrix** banti hai:

$$\text{Cov Matrix} = \begin{bmatrix} 
\text{V}(f_1) & \text{C}(f_1, f_2) & \text{C}(f_1, f_3) \\ 
\text{C}(f_2, f_1) & \text{V}(f_2) & \text{C}(f_2, f_3) \\ 
\text{C}(f_3, f_1) & \text{C}(f_3, f_2) & \text{V}(f_3) 
\end{bmatrix}$$

* **Diagonal Elements:** $\text{V}(f_i)$ = Individual feature ka **Variance** (spread).
* **Off-diagonal Elements:** $\text{C}(f_i, f_j)$ = Do features ke beech ka **Covariance** (relation).

---

### Step 3: Eigenvalues & Eigenvectors Calculate Karna

* Ab hum is $3 \times 3$ Covariance Matrix ka **Eigen-decomposition** nikalte hain.
* 3-D space me hume:
* **3 Eigenvalues:** $\lambda_1, \lambda_2, \lambda_3$ (Maan lo $\lambda_1 > \lambda_2 > \lambda_3$)
* **3 Eigenvectors:** $v_1, v_2, v_3$


* **Principal Components (Directions):**
* $v_1$ (highest $\lambda_1$) $\rightarrow$ **$PC_1$** (Maximum variance wali direction)
* $v_2$ (second highest $\lambda_2$) $\rightarrow$ **$PC_2$** (Orthogonal to $PC_1$, second highest variance)
* $v_3$ (lowest $\lambda_3$) $\rightarrow$ **$PC_3$** (Third direction)



---

### Final Decision: Dimensionality Reduction

Ab yeh hamare upar depend karta hai ki kitni dimensions chahiye:

* **3-D $\rightarrow$ 1-D:** Agar sirf 1 dimension me transform karna hai, toh hum sirf **$PC_1$ ($v_1$)** select karenge aur data ko us par project kar denge.
* **3-D $\rightarrow$ 2-D:** Agar 2D projection chahiye, toh **$PC_1$ aur $PC_2$ ($v_1, v_2$)** select karenge.
