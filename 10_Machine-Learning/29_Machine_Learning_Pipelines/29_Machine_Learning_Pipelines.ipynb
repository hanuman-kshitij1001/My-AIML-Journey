# *****Scikit-Learn Pipelines*****

> **Core Definition:**  
> **Pipelines chains together multiple steps so that the output of each step is used as input to the next step.**  
> **Pipelines makes it easy to apply the same preprocessing to train and test!**

---

### Key Advantages of Using Pipelines

* **End-to-End Workflow:** Data Imputation $\rightarrow$ Encoding $\rightarrow$ Scaling $\rightarrow$ Model Training sab ek single object ke andar chain ho jate hain.
* **Prevents Data Leakage:** Training data par fit hone wale exact parameters automatically test data par apply hote hain bina kisi code leakage ke.
* **Clean & Modular Code:** Multiple lines ke manual `fit_transform` aur `transform` calls likhne ki zaroorat nahi padti.
* **Easy Hyperparameter Tuning:** Poore pipeline (preprocessing + model) ko direct `GridSearchCV` ya `RandomizedSearchCV` me pass kar sakte hain.

---

### Pipeline Architecture Simulation

$$\text{Raw Data} \xrightarrow{\text{Imputation}} \text{Cleaned Data} \xrightarrow{\text{Encoding/Scaling}} \text{Preprocessed Data} \xrightarrow{\text{Model Fit/Predict}} \text{Predictions}$$

---

### Python Code Implementation

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression

# Pipeline step-by-step assemble karna
pipe = Pipeline([
    ('imputer', SimpleImputer(strategy='mean')),
    ('scaler', StandardScaler()),
    ('classifier', LogisticRegression())
])

# Training step: Ek hi call me fit ho jata hai
pipe.fit(X_train, y_train)

# Inference/Test step: Same preprocessing automatically apply hoti hai
y_pred = pipe.predict(X_test)
