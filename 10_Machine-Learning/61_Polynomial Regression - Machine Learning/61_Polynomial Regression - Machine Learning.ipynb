### ****Polynomial — Definition****

* ****Polynomial ek mathematical expression hota hai jisme variables ki powers non-negative integers (0, 1, 2, 3, …) hoti hain, aur variables ke terms ko + ya − se combine kiya jata hai.****

* ****Example:****



$$
P(x) = 3x^3 + 2x^2 - 5x + 7
$$

****Yahan:****

* ****\(x\) → variable****
* ****\(3, 2, -5, 7\) → coefficients/constants****
* ****\(3,2,1,0\) → powers****

👉 **Simple words:**
**Polynomial = Variables + Non-negative integer powers + Coefficients**

- ❌ $x^{-2}+3$ → Polynomial nahi hai
- ❌ $\sqrt{x}+2$ → Polynomial nahi hai
- ✅ $x^2+3x+5$ → Polynomial hai
