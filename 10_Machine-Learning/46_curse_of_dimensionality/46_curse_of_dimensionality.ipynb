## ******What is Curse of Dimensionality (CoD)?******

**Machine Learning me hamare paas columns hote hain jinko hum **features** bolte hain, aur har dataset me multiple columns hote hain**.

* **Features = Dimensions:** CoD me sabse pehle toh features ko hi **dimensions** bola jata hai. Isliye **Curse of Dimensionality is basically "Curse of Features"**.
* **Core Problem:** Jaise-jaise number of features badhne lagte hain, ek limit tak toh theek rehta hai, lekin har dataset me sirf ek **Optimal Number of Features** hote hain.
* **Over-addition of Features:** Agar us limit ke baad bhi aap model me aur features zabardasti add karte jaoge, toh koi fayda nahi hone wala:
  * Model ki accuracy nahi badhegi.
  * Model ki performance nahi badhegi, ulta **performance drop (niche)** ho sakti hai (Overfitting hone lagti hai).
  * Computationally model bohot zyada heavy aur complex ho jata hai (training time explode ho jata hai).

> **Conclusion:** There is always an optimal number of features. Usse zyada badhane par accuracy badhne ki jagah ulta nuksan hota hai.
