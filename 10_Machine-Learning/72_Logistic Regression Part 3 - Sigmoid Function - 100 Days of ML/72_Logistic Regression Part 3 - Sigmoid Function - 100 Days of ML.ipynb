### **Consider Four Case**

* **1- line se bahut dur hai toh jada manitude se line ko apn etraf khecega for uncorect point**

* **2- correctly points ke liye jab bahut dur honga to ye slowly push karega aur pass honga to speed se push karega postive point ke liye ye hai**

##### **Es points ko maine gemini se video se nikala hai**
* 1- Mis-classified point (line ke paas): Agar point mis-classified hai aur line ke bahut kareeb hai, toh ye line ko apni taraf bahut kam magnitude se pull karega.

* 2- Mis-classified point (line se door): Agar point mis-classified hai aur line se kaafi door hai, toh ye line ko apni taraf zyada magnitude se zor se pull karega.

* 3- Correctly classified point (line ke paas): Agar point correctly classified hai aur line ke bahut paas hai, toh ye line ko apne se door bahut tezi se push karega.

* 4- Correctly classified point (line se door): Agar point correctly classified hai lekin line se kaafi door hai, toh ye line ko apne se door bahut dheemi gati se (weak push) dhakka dega.

### **Sigmoid function ke aane ke fayde:**

* **Probabilistic Output: Sigmoid function humein 0 aur 1 ke beech ki continuous value deta hai, na ki sirf discrete 0 ya 1**

* **Continuous Updates: Kyunki sigmoid ka output continuous hota hai, isliye (y - y_hat) kabhi bhi zero nahi hota. Iska matlab hai ki har point, chahe wo correctly classified ho ya nahi, model ke coefficients ko update karne mein madad karta hai**

* **Distance ka asar: Sigmoid use karne se hum ye bhi track kar paate hain ki point line se kitni door hai, jisse model aur zyada behtar tarike se train ho pata hai**
