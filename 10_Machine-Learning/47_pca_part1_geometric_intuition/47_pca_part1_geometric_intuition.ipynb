# ***PCA Part 1 - Geometric Intuition***


### **PCA (Principal Component Analysis)**

* **1.** PCA ek **Unsupervised learning** problem hai.
* **2.** Yaha pe aapke paas me **sirf input features** hote hain.
* **3.** It is a **complex technique**.
* **4.** Bahut **useful and old technique** hai.
* **5.** Iska maths bahut **common** hai.
* **6.** Iska main kaam ye hota hai tab jab aapke data me bahut saare columns hain, toh ye method ke wajah se aapke **columns reduce ho jaate hain**.

---

### 📸 ***PCA Kaam Kaise Karta Hai Behind the Scene***

* **Real-World Example:** Ek photographer ek photo le raha hai. Wo ek soccer match ki photo le raha hai aur ye match **3D environment** me khela ja raha hai. Is photographer ko photo isliye lena hai kyunki wo use newspaper me publish karega. Hum sab jaante hain ki photos **2D** hote hain. Ab ye match jo ho raha hai, wo use apne news me dikhana chahta hai — matlab **3D movement ko 2D me convert karna chahta hai**. 
* **The Connection:** PCA exactly yahi karta hai!

---

### 🎯 ***Core Goal of PCA***

* PCA hamesha ek **Higher Dimension dataset ko Lower Dimension** me laane ki koshish karta hai.
* Isme khaas baat yeh hai ki PCA **best possible lower dimension** leke chalta hai, jisse aap us data ka jo **essence** hai higher dimension me, wo maximum capture kar pao.
