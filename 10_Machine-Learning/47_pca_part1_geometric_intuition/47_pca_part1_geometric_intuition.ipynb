# ***PCA Part 1 - Geometric Intuition***


### **PCA (Principal Component Analysis)**

* **1.** PCA ek **Unsupervised learning** problem hai.
* **2.** Yaha pe aapke paas me **sirf input features** hote hain.
* **3.** It is a **complex technique**.
* **4.** Bahut **useful and old technique** hai.
* **5.** Iska maths bahut **common** hai.
* **6.** Iska main kaam ye hota hai tab jab aapke data me bahut saare columns hain, toh ye method ke wajah se aapke **columns reduce ho jaate hain**.

---

### 📸 ***PCA Kaam Kaise Karta Hai Behind the Scene***

* **Real-World Example:** Ek photographer ek photo le raha hai. Wo ek soccer match ki photo le raha hai aur ye match **3D environment** me khela ja raha hai. Is photographer ko photo isliye lena hai kyunki wo use newspaper me publish karega. Hum sab jaante hain ki photos **2D** hote hain. Ab ye match jo ho raha hai, wo use apne news me dikhana chahta hai — matlab **3D movement ko 2D me convert karna chahta hai**. 
* **The Connection:** PCA exactly yahi karta hai!

---

### 🎯 ***Core Goal of PCA***

* PCA hamesha ek **Higher Dimension dataset ko Lower Dimension** me laane ki koshish karta hai.
* Isme khaas baat yeh hai ki PCA **best possible lower dimension** leke chalta hai, jisse aap us data ka jo **essence** hai higher dimension me, wo maximum capture kar pao.

### 🔍 Is Diagram Me Actually Ho Kya Raha Hai?

Humare paas 2 features ka scatter plot hai:

* **X-axis:** `No. of rooms`
* **Y-axis:** `No. of grocery shops`

Ab agar humein is 2D data ko **1D me convert (reduce)** karna hai:

* **1. Room Axis Ka Spread ($d$):**
* Data points ko agar X-axis (`room`) par project karein, toh unka spread (range) bohot lamba hai, jise sir ne **$d$** likha hai.
* Lamba spread hone ka matlab hai **High Variance** (yaani information bohot zyada hai).


* **2. Grocery Axis Ka Spread ($d'$):**
* Agar data points ko Y-axis (`grocery`) par project karein, toh unka spread bohot chhota hai, jise sir ne **$d'$** likha hai.
* Chhota spread hone ka matlab hai **Low Variance** (yaani sabhi gharo ke aas-pass grocery shops lagbhag similar hain, zyada farq nahi hai).


* **3. Mathematical Condition ($d > d'$):**
* Since **$d > d'$**, `room` axis ka variance `grocery` se bohot zyada hai.
* Agar hume dono me se kisi ek ko chunna ho (Feature Selection), toh hum `grocery` ko drop karenge aur `room` ko retain karenge kyunki `room` me data ka असली signal (variance) chhipa hai.



---

### 💡 PCA Me Yeh Step Kyu Padhaya Jata Hai?

PCA ka foundation samajhne ke liye yeh **3 logical steps** ka pehla step hai:

* **Core Rule of PCA:** Machine learning aur statistics me **$\text{Spread (Spread of Data)} = \text{Variance} = \text{Information}$**. Jis direction me data jitna faila hua hai, waha utni hi unique information hai.
* **Feature Selection Ki Limitation:** Agar hum sirf X ya Y axis chun rahe hain, toh yeh **Feature Selection** hai. Isme hum `grocery` ka jo thoda bohot spread ($d'$) tha, use 100% loss kar dete hain.
* **PCA Ka Asli Jump:**
* Sir yaha background set kar rahe hain ki *“Abhi toh humne seedhe X aur Y axis dekhe. Lekin kya ho agar data ka maximum spread kisi diagonal (tedhe) angle par ho?”*
* Yahi se **PCA (Feature Extraction)** ka janm hota hai — PCA X ya Y axis chunne ke bajaye ek naya axis draw karta hai jo maximum variance ko capture kare bina dusre feature ko poora waste kiye.

###  ***Case 2: Jab Dono Features Ka Variance Barabar Ho ($d \simeq d'$)***

* **Dilemma (Dharamsankat):**
  * Yaha do features hain: `No. of rooms` ($X$-axis) aur `No. of washrooms` ($Y$-axis).
  * Agar dono axis par data points ka spread dekhein, toh $X$-axis ka spread ($d$) aur $Y$-axis ka spread ($d'$) lagbhag **barabar ($d \simeq d'$)** hai.
  * Matlab dono features me lagbhag **equal amount of variance (information)** hai.

* **Feature Selection Ka Failure:**
  * Agar hum yaha Feature Selection lagane jayenge, toh hum fas jayenge.
  * Agar `washrooms` ko drop karoge, toh 50% valuable information khatam ho jayegi.
  * Agar `rooms` ko drop karoge, tab bhi 50% information chali jayegi.
  * Aap decide hi nahi kar paoge ki kisse drop karein aur kise retain karein!

* **PCA (Feature Extraction) Ki Zaroorat:**
  * Is graph me data points diagonal (tedhe) spread hain.
  * PCA yahi par kaam aata hai: wo $X$ ya $Y$ chunne ke bajaye **data ke beech se ek naya diagonal axis create karta hai**, jisse dono columns ka essence ek single new feature me combine hokar preserve ho jata hai bina information drop kiye.

### ***📐 Mathematical Formula of Variance***

$$\sigma^2 = \frac{\sum_{i=1}^{n} (x_i - \bar{x})^2}{n}$$

* $x_i$ = Har individual data point
* $\bar{x}$ = Data ka Mean (Average)
* $(x_i - \bar{x})$ = Har point ka mean se distance
* Square karne ka reason: Negative aur positive distances aapas me cancel na ho jayein, aur bade differences ko zyada weight mile.
* $n$ = Total number of data points

___

Spread aur variance dono technically alag concepts hain, lekin dono ek dusre ke directly proportional hote hain.

* **Spread (Phailav):** Yeh data ka visual ya physical spread hai jo batata hai ki points kitni door tak bikhre hue hain.
* **Variance (Mathematical Measurement):** Yeh ek formula-based calculation hai jo us spread ko numbers me measure karta hai (squared distances ka average).
* **The Golden Rule:** Agar data ka spread zyada hai, toh uska variance bhi zyada hoga. Par spread khud variance nahi hai—spread ek physical ya general observation hai, aur variance us spread ko exactly quantify karne ka mathematical tareeqa hai!

###  ***Is Figure Me Kya Ho Raha Hai?***

* **Projection on X-Axis:** 
  * Graph me 2D space ke data points ($X, Y$) hain. Jab hume dimensionality reduce karke 1D par laana hota hai, toh hum har point se X-axis par ek seedhi line drop karte hain (ise mathematically **Orthogonal Projection** bolte hain).
  * Projection ke baad saare 2D points X-axis par ek single line ke upar dots ban jaate hain.

* **Spread Ka Calculation:**
  * Ab yeh saare dots X-axis par kitni door tak phailte hain, wahi unka physical **Spread** hai.
  * Aur is physical spread ko mathematically napne (measure karne) ke liye hum X-axis par mapped points ka **Variance** calculate karte hain.

---

### ❓ ***Variance Kya Hai Aur Spread Ke Liye Kyu Important Hai?***

* **Spread is Physical, Variance is Mathematical:**
  * **Spread** sirf aankhon se dekhne wala visual observation hai (ki points kitne bikhre hue hain).
  * **Variance** us bikhrav ki precise numerical value hai:
    $$\text{Variance } (\sigma^2) = \frac{\sum (x_i - \bar{x})^2}{n}$$
* **Spread Hamesha Variance ke Proportional Hota Hai:**
  $$\text{Spread} \propto \text{Variance}$$
  * Spread zyada hoga $\longrightarrow$ Points mean se door honge $\longrightarrow$ Variance bada aayega.
  * Spread kam hoga $\longrightarrow$ Points aapas me chipke honge $\longrightarrow$ Variance chhota aayega.

* **PCA Me Yeh Kyu Zaroori Hai:**
  * PCA ka main objective hai: **"Data ko kisi aise axis par project karo jaha points ka spread (variance) sabse maximum ho."**
  * Agar hum kisi aise axis par project kar dein jaha spread lagbhag 0 ho, toh saare points ek dusre ke upar overlap ho jayenge aur unke beech ka differentiation (information) destroy ho jayega!

### ***Variance Kyu Important Hai PCA Ke Liye? (Detailed Notes)***

* **1. Setup:** Diagram me $X-Y$ 2D space me data points spread out hain.
* **2. The Goal:** Machine Learning model ka main task hai in **Green** aur **Red** classes ke beech differentiate karna.
* **3. The 1D Reduction Dilemma:** Agar hum is 2D space ko reduce karke **1-Dimension** me project karein, toh hume sahi axis chunna padega:
  * **Agar X-Axis par project karein:** Red aur Green dono classes ka spread (variance) wide rehta hai. Actual 2D distance (blue line) ke comparison me X-axis par bhi dono ke beech ek **clearly comparable large gap** bachta hai.
  * **Agar Y-Axis par project karein:** Y-axis par dono classes aapas me bohot paas aa jaati hain aur unke beech ka distance lagbhag kho jata hai.
* **4. Impact on Algorithms (e.g., KNN):** Agar Y-axis par projection le liya, toh KNN jaisa distance-based algorithm fail ho jayega kyunki use lagega dono classes aapas me chipki hui hain. Lekin X-axis par variance zyada hone ki wajah se clear separation maintain rehta hai.
* **5. Photographer Analogy Connection:**
  * Agar photographer sahi angle se click kare, toh ground par door khade do players ke beech ka gap 2D photo me clearly dikhta hai.
  * Agar wo galat side se angle le, toh ek player dusre ke peeche chhip jata hai aur gap khatam ho jata hai.
* **6. The Core Objective of PCA:**
  $$\text{PCA End Goal} = \mathbf{\text{Maximize the Variance}}$$
  PCA ka single objective yahi hota hai: ek aisa naya axis dhoondhna jaha data points ka spread (variance) sabse maximum ho, taaki class separation aur actual information 100% preserve rahe!
