### *Power Transformation Kya Hai?*

***Power Transformation ek aisi technique hai jisme data ki non-linear values ko mathematical power ($\lambda$ parameter) ke through transform kiya jata hai, taaki skewed data ek **Normal / Gaussian Distribution** me convert ho sake aur variance stabilize ho jaye.***

*Scikit-Learn ke `sklearn.preprocessing.PowerTransformer` ke andar do main power transforms aate hain:*

---

### **1. Box-Cox Transform**
*  **Working:** Ye internal statistical optimization use karke best $\lambda$ (lambda) value dhoondhta hai jo data ko best normal curve deta hai.
* **Strict Condition:** Ye **strictly positive values ($x > 0$)** par hi kaam karta hai (zeros ya negative numbers par fail ho jata hai).

<div style="background-color: #1a1a1a; padding: 25px; border-radius: 12px; font-family: -apple-system, BlinkMacSystemFont, 'Segoe UI', Roboto, sans-serif; max-width: 750px; margin: 20px auto; color: #ffffff;">



<div style="background-color: #ffffff; color: #000000; border-radius: 6px; padding: 25px 20px; text-align: center; margin-bottom: 25px;">

$$x_i^{(\lambda)} = \begin{cases} \dfrac{x_i^\lambda - 1}{\lambda} & \text{if } \lambda \neq 0, \\[2ex] \ln(x_i) & \text{if } \lambda = 0 \end{cases}$$

</div>

<div style="font-size: 15.5px; line-height: 1.6; color: #cbd5e1; text-align: left;">
The exponent here is a variable called lambda ($\lambda$) that varies over the range of -5 to 5, and in the process of searching, we examine all values of $\lambda$. Finally, we choose the optimal value (resulting in the best approximation to a normal distribution) for your variable.
</div>

</div>

### **2. Yeo-Johnson Transform**
* **Working:** Ye Box-Cox ka advanced aur modern variation hai.
* **Advantage:** Ye **positive, zero aur negative ($x \le 0$)** har type ke numbers par bina kisi error ke successfully kaam karta hai.

<div style="background-color: #1a1a1a; padding: 25px; border-radius: 12px; font-family: -apple-system, BlinkMacSystemFont, 'Segoe UI', Roboto, sans-serif; max-width: 780px; margin: 20px auto; color: #ffffff;">

<div style="background-color: #ffffff; color: #000000; border-radius: 6px; padding: 25px 20px; text-align: center; margin-bottom: 25px;">

$$x_i^{(\lambda)} = \begin{cases} 
{[(x_i + 1)^\lambda - 1]} / {\lambda} & \text{if } \lambda \neq 0, x_i \ge 0, \\[2ex]
\ln(x_i + 1) & \text{if } \lambda = 0, x_i \ge 0, \\[2ex]
-{[( -x_i + 1)^{2 - \lambda} - 1]} / {(2 - \lambda)} & \text{if } \lambda \neq 2, x_i < 0, \\[2ex]
-\ln(-x_i + 1) & \text{if } \lambda = 2, x_i < 0
\end{cases}$$

</div>

<div style="font-size: 15.5px; line-height: 1.6; color: #cbd5e1; text-align: left;">
This transformation is somewhat of an adjustment to the Box-Cox transformation, by which we can apply it to negative numbers.
</div>

</div>
