### 📐 Distance Formula with Missing Values (Nan-Euclidean Distance)

$$\text{dist}(x, y) = \sqrt{\text{weight} \times \sum (\text{sq. distance from present coordinates})}$$

**Where:**

$$\text{weight} = \frac{\text{Total \# of coordinates}}{\text{\# of present coordinates}}$$

### 🔍 Dataset ke saath Step-by-Step Working:

Maan lo hume **Row 1** ka **Feature 2 (Missing Value)** fill karna hai aur hum **K = 2** le rahe hain.

---

#### 1. Nan-Euclidean Distance Formula:
$$\text{dist}(x, y) = \sqrt{\frac{\text{Total Features}}{\text{Common Features}} \times \sum (\text{diff})^2}$$

---

#### 2. Row 1 ka doosri Rows se Distance:

* **Row 1 vs Row 3:**
  * Common features jaha dono present hain: **Feature 1, 3, 4** (Total = 4, Common = 3)
  * Distance:
    $$\text{dist}(R_1, R_3) = \sqrt{\frac{4}{3} \times \left[(33 - 23)^2 + (67 - 71)^2 + (21 - 18)^2\right]}$$
    $$= \sqrt{\frac{4}{3} \times [100 + 16 + 9]} = \sqrt{\frac{4}{3} \times 125} \approx \mathbf{12.91}$$

* **Row 1 vs Row 5:**
  * Common features: **Feature 1, 3** (Total = 4, Common = 2)
  * Distance:
    $$\text{dist}(R_1, R_5) = \sqrt{\frac{4}{2} \times \left[(33 - 35)^2 + (67 - 79)^2\right]}$$
    $$= \sqrt{2 \times [4 + 144]} = \sqrt{296} \approx \mathbf{17.20}$$

---

#### 3. Top-K Nearest Neighbors Select Karna:
Sabhi rows ka distance nikalne ke baad, sabse kam distance wale top 2 neighbors chune jaayenge (maan lete hain **Row 3** aur **Row 5** sabse paas hain).

* Row 3 me Feature 2 ki value = **51**
* Row 5 me Feature 2 ki value = **60**

---

#### 4. Final Missing Value Impute Karna:
Donon values ka average (mean) nikal kar fill kar diya jata hai:

$$\text{Imputed Value} = \frac{51 + 60}{2} = \mathbf{55.5}$$

Is tareeqe se Row 1 ka missing Feature 2 **55.5** se replace ho jata hai.

### 📐 Nan-Euclidean Distance Kya Hai?

Normal **Euclidean Distance** do points ke beech tabhi calculate ho sakti hai jab dono ke sabhi coordinates (features) present hon. Agar kisi bhi row me ek bhi **NaN (Missing Value)** aa jaye, toh normal Euclidean formula fail ho jata hai. 

Is problem ko solve karne ke liye **Nan-Euclidean Distance** use hota hai.

---

### ⚙️ Ye Kaise Kaam Karta Hai?

1. **Ignore Missing:** Jin coordinates me koi ek ya dono points NaN hain, unhe calculation se temporary drop kar diya jata hai.
2. **Present Coordinates par Distance:** Sirf unhi features par square difference nikala jata hai jo **dono rows me present** hain.
3. **Weight (Scaling) Factor:** Kam features hone ki wajah se distance artificially chhota na ho jaye, isliye distance ko scale-up kiya jata hai ek weight factor se.

---

### 🧮 Formula:

$$\text{dist}(x, y) = \sqrt{\text{weight} \times \sum_{i \in \text{present}} (x_i - y_i)^2}$$

Jahan:

$$\text{weight} = \frac{\text{Total Coordinates (Features)}}{\text{Common Present Coordinates}}$$

---

### 💡 Simple Example:
Maan lo dataset me total **4 features** hain, par do rows ke beech sirf **2 features common** hain:
* Weight ho jayega: $4 / 2 = 2$
* Dono common features ke distance ko 2 se multiply kiya jayega taaki 4 features ke barabar ka fair estimate mil sake.

### ⚖️ Advantages & Disadvantages of KNN Imputer

---

#### 1. More Accurate ✅ *(Advantage)*
* Simple Mean / Median imputation ke mukable yeh zyada accurate results deta hai kyunki yeh similar data points (neighbors) ke pattern aur correlation ko follow karta hai.

---

#### 2. More Number of Calculations ❌ *(Disadvantage)*
* Har ek missing value ke liye pure dataset ke har point se distance calculate karna padta hai ($O(n^2)$ time complexity).
* Is wajah se bade datasets par execution time bohot zyada lagta hai (Computationally expensive).

---

#### 3. Production Deployment Challenge ⚠️ *(Disadvantage)*
* KNN ek **Lazy Learner** (instance-based learner) hai.
* Iska matlab production environment me naye/unseen test data par imputation perform karne ke liye bhi pure `X_train` dataset ko memory me deploy aur store karke rakhna padta hai.

### ⚖️ KNN Imputer me `weights` Parameter: `uniform` vs `distance`

`KNNImputer(weights='uniform')` ya `weights='distance'` ye decide karta hai ki chune hue **K-Nearest Neighbors** ki values ko combine (average) karte waqt unko kitna importance (vazan) dena hai.

---

#### 1. `weights='uniform'` (Default)
* **Concept:** Sabhi neighbors ko **barabar importance (equal weight)** milti hai, chahe koi neighbor kitna bhi paas ho ya thoda door ho.
* **Kaise Kaam Karta Hai:** Yeh seedha chune hue $K$ neighbors ka **Simple Average (Mean)** nikal leta hai.
* **Kab Use Karein:** Jab aapka data kafi dense aur evenly distributed ho, yaani paas aur thode door wale neighbors ke beech ka farq bohot zyada matter na karta ho.

$$\text{Value} = \frac{v_1 + v_2 + \dots + v_k}{K}$$

---

#### 2. `weights='distance'`
* **Concept:** Jo neighbor **jitna paas** hoga, uska **weight utna hi zyada** hoga. Jo neighbor door hoga, uska asar missing value par kam padega.
* **Kaise Kaam Karta Hai:** Yeh distance ke inverse ($1 / \text{distance}$) ke basis par **Weighted Average** nikalta hai. Kareebi padosi ki value ka contribution zyada hota hai.
* **Kab Use Karein:** Jab data points thode scattered hon aur aap chahte hon ki jo bilkul nazdeek ka record hai, wahi missing value ko sabse zyada dominate kare.

$$\text{Weight}_i = \frac{1}{\text{dist}_i}$$

$$\text{Value} = \frac{\sum (\text{Weight}_i \times v_i)}{\sum \text{Weight}_i}$$

---

### 📊 Quick Comparison

| Feature | `uniform` | `distance` |
| :--- | :--- | :--- |
| **Weightage** | Sabhi neighbors ko equal weight | Nazdeek wale ko zyada, door wale ko kam |
| **Calculation** | Simple Mean (Average) | Weighted Mean (Inverse Distance) |
| **Outlier Sensitivity** | Thoda kam sensitive | Agar koi neighbor bohot close hai toh wo output ko zyada influence karega |
