* **Dimension:** 3 Input Columns ($x_1, x_2, x_3$) + 1 Output Column ($y$) $\rightarrow$ **4D Structure**

* **Step 1:** Multiple Regression ko hum graph me visually nahi dekh sakte, kyunki kabhi-kabhi isme $n$ number of columns (high dimensions) hote hain.

* **Step 2:** Lekin agar yahi 2D me hota, toh iska equation ye hota[cite: 3]:

$$y = mx + b$$
$$y = \beta_0 + \beta_1 x$$

* **Step 3:** Agar hum usi 4D wale structure (3 inputs + 1 output) ko equation me likhein, toh ye hoga:

$$\hat{y} = \beta_0 + \beta_1 x_1 + \beta_2 x_2 + \beta_3 x_3$$

* **Note:** Yahan humne $y$ ke badle $\hat{y}$ use kiya hai kyunki:
  * **$y$**: Hamara **actual value** hai jo dataset ke output column me already maujood hai.
  * **$\hat{y}$**: Hamara **predicted value** hai jo hum is equation ki help se calculate karte hain.

* **Step 4:** Let say hum ek example lete hain jahan hamare paas 100 students ka data hai, toh hamare dataset ka shape hoga:

$$\text{Shape} = (100, 4)$$

    (Yahan 100 rows/students hain aur 4 columns: 3 inputs + 1 output).

* **Step 5:** Ab hum har student ka predicted value ($\hat{y}$) nikalna chahte hain.
  * Yahan hum assume kar rahe hain ki hume $\beta_0$ se $\beta_3$ tak ki values already mil chuki hain.
  * Humne sirf $\beta_3$ tak hi use kiya kyunki hamare paas total 4 columns hain (3 input features + 1 output target). Har feature ke liye ek weight ($\beta_1, \beta_2, \beta_3$) aur ek bias/intercept ($\beta_0$).


* **Step 6:** Har ek student ka predicted output ($\hat{y}$) matrix form me is tarah represent hoga[cite: 4]:

$$\begin{bmatrix} \hat{y}_1 \\ \hat{y}_2 \\ \vdots \\ \hat{y}_{100} \end{bmatrix} = \begin{bmatrix} \beta_0 + \beta_1 x_{11} + \beta_2 x_{12} + \beta_3 x_{13} \\ \beta_0 + \beta_1 x_{21} + \beta_2 x_{22} + \beta_3 x_{23} \\ \vdots \\ \beta_0 + \beta_1 x_{100,1} + \beta_2 x_{100,2} + \beta_3 x_{100,3} \end{bmatrix}$$

---

**Individual Calculations (1st, 2nd, aur Last Student ke liye):**
* **1st Student ($\hat{y}_1$):**
  $$\hat{y}_1 = \beta_0 + \beta_1 x_{11} + \beta_2 x_{12} + \beta_3 x_{13}$$
* **2nd Student ($\hat{y}_2$):**
  $$\hat{y}_2 = \beta_0 + \beta_1 x_{21} + \beta_2 x_{22} + \beta_3 x_{23}$$
* **100th Student ($\hat{y}_{100}$):**
  $$\hat{y}_{100} = \beta_0 + \beta_1 x_{100,1} + \beta_2 x_{100,2} + \beta_3 x_{100,3}$$

---

> 📌 **Notation Samajhne Ke Liye ($x_{ij}$):**
> * Yahan $x_{ij}$ me pehla index ($i$) **Student (Row number)** ko darshata hai aur doosra index ($j$) **Feature (Column number)** ko:
>   * $x_{11}$: 1st row (Student 1) ka 1st input (cgpa)[cite: 6]
>   * $x_{12}$: 1st row (Student 1) ka 2nd input (iq)[cite: 6]
>   * $x_{13}$: 1st row (Student 1) ka 3rd input (gender)[cite: 6]
```

______________________________


* **Step 7: Generalized Form ($n$ Rows aur $m$ Features)**

Ab hum assume karte hain ki dataset me:
* Total rows = $n$ (Students / Data points)
* Total input features = $m$ ($x_1, x_2, \dots, x_m$)
* Total columns = $m$ inputs + $1$ output ($y$)

---

**Generalized Matrix Equation:**

$$\hat{Y} = \begin{bmatrix} \hat{y}_1 \\ \hat{y}_2 \\ \vdots \\ \hat{y}_n \end{bmatrix} = \begin{bmatrix} \beta_0 + \beta_1 x_{11} + \beta_2 x_{12} + \dots + \beta_m x_{1m} \\ \beta_0 + \beta_1 x_{21} + \beta_2 x_{22} + \dots + \beta_m x_{2m} \\ \vdots \\ \beta_0 + \beta_1 x_{n1} + \beta_2 x_{n2} + \dots + \beta_m x_{nm} \end{bmatrix}$$

* Yahan pure prediction vector ko humne capital **$\hat{Y}$** maan liya hai.

---

> 📌 **General Term ($i$-th Data Point ke liye):**
> $$\hat{y}_i = \beta_0 + \sum_{j=1}^{m} \beta_j x_{ij}$$
> * Yahan $i$ row index ($1$ se $n$) hai aur $j$ feature index ($1$ se $m$) hai.

___

* **Step 8:** Ab hum thoda dimaag laga rahe hain aur saare $\beta$ hata de rahe hain:

$$= \begin{bmatrix} 
1 & x_{11} & x_{12} & x_{13} & \dots & x_{1m} \\ 
1 & x_{21} & x_{22} & x_{23} & \dots & x_{2m} \\ 
\vdots & \vdots & \vdots & \vdots & \ddots & \vdots \\ 
1 & x_{n1} & x_{n2} & x_{n3} & \dots & x_{nm} 
\end{bmatrix} \begin{bmatrix} 
\beta_0 \\ 
\beta_1 \\ 
\beta_2 \\ 
\vdots \\ 
\beta_m 
\end{bmatrix}$$

* Bas sochna kya hai ki yahan pe humne is matrix ko do matrices ke product ke form me decompose kar diya hai.

> 💡 **Dhyan dene wali baat:**
> * Pehle column me **$1$** isliye add kiya hai taaki intercept ($\beta_0$) multiply hokar directly har equation me add ho sake ($1 \times \beta_0 = \beta_0$).
> * First matrix hamara **Feature Matrix ($X$)** ban gaya aur second matrix hamara **Parameter Vector ($\beta$)** hai.

___

* **Step 9: Compact Matrix Form**

Pichle decomposed matrices ko compact notation mein aise likhte hain[cite: 8]:

$$\hat{Y} = X \beta$$

* **$\hat{Y}$:** Prediction vector[cite: 8]
* **$X$:** Feature matrix (jisme 1s ka pehla column add kiya gaya hai)[cite: 8]
* **$\beta$:** Coefficients/weights ka vector[cite: 8]

___

* **Step 10:** Ye mera $X$ hai:

$$X \rightarrow \begin{matrix}
 & \text{cgpa} & \text{gender} & \text{iq} \\
 & x_1 & x_2 & x_3 \\
\begin{matrix} 1 \\ 1 \\ 1 \end{matrix} & \begin{array}{|c|c|c|}
\hline
x_{11} & x_{12} & x_{13} \\
\hline
x_{21} & x_{22} & x_{23} \\
\hline
x_{31} & x_{32} & x_{33} \\
\hline
\end{array}
\end{matrix}$$

___

* **Step 11:** Humne $\hat{Y}$ bana liya hai, ab hum ek **$Y$** matrix (vector) banate hain[cite: 10]:

$$Y = \begin{bmatrix} y_1 \\ y_2 \\ \vdots \\ y_n \end{bmatrix}$$

* Matlab saare **actual outputs** ki values ka matrix[cite: 10].

___

* **Step 12: Error Vector ($e$)**

Ab hum ek aur naya matrix (vector) bana rahe hain jise **$e$** bolenge, jo ki actual aur predicted values ka difference hoga ($Y - \hat{Y}$)[cite: 11]:

$$e = Y - \hat{Y} = \begin{bmatrix} y_1 \\ y_2 \\ \vdots \\ y_n \end{bmatrix} - \begin{bmatrix} \hat{y}_1 \\ \hat{y}_2 \\ \vdots \\ \hat{y}_n \end{bmatrix}$$

$$e = \begin{bmatrix} y_1 - \hat{y}_1 \\ y_2 - \hat{y}_2 \\ \vdots \\ y_n - \hat{y}_n \end{bmatrix}$$

* **Step 13: Error Function (Recall & Matrix Formulation)**

Thoda piche chalte hain aur ek baar error wala formula yaad karte hain:

> ⚠️ **Important:** Ye formula **Single Linear Regression** ka hai, is baat ko yaad rakhna!

$$E = \sum_{i=1}^{n} (y_i - \hat{y}_i)^2$$

---

Is same formula ko ab hum matrix form me aise likh sakte hain:

$$E = e^T e$$

* Yahan **$e^T$** ka matlab hai error vector $e$ ka **Transpose**.
* **Note:** Jab aap $e^T \cdot e$ calculate karoge, toh aapko exact wahi sum of squared errors ($E$) mil jayega:

$$\begin{bmatrix} (y_1 - \hat{y}_1) & (y_2 - \hat{y}_2) & \dots & (y_n - \hat{y}_n) \end{bmatrix} \begin{bmatrix} y_1 - \hat{y}_1 \\ y_2 - \hat{y}_2 \\ \vdots \\ y_n - \hat{y}_n \end{bmatrix} = \sum_{i=1}^{n} (y_i - \hat{y}_i)^2$$

____

* **Step 14: Error Function Expansion in Matrix Terms**

Ab hum $E = e^T e$ ko expand karte hain:

$$E = e^T e = (Y - \hat{Y})^T (Y - \hat{Y})$$

$$= (Y^T - \hat{Y}^T)(Y - \hat{Y})$$

Hum jaante hain ki $\hat{Y} = X\beta$, toh $\hat{Y}$ ki jagah $X\beta$ rakhne par:

$$= [Y^T - (X\beta)^T](Y - X\beta)$$

Ab brackets ko multiply karke open karenge:

$$E = Y^T Y - Y^T X\beta - (X\beta)^T Y + (X\beta)^T X\beta$$

### ***This is the proof Why Both Tearm are Equal***

___

### ****Complete Mathematial Formulation****

___

# ****Why Gradient Descent****


#### **Step 15: Why Gradient Descent? (OLS vs Gradient Descent)**

Scikit-Learn Linear Regression ki problem ko do tareeqon se solve karta hai:

```text
               Linear Regression Problems
                      /        \
                     /          \
                   OLS          SGDRegressor
            (LinearRegression)  (Gradient Descent)

```

1. **Closed Form Solution (OLS):** `LinearRegression` class me use hota hai.
2. **Gradient Descent:** `SGDRegressor` class me use hota hai.

---

**Sawaal: Jab OLS se direct answer mil raha hai toh Gradient Descent ki kya zaroorat?**

* OLS formula se directly saare coefficients ($\beta$) ek hi baar me nikal jaate hain:

$$\beta = (X^T X)^{-1} X^T Y$$

* **Dikkat (The Bottleneck):** Formula me jo **$(X^T X)^{-1}$** (Matrix Inversion) operation hai, uski computational complexity bohot zyada hoti hai ($O(m^3)$). Agar data me columns/features bohot zyada ho gaye, toh ye algorithm bohot slow ho jata hai aur system memory crash ho sakti hai.

---

**Solution: Gradient Descent (SGDRegressor)**

* Yahan hum direct inverse wala formula nahi lagate.
* Hum weights ko random value se initialize karte hain aur step-by-step (dheere-dheere) error ko minimize karte hue optimal coefficients ke paas pahunchte hain.
* Ye large datasets aur high-dimensional data par OLS se kafi fast aur memory-efficient hota hai.
