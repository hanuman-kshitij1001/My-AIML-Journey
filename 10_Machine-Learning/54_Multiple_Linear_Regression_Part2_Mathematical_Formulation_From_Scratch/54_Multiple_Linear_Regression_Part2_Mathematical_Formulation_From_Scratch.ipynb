### 🎯 ****Session Agenda: Finding Optimal Coefficients ($\beta$)****

Aaj ke session ka core goal ye samajhna hai ki **Multiple Linear Regression** equation mein coefficients ($\beta$) kaise calculate hote hain:

$$\hat{y} = \beta_0 + \beta_1 x_1 + \beta_2 x_2 + \dots + \beta_n x_n$$

Hum inhi **optimal coefficients** $(\beta_0, \beta_1, \beta_2, \dots, \beta_n)$ ki calculation aur intuition par baat karenge.

---

> 💡 **Quick Intuition on $\hat{y}$ vs $y$:**
> * Hum yahan **$\hat{y}$** (y-hat) use kar rahe hain kyunki ye hamare model ki **predicted value** hai, na ki actual target ($y$).
> * Training ke dauran hamara objective actual error ($\varepsilon$) ko minimize karke residual baseline ko optimize karna hota hai.
