# ****Multiple Linear Regression (MLR)****

## ***Definition***

Multiple Linear Regression (MLR) ek supervised machine learning algorithm hai jahan hum **ek se zyada independent features ($X_1, X_2, \dots, X_n$)** ka use karke **ek continuous target variable ($Y$)** ko predict karte hain.

Simple Linear Regression mein sirf $1$ input feature hota tha:


$$Y = \beta_0 + \beta_1 X_1$$

Multiple Linear Regression mein multiple input features hote hain:


$$Y = \beta_0 + \beta_1 X_1 + \beta_2 X_2 + \beta_3 X_3 + \dots + \beta_n X_n + \varepsilon$$

Yahan:

* $Y$ = Target variable (e.g., Placement Package)
* $\beta_0$ = Intercept (bias term)
* $\beta_1, \beta_2, \dots, \beta_n$ = Coefficients (har feature ke weights)
* $X_1, X_2, \dots, X_n$ = Independent input features (e.g., CGPA, IQ, Coding Score)
* $\varepsilon$ = Error term (residual noise)

---

## ****Geometric Intuition: Line se Hyperplane tak****
* **2D Space (Simple Linear Regression):**
* $1$ Input Feature ($X$) + $1$ Output ($Y$)
* Data points 2D plane par scatter hote hain.
* Model ka solution ek **2D Best Fit Straight Line** banti hai.


* **3D Space (Multiple Linear Regression with 2 Features):**
* $2$ Input Features ($X_1$ = CGPA, $X_2$ = IQ) + $1$ Output ($Y$ = Package)
* Data points ek 3D room/cube ke andar points ki tarah bikhre hote hain.
* Best fit solution ab line nahi hoti, balki ek flat sheet yaani **2D Best Fit Plane** banta hai.


* **n-D Space (Higher Dimensions / >2 Features):**
* Jab features 3 ya usse zyada ho jate hain, toh hum use visual graph me draw nahi kar sakte.
* Geometrically use hum **Hyperplane** bolte hain.
* Model ka objective wahi rehta hai: aisa hyperplane dhundhna jo sabhi data points ke vertical distance (residuals) ke squared error ko minimize kare.

##### ****Image me jo likha hai, wo **3D Space (2 Inputs + 1 Output)** ke liye linear regression ki equation hai:****

$$y = m x_1 + n x_2 + b \quad \longrightarrow \quad y = \beta_0 + \beta_1 X_1 + \beta_2 X_2$$

Short me har component ka matlab:

* **$y$:** Output / Target variable (jaise **Package**).
* **$X_1$ aur $X_2$:** Do input features (jaise $X_1$ = **CGPA**, $X_2$ = **IQ / Coding Score**).
* **$\beta_0$ (ya $b$):** Intercept / Bias term — agar dono inputs zero hon, toh baseline output kitna hoga.
* **$\beta_1$ (ya $m$):** Feature $X_1$ ka slope / weight — $X_1$ ke 1 unit badhne par $y$ kitna change hoga.
* **$\beta_2$ (ya $n$):** Feature $X_2$ ka slope / weight — $X_2$ ke 1 unit badhne par $y$ kitna change hoga.

**3D Intuition:**
2D me ek input tha toh equation line banati thi ($y = mx + b$). Yahan 2 inputs hain, isliye yeh equation 3D space ke andar ek **2D Flat Plane** ko represent karti hai.

### `Note - 4 -D data ko hyper Equation bolte hai`

**Coefficients ($\beta_1, \beta_2, \dots, \beta_n$)** ka matlab machine learning mein simple shabdon mein har feature ka **weight ya importance** hota hai.

Equation ke hisaab se:


$$y = \beta_0 + \beta_1 X_1 + \beta_2 X_2$$

Yahan do main baatein samajhni hoti hain:

* **Unit Change ka Asar:** $\beta_1$ batata hai ki agar baki sabhi features constant (unchanged) rahein aur feature $X_1$ ko **$1$ unit** badha diya jaye, toh output $y$ mein kitna badlaav aayega.
* *Example:* Maan lo $X_1$ = CGPA aur $\beta_1 = 1.5$. Iska matlab hai agar bande ki CGPA $1$ point badhti hai, toh uska package **$1.5$ LPA** badh jayega.


* **Sign ka Matlab (Positive vs Negative):**
* **Positive Coefficient ($+\beta$):** Feature badhne par output bhi badhega (jaise CGPA ya DSA score badhne par Package badhta hai).
* **Negative Coefficient ($-\beta$):** Feature badhne par output ghatega (jaise college me backlogs ya bunk classes badhne par Package ghat-ta hai).



Short mein: Coefficient yeh decide karta hai ki prediction nikaalte waqt kis feature ki baat ko kitni taqat (importance) deni hai.
