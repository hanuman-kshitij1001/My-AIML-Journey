* Suppose you have 1 million training examples.
* In Batch Gradient Descent, for one parameter update, you calculate the gradient using all 1 million examples:

**So:**
* ❌ Very slow for huge datasets
* ❌ Requires a lot of memory/computation
* ❌ You have to process the entire dataset before one update
* ❌ If dataset is huge, training becomes expensive
`

But it has one advantage: the gradient is relatively stable/smooth.

### ****Then came Stochastic Gradient Descent (SGD)****

* Instead of using all examples, SGD uses only one example to calculate the gradient and update weights.

* "Ek-ek example dekhunga aur har example ke baad step lunga."


___


___

#### ****Steps in Stochastic Gradient Descent (SGD)****

1. **Initialize the weights**  
   Start with random or zero values for the weights.

2. **Pick one training example**  
   Select a single data point `(Xᵢ, yᵢ)`.

3. **Make a prediction**  
   Calculate the predicted value `ŷᵢ`.

4. **Calculate the Loss**  
   Find the error between actual value `yᵢ` and predicted value `ŷᵢ`.

5. **Calculate the Gradient**  
   Calculate how much the loss changes with respect to the weights.

6. **Update the Weights**
