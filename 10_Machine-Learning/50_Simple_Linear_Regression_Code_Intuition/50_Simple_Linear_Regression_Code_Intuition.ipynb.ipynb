`Problem Statment-`  ***Hamra Kaam Hai ek Model Banna Jo ki Lenear Regration Model ho Jo CGPA Ka Input Leke apko Package Bataye Done***

****Step 1 - Sabse pahele Apko Ye data ko Plot karna Honaga bhai****

```python
import pandas as pd
import numpy as np
```

```python
# cgpa = [7.1, 4.7, 8.9, 8.1]
# package = [3.5, 1.2, 4.2, 3.9]
```

```python
np.random.seed(42)
# cgpa = np.random.uniform(4, 10, 100)
# cgpa = np.round(cgpa, 2)
cgpa = np.round(np.random.uniform(4, 10, 100), 2)
cgpa
```

```python
# np.random.seed(42)
# package = np.random.uniform(1, 15, 100)
# package = np.round(package, 2)
# # CGPA ke basis par package + random noise
package = 1.5 * cgpa - 6 + np.random.normal(0, 1, 100)
package = np.round(package, 2)
package
```

```python
df = pd.DataFrame({
    'CGPA' : cgpa,
    'Package':package
})
```

```python
df
```

```python
import matplotlib.pyplot as plt
```

```python
plt.scatter(df['CGPA'], df['Package'])
plt.xlabel('CGPA')
plt.ylabel('package')
plt.title('CGPA vs Package')
plt.show()
```

___

### *******Now Hamne Khud se Sikh Liye ab Ham data import kar lete hai*******

```python
data = pd.read_csv("placement.csv")
```

```python
data
```

```python
plt.scatter(data['cgpa'], data['package'])
plt.xlabel('CGPA')
plt.ylabel('Package (in-LPA)')
plt.title("CGPA vs PACKAGE")
```

****Note- ye Data apna compelte liner nahi hai**** 

#### ****So From Here we Can Draw Best Fit Line That is What Linear Regration Does****

```python
from sklearn.linear_model import LinearRegression
```

```python
X = data[['cgpa']]      # Input → 2D hona chahiye
y = data['package']     # Output → 1D
```

```python
model = LinearRegression()

model.fit(X, y)
```

```python
y_pred = model.predict(X)
```

```python
plt.scatter(data['cgpa'], data['package'], label='Actual Data')
plt.plot(data['cgpa'], y_pred, color='red', label='Best Fit Line')

plt.xlabel('CGPA')
plt.ylabel('Package (in LPA)')
plt.title('CGPA vs Package')

plt.legend()
plt.show()
```

___

### ***Core Intuition Behind Linear Regression***

Linear Regression ka main goal hai **data points ke beech ek best-fit straight line find karna**, jisse hum naye data ki prediction kar sakein.

Example: CGPA → Package

Model ek equation seekhta hai:

**y = mx + c**

- **m (slope):** CGPA badhne par package kitna change hoga.
- **c (intercept):** Jab CGPA 0 ho, tab line ka starting value.
- **y:** Predicted package.

Model aisi line choose karta hai jisme **actual values aur predicted values ka difference (error) minimum ho**.

**Simple intuition:**  
Actual data → Best-fit line → Prediction
