### **Now Lets Find Out Loss Function**

`Maximum Likelihood (4:48): Is technique se hum model ki probabilities ko multiply karte hain. Jis model ka product sabse zyada hota hai, wahi best model mana jata hai kyunki wo data ko ache se classify kar raha hota hai.`

* 1. Problem with Product (Maximum Likelihood):
* `Issue:` Jab dataset bada hota hai (jaise 10,000 points), to bahut saare chhote numbers (0 se 1 ke beech) ko multiply karne se result itna chhota ho jata hai ki computer ke liye calculate karna mushkil (underflow issue) ho jata hai

* 2. The Solution (Using Logarithms):
* Maths ka ek rule hai:
* log(AxB) = log(A) + log(B)
* Isliye, product ko sum mein badalne ke liye hum Log ka use karte hain 
* Isse multiplication ka complicated operation, addition (summation) mein convert ho jata hai, jo calculation ke liye bahut stable hota hai 

* 3. Cross Entropy:

* Log nikaalne ke baad values negative ho jati hain (kyunki 0-1 ke beech ke numbers ka log negative hota hai). Isliye hum poore equation ko negative sign (-) se multiply karte hain 
* Is final formula ko Cross Entropy kaha jata hai

#### **Binary Cross Entropy (Log- Loss) equation** 


$$Loss = -\frac{1}{n} sum_{i=1}^{n} [y_i log(hat{y}_i) + (1 - y_i) log(1 - hat{y}_i)]$$

**Is equation ke components ka matlab:**

*   **$n$**: Total number of data points.
*   **$y_i$**: Aapka actual label (0 ya 1).
*   **$hat{y}_i$**: Aapke model dwara predict ki gayi probability.
*   **$sum$**: Sabhi points ka sum (addition).

Yeh equation batata hai ki aapka model kitni galti (error) kar raha hai, jise humein **Gradient Descent** ka use karke minimize karna hota hai.
