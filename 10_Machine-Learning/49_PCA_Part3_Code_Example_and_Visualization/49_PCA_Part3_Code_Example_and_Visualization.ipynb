## ****Linear Regration****

#### *****Problem***** - ****"Sabse badi problem ye aa rahi hai mujhe samajhne me ki kaha pe mai ye algorithm lagaunga, ye ratna padega kya mujhe?"****

___

*  **Nahi bhai, ratna nahi padega. Ye confusion ML shuru karte waqt almost sabko hota hai: “Algorithm toh samajh aa gaya, lekin real problem mein kaunsa lagana hai?”**

* **Iska solution algorithm yaad karna nahi, problem ko identify karna seekhna hai.**

### ****Sabse pehle ye 2 questions pucho:****

1. ****Mujhe kya predict karna hai?****

* ***Number → Regression***
* ***Category / Class → Classification***
* ***Groups banane hain → Clustering***

2. ****Agar Regression hai, toh relationship kaisa hai?****

* ***Ek feature + straight relationship → Simple Linear Regression***
* ***Multiple features + straight relationship → Multiple Linear Regression***
* ***Relationship curve jaisa hai → Polynomial Regression***

### *****Tumhe abhi algorithms ratne ki jagah ye skill develop karni hai:*****

`Problem → Input → Output → Relationship → Algorithm`
