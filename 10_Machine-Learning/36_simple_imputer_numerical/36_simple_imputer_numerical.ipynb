# ***Handling Missing Numerical Data: Univariate vs Multivariate***

Agar hum numerical data me missing values fill (impute) karne ki baat karein, toh 2 main tareeqe hote hain:

---

### ***1. Univariate Imputation***
* **Matlab:** Agar aapke paas multiple columns hain (jaise column 1, 2, 3, 4) aur maan lo **column 2 me koi value missing hai**.
* Is missing value ko fill karne ke liye aap **sirf aur sirf column 2 ki baaki values** ko dekhoge (jaise unka mean, median nikaal loge).
* Aapko baaki columns (1, 3, 4) se koi matlab nahi hota.

---

### ***2. Multivariate Imputation***
* **Matlab:** Agar **column 2 me value missing hai**, toh aap usko akele column 2 ke hisaab se nahi bharoge.
* Aap us missing value ka andaza lagane ke liye **baaki sabhi columns (1, 3, 4) ki values ka sahara loge** (unki aapas ki relationship ya pattern dekh kar value predict karoge).
* **Main Algorithms:**
  * **KNN Imputer:** Baaki columns ke basis par sabse milte-julte (nearest) rows dhoondh kar value fill karta hai.
  * **Iterative Imputer (MICE):** Baaki columns ko use karke regression equation banata hai aur value fill karta hai.

# ***Univariate Imputation Techniques (Numerical Data)***

Aaj hum yahan pe **4 alag-alag tarike** se missing values fill karna seekhne wale hain:

1. **Mean / Median Imputation**
2. **Arbitrary Value Imputation** (Koi bhi constant value fill kar dena jaise -1, 999)
3. **End of Distribution Imputation** (Distribution ke extreme end wali value fill karna)
4. **Random Sample Imputation** (Dataset me se hi randomly koi value utha kar fill karna)

---

> **Note:**  
> Last me hum ye bhi dekhenge ki Scikit-Learn (`sklearn`) ka use karke inme se kaun sa tareeqa sabse best aur efficiently kaam karta hai.

____

# ***1. Mean / Median Imputation***

Kuch nahi, bas column ki missing values ko us column ke **Mean** ya **Median** se bhar do.

---

### ***Sawaal: Kab Kaun Sa Use Karein? (Mean vs Median)***

1. **Agar Normal Distribution hai (Symmetric / Bell-shaped):**
   * Toh chahe Mean use karo ya Median, dono lagbhag barabar hote hain aur ek jaisa hi behave karenge.
2. **Agar Distribution Skewed hai (Right-skewed / Left-skewed):**
   * Is case me outliers ki wajah se Mean shift ho jata hai (skew ho jata hai).
   * Yahan par **Median use karna zyada better** aur safe rehta hai kyunki Median outliers se affect nahi hota.

---

### ***Benefits (Faayde):***

1. **Server / Production Simplicity:**
   * Jab production server par model deploy karte ho ya pipeline recreate karte ho, toh Mean/Median bas ek single number hota hai, implement karna bohot aasan ho jata hai.
2. **Easy to Implement:**
   * Pandas me bas ek line ka code hota hai, bohot aasaani se missing values handle ho jaati hain.
3. **Decent Results:**
   * Agar missing values ka count kam hai, toh ye bina kisi complex calculation ke achha result de deta hai.

---

### ***Disadvantages (Nuksaan):***

1. **Distribution ka Shape Change Ho Jata Hai:**
   * Ek hi value (Mean ya Median) baar-baar fill hone se us point par data ka spike (peak) ban jata hai aur original distribution ka shape badal jata hai.
2. **False Outliers Create Hote Hain:**
   * Actually wo values outlier nahi hoti, lekin distribution shrink/distort hone ki wajah se system/model normal values ko bhi outliers samajhne lagta hai.
3. **Covariance & Correlation Change Ho Jata Hai:**
   * Jis column me aapne values fill ki hain, uska baaki columns ke saath jo original relationship/correlation tha wo change ho jata hai (Covariance distort ho jata hai).

---

### ***Kab Use Karna Chahiye?***

1. **Data MCAR ho** (Missing Completely at Random).
2. **Missing data 5% se kam ho** (< 5%).
3. **Jab simple aur quick baseline model** banana ho bina heavy computation ke.

---

### ***Quick Implementation (Pandas):***

```python
# Mean Imputation
df['Age_mean'] = df['Age'].fillna(df['Age'].mean())

# Median Imputation
df['Age_median'] = df['Age'].fillna(df['Age'].median())

____

# ***2. Arbitrary Value Imputation***

Is method me hum kya karte hain ki missing values ko **koi bhi ek random/arbitrary constant value** se fill kar dete hain. 

Jaise categorical data me hum missing values ko seedha `"Missing"` text se replace kar dete hain, waise hi numerical data me hum koi extreme ya unique number daal dete hain (jaise `-1`, `99`, `999`, `-999`).

Yahan pe aap value choose karne ke liye free hote ho, aapka main goal bas missing data aur present data ke beech ek clear **difference/flag create karna** hota hai.

---

### **Benefits (Faayde):**
* **Super Easy to Apply:** Isko apply karna bohot hi simple aur fast hai, koi calculation nahi karni padti.

---

### Disadvantages (Nuksaan):
1. **PDF (Probability Density Function) Graph Distort Ho Jata Hai:**
   * Ek hi arbitrary value bohot baar repeat hone se distribution graph ka shape poori tarah bigad jata hai.
2. **Variance Change Ho Jata Hai:**
   * Data ka overall spread/variance drastically change ho jata hai.
3. **Covariance & Correlation Change Ho Jata Hai:**
   * Uss column ka baaki features ke saath relation aur covariance distrupt ho jata hai.

---

> **Note:**  
> Ye wali technique real-world me bohot zyada use nahi hoti hai, par kyunki hum sab kuch comprehensively cover kar rahe hain, isliye iska idea hona zaroori hai.

---

### Quick Implementation:

```python
# Pandas
df['Age_arbitrary'] = df['Age'].fillna(99)
# ya negative extreme value
df['Age_arbitrary_neg'] = df['Age'].fillna(-1)

# Scikit-Learn
from sklearn.impute import SimpleImputer
imputer = SimpleImputer(strategy='constant', fill_value=99)
df['Age_imputed'] = imputer.fit_transform(df[['Age']])

___

### ***Sabse Important Baat (MCAR vs MNAR):***

* **Mean / Median Imputation:**  
  Aap missing values ko Mean ya Median se tab replace karte ho jab aapka data **Randomly Missing (MCAR)** hota hai.

* **Arbitrary Value Imputation:**  
  Ye technique aap specially tab use karte ho jab aapko lagta hai ki **Data is NOT Missing at Random (MNAR)**.

---

### ***Aisa Kyun?***

Jab data randomly missing nahi hota (matlab missing hone ke peeche koi specific reason ya pattern hota hai), tab agar aap Mean/Median daal doge toh model ko lagega ki wo ek normal average value hai. 

Arbitrary value (jaise `-1` ya `999`) daalne se model ko ek clear signal milta hai ki **ye data actually missing tha aur iska missing hona bhi ek important pattern hai.**

# ***3. End of Distribution Imputation***

Ye imputation technique **Arbitrary Value Imputation ka hi ek direct extension** hai.

Arbitrary wale me dikkat kya thi ki kaun sa random number choose karein (jaise -1, 999 ya koi aur), wo manually decide karna thoda mushkil tha. 

Is technique me hum kya karte hain ki distribution ke bilkul **extreme end (tail) ki value** utha kar missing values me fill kar dete hain. Iska main goal bhi model ko yahi signal dena hota hai ki yahan missing values the.

---

### ***End Value Calculate Karne Ke 2 Cases:***

* **Case 1: Agar Column Normally Distributed Hai (Normal Distribution)**
  * Yahan hum **3-Sigma Rule** use karte hain:
    $$\text{Fill Value} = \text{Mean} + 3\sigma \quad \text{ya} \quad \text{Mean} - 3\sigma$$
  * Kyunki normal curve me $3\sigma$ ke aage ka region extreme/outlier zone hota hai.

* **Case 2: Agar Column Skewed Hai (Right / Left Skewed)**
  * Yahan hum **IQR Proximity Rule** use karte hain:
    $$\text{Fill Value} = Q_3 + 1.5 \times \text{IQR} \quad (\text{Right End})$$
    $$\text{ya}$$
    $$\text{Fill Value} = Q_1 - 1.5 \times \text{IQR} \quad (\text{Left End})$$
  * Jahan $Q_1 = 25\text{th percentile}$, $Q_3 = 75\text{th percentile}$, aur $\text{IQR} = Q_3 - Q_1$.

---

### Code Implementation:

```python
# 1. Normal Distribution ke case me:
mean = df['Age'].mean()
std = df['Age'].std()
extreme_value = mean + 3 * std
df['Age_end_dist'] = df['Age'].fillna(extreme_value)

# 2. Skewed Distribution ke case me (IQR Rule):
q1 = df['Fare'].quantile(0.25)
q3 = df['Fare'].quantile(0.75)
iqr = q3 - q1
extreme_fare = q3 + 1.5 * iqr
df['Fare_end_dist'] = df['Fare'].fillna(extreme_fare)

____

## ***End of Distribution Imputation: Summary & Trade-offs***

---

### Benefit:
* **Easy to Implement:** Arbitrary number manually dhoondhne ke jhanjhat ke bina extreme tail value mathematical formula (Mean $\pm 3\sigma$ ya IQR Rule) se asani se nikal kar apply ho jati hai.

---

### Disadvantages:
1. **PDF (Probability Density Function) Distort:** Tail par bohot saari duplicate values aane se distribution graph ka shape bigad jata hai.
2. **Covariance Change:** Baaki features ke saath jo correlation/relationship tha wo disturb ho jata hai.
3. **Variance Change:** Overall data ka spread aur variance badal jata hai.

---

### When to Use?
* **NMAR (Not Missing at Random):**  
  Ye technique tabhi use karni hai jab data **randomly missing na ho** aur missingness ka pattern model ko ek extreme signal ke roop me batana zaroori ho.

___
