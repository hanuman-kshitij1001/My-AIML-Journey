### ***Explination***

### 1. **Numerical Data:** 
Discrete ya continuous numbers (e.g., Age, Weight, Salary) $\rightarrow$ Inhe scale karne ke liye **Feature Scaling (Standardization / Min-Max Scaling)** ka use hota hai.


### ***1. Ordinal Data***
**Ordinal Data** wo categorical data hota hai jahan categories ke beech ek natural order, rank ya mathematical relation exist karta hai (yaani ek category dusri se badi ya behtar hoti hai).

---

#### Real-World Examples

* **Education / Academic Grades:**
  $$\text{Distinction} > \text{1st Class} > \text{2nd Class} > \text{Fail}$$
* **Customer Reviews / Feedback:**
  $$\text{Excellent} (5\star) > \text{Good} (4\star) > \text{Average} (3\star) > \text{Poor} (2\star) > \text{Terrible} (1\star)$$
* **T-Shirt Sizes:**
  $$\text{XL} > \text{L} > \text{M} > \text{S}$$

---

### Encoding Rule

Kyunki yahan hierarchy aur sequence matter karta hai, isliye is data ko **integers ($0, 1, 2, 3 \dots$)** me encode kiya jata hai taaki model categories ke beech ka relation aur relative weight samajh sake:

* **Independent Features ($X$):** Use karte hain **`OrdinalEncoder`**
* **Target Variable ($y$):** Use karte hain **`LabelEncoder`**

> **Important Note:** Dataset me Categorical Data hamesha **strings / text** ke format me hota hai, jabki Machine Learning ke algorithms strictly **numbers (math equations)** accept karte hain. As a Machine Learning Engineer, ye aapki duty hai ki aap un textual categories ko numbers me convert karein.

Text ko numbers me convert karne ke liye bohot saari encoding techniques hoti hain, lekin hum **2 sabse popular aur widely used techniques** detail me padhenge:

1. **Ordinal Encoding:** 
   * Naam se hi clear hai ki ye technique **Ordinal Data** ke upar apply hoti hai (jahan values ke beech rank ya natural relation hota hai, jaise: Good/Average/Poor).
2. **One-Hot Encoding (OHE):** 
   * Ye technique **Nominal Data** ke upar apply hoti hai (jahan values ke beech koi hierarchy ya order nahi hota, jaise: Male/Female, Delhi/Mumbai).

# ***Label Encoding (`LabelEncoder`)***

Machine Learning me categorical data ko encode karte waqt sabse bada confusion `OrdinalEncoder` aur `LabelEncoder` ke beech hota hai. Dono ka mathematical kaam categories ko integer numbers ($0, 1, 2, \dots$) me convert karna hi hai, lekin **dono ki application domain alag hoti hai**.

---

### ***Kaha Pe Kya Lagana Hai? (The Golden Rule)***
* **Independent Input Features ($X$):** 
  * Agar aapke input features ($X$) me koi categorical/ordinal column hai (jaise: Education Level, Rating, Customer Feedback), toh wahan **`OrdinalEncoder`** use hoga.
* **Target Output Variable ($y$):** 
  * Agar aapka Target Column ($y$) categorical hai (jaise classification problems me hota hai: `"Yes"`/`"No"`, `"Spam"`/`"Not Spam"`, `"Setosa"`/`"Versicolor"`), toh wahan `OrdinalEncoder` nahi balki specifically **`LabelEncoder`** use hota hai.

---

### ***Core Difference Table***

| Parameter | `OrdinalEncoder` | `LabelEncoder` |
| :--- | :--- | :--- |
| **Designed For** | **Input Features ($X$)** | **Target Variable ($y$)** |
| **Input Shape** | 2D Array / DataFrame ($[n\_samples, n\_features]$) | 1D Array / Series ($[n\_samples]$) |
| **Expected Behavior** | Multiple features ko ek sath encode kar sakta hai | Sirf ek single target column ko encode karta hai |
| **Typical Problem** | Feature Preprocessing / Transformation | Target Variable Encoding (Classification) |

---

### ***Scikit-Learn Implementation Code***

```python
from sklearn.preprocessing import LabelEncoder

# Target column (y) ke liye data
y_train = ['No', 'Yes', 'Yes', 'No', 'Yes']

# LabelEncoder initialize aur fit-transform karna
le = LabelEncoder()
y_encoded = le.fit_transform(y_train)

# Output: array([0, 1, 1, 0, 1])
# Mapping dekhne ke liye:
print(le.classes_)  # Output: array(['No', 'Yes'], dtype='<U3') --- ```> **Summary:** `OrdinalEncoder` is designed for 2D inputs ($X$), whereas `LabelEncoder` is strictly designed for 1D outputs ($y$).

# ***Ordinal Encoding (`OrdinalEncoder`)***

Dataset me jab kisi categorical feature ke andar **ek natural rank ya hierarchy** hoti hai, tab hum **Ordinal Encoding** apply karte hain.

---

### Step 1: Identify - Nominal ya Ordinal?

Maan lijiye hamare dataset me ek column hai: **`Education`**

Values: `['HS', 'UG', 'PG', 'PG', 'UG', 'HS', 'UG']`

* **Nominal:** Agar values me koi order nahi hota (jaise `City`), toh One-Hot Encoding lagate.
* **Ordinal:** Yahan clear sequence exist karta hai:
  $$\text{Post Graduation (PG)} > \text{Under Graduation (UG)} > \text{High School (HS)}$$
  Isliye yeh **Ordinal Data** hai.

---

### Step 2: Mapping Logic (Assigning Values by Order)

Hum Scikit-Learn ko specific hierarchy list pass karte hain:
* **High School (HS):** Lowest value $\rightarrow$ `0`
* **Under Graduation (UG):** Middle value $\rightarrow$ `1`
* **Post Graduation (PG):** Highest value $\rightarrow$ `2`

---

### Step 3: Transformation Simulation

| Row Index | Original Education | Assigned Logic | Encoded Value |
| :--- | :--- | :--- | :--- |
| **Row 1** | **HS** | Lowest Rank | **0** |
| **Row 2** | **UG** | Middle Rank | **1** |
| **Row 3** | **PG** | Highest Rank | **2** |
| **Row 4** | **PG** | Highest Rank | **2** |
| **Row 5** | **UG** | Middle Rank | **1** |
| **Row 6** | **HS** | Lowest Rank | **0** |
| **Row 7** | **UG** | Middle Rank | **1** |

---

### Scikit-Learn Implementation Code

Scikit-Learn ke `OrdinalEncoder` ko categories parameter me increasing order (`['HS', 'UG', 'PG']`) pass karna padta hai:

```python
import pandas as pd
from sklearn.preprocessing import OrdinalEncoder

# Sample DataFrame
df = pd.DataFrame({'Education': ['HS', 'UG', 'PG', 'PG', 'UG', 'HS', 'UG']})

# Categories list specify karna (Increasing Order: 0 -> 1 -> 2)
categories_order = [['HS', 'UG', 'PG']]

# OrdinalEncoder initialize aur transform karna
oe = OrdinalEncoder(categories=categories_order)
df['Education_Encoded'] = oe.fit_transform(df[['Education']])

print(df)
