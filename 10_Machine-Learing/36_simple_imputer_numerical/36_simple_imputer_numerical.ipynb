# ***Handling Missing Numerical Data: Univariate vs Multivariate***

Agar hum numerical data me missing values fill (impute) karne ki baat karein, toh 2 main tareeqe hote hain:

---

### ***1. Univariate Imputation***
* **Matlab:** Agar aapke paas multiple columns hain (jaise column 1, 2, 3, 4) aur maan lo **column 2 me koi value missing hai**.
* Is missing value ko fill karne ke liye aap **sirf aur sirf column 2 ki baaki values** ko dekhoge (jaise unka mean, median nikaal loge).
* Aapko baaki columns (1, 3, 4) se koi matlab nahi hota.

---

### ***2. Multivariate Imputation***
* **Matlab:** Agar **column 2 me value missing hai**, toh aap usko akele column 2 ke hisaab se nahi bharoge.
* Aap us missing value ka andaza lagane ke liye **baaki sabhi columns (1, 3, 4) ki values ka sahara loge** (unki aapas ki relationship ya pattern dekh kar value predict karoge).
* **Main Algorithms:**
  * **KNN Imputer:** Baaki columns ke basis par sabse milte-julte (nearest) rows dhoondh kar value fill karta hai.
  * **Iterative Imputer (MICE):** Baaki columns ko use karke regression equation banata hai aur value fill karta hai.

# ***Univariate Imputation Techniques (Numerical Data)***

Aaj hum yahan pe **4 alag-alag tarike** se missing values fill karna seekhne wale hain:

1. **Mean / Median Imputation**
2. **Arbitrary Value Imputation** (Koi bhi constant value fill kar dena jaise -1, 999)
3. **End of Distribution Imputation** (Distribution ke extreme end wali value fill karna)
4. **Random Sample Imputation** (Dataset me se hi randomly koi value utha kar fill karna)

# ***1. Mean / Median Imputation***

Kuch nahi, bas column ki missing values ko us column ke **Mean** ya **Median** se bhar do.

---

### ***Sawaal: Kab Kaun Sa Use Karein? (Mean vs Median)***

1. **Agar Normal Distribution hai (Symmetric / Bell-shaped):**
   * Toh chahe Mean use karo ya Median, dono lagbhag barabar hote hain aur ek jaisa hi behave karenge.
2. **Agar Distribution Skewed hai (Right-skewed / Left-skewed):**
   * Is case me outliers ki wajah se Mean shift ho jata hai (skew ho jata hai).
   * Yahan par **Median use karna zyada better** aur safe rehta hai kyunki Median outliers se affect nahi hota.

---

### ***Benefits (Faayde):***

1. **Server / Production Simplicity:**
   * Jab production server par model deploy karte ho ya pipeline recreate karte ho, toh Mean/Median bas ek single number hota hai, implement karna bohot aasan ho jata hai.
2. **Easy to Implement:**
   * Pandas me bas ek line ka code hota hai, bohot aasaani se missing values handle ho jaati hain.
3. **Decent Results:**
   * Agar missing values ka count kam hai, toh ye bina kisi complex calculation ke achha result de deta hai.

---

### ***Disadvantages (Nuksaan):***

1. **Distribution ka Shape Change Ho Jata Hai:**
   * Ek hi value (Mean ya Median) baar-baar fill hone se us point par data ka spike (peak) ban jata hai aur original distribution ka shape badal jata hai.
2. **False Outliers Create Hote Hain:**
   * Actually wo values outlier nahi hoti, lekin distribution shrink/distort hone ki wajah se system/model normal values ko bhi outliers samajhne lagta hai.
3. **Covariance & Correlation Change Ho Jata Hai:**
   * Jis column me aapne values fill ki hain, uska baaki columns ke saath jo original relationship/correlation tha wo change ho jata hai (Covariance distort ho jata hai).

---

### ***Kab Use Karna Chahiye?***

1. **Data MCAR ho** (Missing Completely at Random).
2. **Missing data 5% se kam ho** (< 5%).
3. **Jab simple aur quick baseline model** banana ho bina heavy computation ke.

---

### ***Quick Implementation (Pandas):***

```python
# Mean Imputation
df['Age_mean'] = df['Age'].fillna(df['Age'].mean())

# Median Imputation
df['Age_median'] = df['Age'].fillna(df['Age'].median())

____
