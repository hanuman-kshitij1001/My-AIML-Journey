# Exploratory Data Analysis (EDA)

## What is EDA?

**EDA (Exploratory Data Analysis)** is the process of **understanding and analyzing data** before applying Machine Learning or statistical models.

EDA helps us find:

* Missing values
* Outliers
* Data distribution
* Relationships between variables
* Important patterns in data

### Main Purpose

> **First understand the data, then build the model.**

---

## Types of EDA

### 1. Univariate Analysis

Analysis of **one variable** at a time.

Example: Analyzing only the `Age` column.

### 2. Bivariate Analysis

Analysis of **two variables** together.

Example: `Age` vs `Salary`.

### 3. Multivariate Analysis

Analysis of **more than two variables** together.

Example: `Age`, `Experience`, `Education` and `Salary`.

---

## Next Topic

We will start with **Univariate Analysis**, where we analyze **one variable at a time**.


# Univariate Analysis

**Univariate = Single Variable**

Jab hum **sirf ek variable (single column)** par EDA ya Data Analysis perform karte hain, toh usse **Univariate Analysis** kehte hain.

### Example

Agar dataset mein bahut saare columns hain aur hum sirf ek column, jaise `Age`, ko analyze karte hain:

```python
df["Age"]
```

Toh ye **Univariate Analysis** hai.

> **Single Variable / Single Column → Univariate Analysis**


# Types of Data

Data ko mainly **2 types** mein divide kiya jata hai:

## 1. Numerical Data

Numerical data woh data hota hai jo **numbers** mein hota hai aur jiske saath mathematical operations perform kar sakte hain.

**Examples:**

* Age
* Salary
* Height
* Marks

## 2. Categorical Data

Categorical data woh data hota hai jo **categories or labels** represent karta hai.

**Examples:**

* Gender → Male, Female
* City → Delhi, Mumbai, Vadodara
* Department → CSE, IT, AI/ML

### In Short

**Data**

* Numerical
* Categorical
