### Ye Kya Hai aur Kahan Se Aaya? (The Problem Before It)

Real-world datasets me har column ka data type alag hota hai:
* Kuch columns **Numerical** hote hain (e.g., `Age`, `Salary` $\rightarrow$ jinpar **StandardScaler** lagana hota hai).
* Kuch columns **Ordinal Categorical** hote hain (e.g., `Education` $\rightarrow$ jinpar **OrdinalEncoder** lagana hota hai).
* Kuch columns **Nominal Categorical** hote hain (e.g., `City`, `Gender` $\rightarrow$ jinpar **OneHotEncoder** lagana hota hai).
* Kuch numerical columns me missing values hoti hain (jinpar **SimpleImputer** lagana hota hai).

`ColumnTransformer` aane se pehle, hume manually dataset ko alag-alag columns me split karna padta tha, har column par alag se transformation fit-transform karke wapas sabko horizontal concatenate (`np.hstack` ya `pd.concat`) karna padta tha. Ye process bohot messy, error-prone aur data leakage risk se bhara hota tha.

---

### Solution: ColumnTransformer Kya Karta Hai?

**`ColumnTransformer`** Scikit-Learn ka ek aisa wrapper tool hai jo aapko single step me **alag-alag columns par simultaneously alag-alag preprocessing techniques apply karne ki power deta hai**, bina DataFrame ko manually split aur join kiye.
