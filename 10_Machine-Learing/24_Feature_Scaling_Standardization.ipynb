# Day 24 — Feature Scaling & Standardization

Feature Scaling brings numerical features to comparable scales.

Example:

Age → 20 to 80

Salary → 20,000 to 2,00,000

Without scaling, some algorithms may be affected by different feature magnitudes.


## Why Scaling?

Scaling is especially important for algorithms based on:

- distance
- gradient optimization
- magnitude of features

Examples:

KNN
K-Means
SVM
Logistic Regression
Linear Regression
Neural Networks

Tree-based models generally do not require feature scaling.


# Standardization

Standardization transforms data using:

z = (x - mean) / standard_deviation

After standardization:

Mean ≈ 0

Standard deviation ≈ 1


```python
import numpy as np
import pandas as pd

from sklearn.preprocessing import StandardScaler

data = pd.DataFrame({
    "Age": [20, 25, 30, 35, 40],
    "Salary": [20000, 40000, 60000, 80000, 100000]
})

print(data)
```

```python
scaler = StandardScaler()

scaled_data = scaler.fit_transform(data)

scaled_df = pd.DataFrame(
    scaled_data,
    columns=data.columns
)

print(scaled_df)
```

# StandardScaler — Important

`fit()` learns statistics from data.

For StandardScaler:

fit() learns:

- mean
- standard deviation

`transform()` uses those learned statistics to transform data.

`fit_transform()` does both.


## Correct ML Workflow

First split:

X_train, X_test

Then:

scaler.fit(X_train)

X_train_scaled = scaler.transform(X_train)

X_test_scaled = scaler.transform(X_test)

OR:

X_train_scaled = scaler.fit_transform(X_train)

X_test_scaled = scaler.transform(X_test)

Never fit the scaler separately on test data.


```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

X = data

X_train, X_test = train_test_split(
    X,
    test_size=0.2,
    random_state=42
)

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("Training:")
print(X_train_scaled)

print("Testing:")
print(X_test_scaled)
```

# Other Scaling Techniques

## Min-Max Scaling

Transforms values approximately into [0, 1].

Formula:

x_scaled = (x - min) / (max - min)

Use:

MinMaxScaler


```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

scaled = scaler.fit_transform(data)

print(scaled)
```

## Robust Scaling

RobustScaler uses:

- median
- interquartile range (IQR)

It can be useful when the dataset contains significant outliers.


```python
from sklearn.preprocessing import RobustScaler

scaler = RobustScaler()

scaled = scaler.fit_transform(data)

print(scaled)
```

# Quick Comparison

| Technique | Main Idea |
|---|---|
| StandardScaler | Mean = 0, Std = 1 |
| MinMaxScaler | Usually maps to 0–1 |
| RobustScaler | Uses median and IQR |

## Most Important Interview Point

Scaling should be learned only from training data.

Correct:

fit → training data

transform → training + test data

This prevents data leakage.
