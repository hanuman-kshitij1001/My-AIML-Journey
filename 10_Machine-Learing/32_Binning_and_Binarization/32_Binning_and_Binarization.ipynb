# ***Feature Engineering: Encoding Numerical Features (Binning & Binarization)***

*So pichli video me seekha tha ki how to encode the categorical variables, aaj hum uska ek doosra variation dekh rahe hain ki numerical data ko kaise encode karna hai.*

*Matlab let's say hamare paas ek numerical data hai `Age` naam ka, toh ab aap isko categorical data me kaise convert kar sakte ho, ye hum padhenge.*

---

### ***Important: Numerical Data Toh Accha Hota Hai Toh Hum Use Categorical Data Me Kyun Badlein?***

Honestly, its depend. Kabhi-kabhi aisi zaroorat aa jaati hai ki jo numerical data hai uska better representation tab hota hai jab wo categorical me ho.

* **Example (Google Play Store):**
  * Play store ke dataset me ek column tha `Number of Downloads` bolke.
  * Isme ajeeb-ajeeb values thi:
    * Jo bahut kam famous app the, unme ajeeb values thi jaise `23`.
    * Jo koi bahut bade aur famous app the, unme bahut badi values thi jaise `101426426`.
    * Phir koi chhoti values thi.
  * Toh ye jo numerical representation hai ye aapko bahut zyada help nahi karega. Jab isko plot karoge toh ye curve bahut hi right-skewed aayega ya kuch aur bhi ho sakta hai.
  * Phir humne ye numerical data ko categorical data me convert kar diya tha using bins like:
    * `1k+ Downloads`
    * `10M+ Downloads`
    * `100M+ Downloads`
  * Jaise hi maine ye kiya, toh mera poora ka poora problem simplify ho gaya aur result bhi better aane laga.

---

### *****Numerical Data Ko Categorical Me Convert Karne Ke 2 Tarike:*****

* **1. Discretization (Jisko hum Binning bolte hain)**
* **2. Binarization**

# ***Discretization (Binning)***

> **Definition:**  
> Discretization is the process of transforming continuous variables into discrete variables by creating a set of contiguous intervals that span the range of the variable's values. Discretization is also called binning, where bin is an alternative name for interval.

---

### ***Why Use Discretization?***

* **1. To Handle Outliers:** Extreme minimum ya maximum values alag-alag bins me group ho jati hain, jisse outliers ka impact neutral ho jata hai.
* **2. To Improve the Value Spread:** Continuous data ka distribution agar heavily skewed ya unevenly spread hai, toh intervals create karne se data points ka spread balance aur simplify ho jata hai.

# ***4. Equal Width / Uniform Binning***

*Yahan pe funda kya hai ki jo humne Binning ka definition padha, exactly wahi hum yahan pe karte hain.*

*Jaise maan lo `Age` bolke ek data column hai, usme alag-alag numbers hain. Toh yahan pe **khud se batana padta hai aapko bins kitne chahiye**, jaise mujhe `10 bins` chahiye (ye bins hume decide karne padte hain).*

---

### *Formula:*

**$$\text{Bin Width} = \frac{\text{Max} - \text{Min}}{\text{bins}}$$**

---

| **372** | 19.0 | 2.0 | (16.336, 24.294] |

### **Example:**
* Maan lo aapka $\text{Max} = 100$ hai aur $\text{Min} = 0$ hai.
* Toh ab interval banega: $\frac{100 - 0}{10} = 10$.
  * Pehla Interval: `0 - 10`
  * Doosra Interval: `10 - 20`
  * ...aur aise hi total 10 bins banenge.

Ab aapko kya karna hai: data table se har value ko uthana hai aur check karna hai ki ye kaun se interval me aayega, aur us hisaab se bin label assign ho jayega.

---



##### `Note`: ***apne sare interval me ek Equal width raha hai that is called Equal width / uniform Binning***
____________

# ***Equal Width Binning Ke Fayde***

---

### ***1. Outliers Handle Ho Jaate Hain***
* Extreme values (chahe bohot chhoti hon ya bohot badi) unke respective corner bins ke andar compress ho jaati hain.
* Isse outliers ka model ke mathematical weights aur calculations par padne wala bura asar ruk jata hai.

---

### ***2. Spread of Data Change Nahi Hota Hai***
* Har bin ki **width/size equal (constant)** rehti hai.
* Is wajah se data ka jo original natural distribution ya continuous spread ka shape hota hai, wo as-it-is maintain rehta hai aur distort nahi hota.
___

# ***5. Equal Frequency / Quantile Binning***

### **Isme Kya Hota Hai?**

* **1.** Aapko phir se batana padta hai ki aapko kitne bins chahiye.
* **2.** Ab yahan pe kya karoge aap jitne interval aur bins chahiye utne hi quantiles ke upar kaam karte ho.
  * **Quantile Kya Hota Hai?** Quantile wo cut points hote hain jo aapke pure sorted dataset ko barabar parts (percentages) me divide karte hain.
* Agar aapko **10 intervals** chahiye, toh:
  $$\text{Each interval contains } 10\% \text{ of the total observations}$$
* Matlab aap zero se start karke uss age tak jaoge jahan pe aapka **10th percentile** value hai (jaise `0 - 16` me 10% data). Phir `16` se wahan tak jaoge jahan pe aapko **20th percentile** value milegi (maan lo jaise `20` pe mila), toh `16 - 20`. Matlab ab `0 - 20` ke beech me 20% of population lie karti hai. Same aise hi `100%` tak jaate hain:
  * Intervals: `0-16`, `16-20`, `20-22`, `22-25`, ... `50-74`
