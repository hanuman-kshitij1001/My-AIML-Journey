# ***1. Complete Case Analysis (CCA)***

> **Complete-case analysis (CCA)**, jisko **"list-wise deletion"** bhi bolte hain, iska simple matlab hai un saari observations (rows) ko discard/remove kar dena jahan kisi bhi variable (column) me missing value ho.

> **Complete Case Analysis** ka basic matlab yahi hai ki aap dataset me se sirf aur sirf unhi observations ko analyze ya model training ke liye use karte ho jinke saare variables me complete information available hoti hai.

---

### ***Core Concept:***
* Agar kisi row ke 10 columns me se 9 columns me data hai aur sirf 1 column me `NaN` hai, toh CCA ke according wo **puri ki puri row delete** kar di jayegi.
* Aap sirf wahi rows retain karte ho jisme koi bhi missing value na ho.

```python
# Jin rows me koi bhi missing value hai unhe drop karna
df_cleaned = df.dropna()

# Ya kisi specific column ke basis par drop karna
df_cleaned = df.dropna(subset=['Age', 'Salary'])
______________________________________________________________________________________________________________________________________________________________________________________________

## ***Assumptions For CCA (Complete Case Analysis)***

Kab aap data ko remove kar sakte ho?

---

### ***1. Data is Missing Completely at Random (MCAR)***

Ye Complete Case Analysis ka **sabse bada assumption** hota hai.

* **Example se samjho:**
  * Maan lo tumhare paas `Age` column hai.
  * Usme tumhare paas **1000 values** hain aur **50 values missing** hain.
  * Agar aap Complete Case Analysis karne wale ho, toh aap ye 50 missing rows ko hata doge.
  * Transformation ke baad data shape 1000 se ghat kar **950 rows** ho jayega, columns abhi bhi utne hi rahenge.

---

### ***Ye Kab Karna Hai Aur Kab Nahi Karna Hai?***

* **Kab apply nahi karna:**
  * Aisa bhi ho sakta hai ki shuru ke 50 data missing hon ya last ke 50 data missing hon (matlab kisi specific pattern ya reason ki wajah se missing hain).
  * Is tarah ke setup me aap data remove **nahi** karoge, warna data ka distribution gadbad ho jayega aur model biased ban jayega.

* **Kab apply karna:**
  * Aap ye tabhi karoge jab aap iske baare me 100% sure honge ki jo 50 rows missing hain wo **completely randomly missing** hain (kisi particular group, order ya feature par depend nahi karti).
  * Isi condition ko **Missing Completely at Random (MCAR)** bola jata hai.

# ***Complete Case Analysis (CCA): Advantages & Disadvantages***

---

### Advantages:

1. **Easy to Implement:**
   * Isko implement karna bohot asaan hai kyunki isme koi complex data manipulation ya calculation ki zaroorat nahi padti (bas `df.dropna()` kiya aur kaam khatam).

2. **Preserves Variable Distribution:**
   * Agar aapka data **MCAR (Missing Completely at Random)** hai, toh rows drop karne ke baad bacha hua dataset ka distribution exact original dataset ke distribution jaisa hi rehta hai (shape/spread change nahi hota).

---

### Disadvantages:

1. **Bohot Zyada Data Lose Ho Sakta Hai:**
   * Agar dataset me missing values bohot zyada (abundant) hain, toh ye aapke original dataset ka ek bohot bada fraction delete kar deta hai, jisse training data kaafi kam bachta hai.

2. **Useful Information Delete Ho Sakti Hai:**
   * Agar data randomly missing nahi hai (Not MCAR), toh jo rows aap delete kar rahe ho unme bohot informative aur zaroori data ho sakta hai, jisse model biased ban jayega.

3. **Production Issue (Model Fails on Missing Data):**
   * Jab aap is model ko production me use karoge aur user/real-time input me agar koi value missing aayi, toh model ko pata hi nahi hoga ki missing data ko kaise handle karein (kyunki usne missing values ko deal karna seekha hi nahi tha).
