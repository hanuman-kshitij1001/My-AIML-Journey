### ***Sabse Pehla Aur Important Step: Convert String to Datetime***

Jab bhi aap pandas me kisi dataset ko load karte ho, toh jo `Date` ya `Time` wala column hota hai wo **by default ek Object (String)** format me load hota hai.

Agar aap usse String hi rehne doge, toh aap uske upar wo sab kuch perform nahi kar paoge (jaise Day, Month, Year, Day of Week, Quarter nikalna) jo is video me dikhaya jayega. String format par pandas ke datetime functions kaam nahi karte.

Isliye feature extraction start karne se pehle, sabse pehla kaam hota hai us string column ko **`datetime` format me convert karna**.

---

### Implementation:

```python
import pandas as pd

# String column ko datetime object me badalna
df['date'] = pd.to_datetime(df['date'])

# Ab aap asani se features extract kar sakte ho
df['day'] = df['date'].dt.day
df['month'] = df['date'].dt.month
df['year'] = df['date'].dt.year
