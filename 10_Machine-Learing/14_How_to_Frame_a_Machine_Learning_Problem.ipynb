# Day 14 — How to Frame a Machine Learning Problem

Before writing ML code, clearly define the problem.

## Six Important Questions

1. What is the problem?
2. What is the objective?
3. What is the input?
4. What is the output?
5. What type of ML problem is it?
6. How will success be measured?


## Example

Problem:

Predict whether a customer will leave a company.

Input:

Customer information.

Output:

Churn = Yes / No

This is a Classification problem.

Possible metrics:

- Accuracy
- Precision
- Recall
- F1 Score
- ROC-AUC


## Types of ML Problems

### Regression

Output is continuous.

Example:

House price prediction.

### Classification

Output belongs to categories.

Example:

Spam / Not Spam.

### Clustering

No target label.

Example:

Customer segmentation.


## Business Problem → ML Problem

Business problem:

"Reduce customer churn."

ML problem:

"Predict probability that a customer will churn."

This distinction is extremely important.
