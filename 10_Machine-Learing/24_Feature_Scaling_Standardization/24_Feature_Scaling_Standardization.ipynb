# Feature Scaling

Feature Scaling, Feature Engineering pipeline ka aakhri step hota hai. Machine Learning model ko apna data feed karne se theek pehle hum ye step perform karte hain. Ye Feature Engineering ka sabse easy aur important part hota hai.

Aaj hum Feature Scaling ke baare me baat karenge, jisme hamara main focus **Feature Standardization** par hoga.

---

### Definition
> **Feature Scaling** ek aisi technique hai jisse hum dataset ke independent features (input variables) ki values ko ek fixed range ya scale me transform karte hain.

---

### Dataset Components
Hamare dataset me mainly do tarah ke variables hote hain:
1. **Input Features (Independent Variables):** Wo features jinka use karke model pattern seekhta hai.
2. **Target Variable (Dependent Variable):** Wo value jo hum predict karte hain.

### Feature Scaling Ka Core Concept
Jab hamare input features alag-alag units ya ranges me hote hain (jaise Age: 18–60 aur Salary: 20,000–2,00,000), toh hum un sabhi columns ko ek consistent aur comparable range me scale karte hain. Is process ko hi **Feature Scaling** kaha jata hai.

## Why Do We Need Feature Scaling?

Jab hamare dataset ke features alag-alag scale ya units me hote hain, toh distance-based algorithms (jaise KNN, K-Means) ya gradient-based algorithms par iska bura asar padta hai. Isko samajhte hain ek example ke zariye:

---

### Mathematical Intuition (Euclidean Distance)

Maan lete hain hamare paas do points hain:
* $P_1 = (x_1, y_1)$
* $P_2 = (x_2, y_2)$

Do points ke beech ka **Euclidean Distance** nikaalne ka formula hota hai:

$$\text{Distance}(P_1, P_2) = \sqrt{(x_2 - x_1)^2 + (y_2 - y_1)^2}$$

---

### Example Scenario

Maan lete hain hamare features hain:
* **$x$-axis:** Salary (Ranges in thousands: ₹48,000 to ₹83,000)
* **$y$-axis:** Age (Ranges in tens: 27 to 50 years)

Ab agar hum do data rows ke beech distance calculate karein:
* $P_1 = (x_1 = 48000, y_1 = 27)$
* $P_2 = (x_2 = 83000, y_2 = 50)$

#### Step-by-Step Calculation:
1. **Salary Difference ($x$-axis):**
   $$(x_2 - x_1)^2 = (83000 - 48000)^2 = (35000)^2 = 1,225,000,000$$

2. **Age Difference ($y$-axis):**
   $$(y_2 - y_1)^2 = (50 - 27)^2 = (23)^2 = 529$$

3. **Total Distance:**
   $$\text{Distance} = \sqrt{1,225,000,000 + 529} \approx \sqrt{1,225,000,529} \approx 35000.007$$

---

### Main Problem / Conclusion
* Yaha **Salary ka impact $1,225,000,000$** hai jabki **Age ka impact sirf $529$** hai.
* Salary ki value bahut badi hone ki wajah se poora distance calculation **Salary par dominate** kar raha hai aur Age ka contribution neglect ho gaya hai.
* Isi problem ko solve karne ke liye hum **Feature Scaling** karte hain taaki sabhi features ko barabar ka importance (weightage) mile.

## Types of Feature Scaling

Feature Scaling ko broadly **do main types** me divide kiya jata hai:

1. **Standardization (Z-score Scaling)**
2. **Normalization (Min-Max Scaling & Others)**

---

### 1. Standardization (Z-score Scaling)
* Isme data ko aise scale kiya jata hai jiska **Mean ($\mu$) = 0** aur **Standard Deviation ($\sigma$) = 1** ho jata hai.
* Ye data distribution ko normal (Gaussian) shape me lane me help karta hai.

---

### 2. Normalization
Normalization ke andar data ko mainly ek specific range (jaise 0 to 1) me bound kiya jata hai. Isme kai tarah ki techniques aati hain jo hum aage cover karenge:

* **Min-Max Scaler (MinMaxScaling):** Data ko ek fixed bound (usually $[0, 1]$) me scale karta hai.
* **Robust Scaler (RobustScaling):** 
  * Ye Median aur Interquartile Range (IQR) par based hota hai.
  * **Key Advantage:** Ye **Outliers** ke upar bahut accha perform karta hai kyunki outliers se heavily affect nahi hota.
* **MaxAbs Scaler:** Data ko $[-1, 1]$ ke beech scale karta hai absolute maximum value se divide karke.
* **Mean Normalization**

# Standardization (Z-Score Normalization) - Intuition

Standardization ko **Z-Score Normalization** bhi kaha jata hai. Ye Feature Scaling ka sabse widely used method hai.

---

### Basic Concept & Intuition

Maan lete hain hamare paas do input features hain:
1. **Age**
2. **Salary**

Agar hume **Age** column ko standardize karna hai, toh hum har ek value ko transform karke ek naya transformed column (maan lete hain `Age'`) banayenge.

---

### Standardization Formula (Z-Score)

Kisi bhi feature value $x_i$ ko standardize karne ka mathematical formula hota hai:

$$x_i' = \frac{x_i - \bar{x}}{\sigma}$$

Yahan:
* $x_i'$ = Scaled / Transformed value (Z-Score)
* $x_i$ = Original feature value
* $\bar{x}$ (X-bar / $\mu$) = Feature ka **Mean** (Average)
* $\sigma$ (Sigma) = Feature ka **Standard Deviation** (Spread)

---



### Key Concept: Transformation ke baad kya hota hai?

Samajhne wali sabse important baat ye hai ki jab hum formula ($x_i' = \frac{x_i - \bar{x}}{\sigma}$) apply karte hain, toh hum feature ke **har ek particular number (row value)** ko individually transform karte hain. 

Is calculation ke baad hamare paas jo naya transformed column banta hai, usme nayi values (Z-scores) aati hain. In saari nayi values ki do sabse khaas properties hoti hain:

1. **Mean ($\mu$) hamesha $0$ ban jata hai:**
   * Jab aap transformed column ki saari nayi values ka average nikaalenge, toh wo **zero ($0$)** ke barabar aayega.
   * Iska matlab poora data ab **origin ($0$) par center** ho chuka hai.

2. **Standard Deviation ($\sigma$) hamesha $1$ ban jata hai:**
   * Transformed values ka naya spread/variance calculate karne par standard deviation **ek ($1$)** ke barabar aayega.
   * Iska matlab data ka spread ab normalize ho chuka hai aur unit variance par fix ho gaya hai.

> **In Short:** Har individual value ko transform karke pure column ka **Mean = 0** aur **Standard Deviation = 1** set karna hi **Standardization** कहलाता hai.
