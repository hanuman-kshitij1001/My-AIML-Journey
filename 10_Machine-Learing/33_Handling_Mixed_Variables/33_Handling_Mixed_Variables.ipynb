# Handling Mixed Variables in Feature Engineering

In this topic, we are going to discuss how we handle **Mixed Variables**.

---

### 1. What Is Mixed Data?

Machine Learning me jab bhi aapko real-world data milta hai, toh aksar mixed data milta hai.

Iska matlab ye hai ki **aapke single data column ke andar Numerical aur Categorical dono type ka data mixed hota hai.** Aise data ko handle karna kaafi bada headache hota hai kyunki direct model me feed karne se pehle samajh nahi aata ki ise kaise process karein.

---

### Mixed Data Aapko 2 Main Tariko Se Mil Sakta Hai:

* **Type 1: Single Cell Ke Andar Number + Text Dono Mix Hon**
  * **Example:** Train ke coach numbers jaise `B1`, `B2`, `S3`, `A1`.
  * Yahan `B` / `S` / `A` categorical part hai aur `1`, `2`, `3` numerical count hai.
  * **Solution:** Is case me hum string extraction/split karke do naye columns bana dete hain:
    * Ek **Categorical Column** (`Coach_Type`: `B`)
    * Ek **Numerical Column** (`Coach_Number`: `1`)

* **Type 2: Single Column Me Kuch Rows Numbers Hon Aur Kuch Rows Text/Letters Hon**
  * **Example:** Column ke andar values aisi hain: `[2, 3, 'A', 4, 'H', 'K', 6]`.
  * **Solution:** Isko bhi hum do alag-alag columns me split kar dete hain:
    * **Numerical Column:** Jahan sirf numbers aayenge, aur text wali jagah par `NaN` / `Null` store ho jayega.
    * **Categorical Column:** Jahan sirf letters/categories aayengi, aur number wali jagah par `NaN` / `Null` store ho jayega.

---

### Quick Implementation Logic (Pandas):

```python
import pandas as pd
import numpy as np

# Sample Data
df = pd.DataFrame({'mixed_col': [2, 3, 'A', 4, 'H', 'K', 6]})

# 1. Numerical alag karna (jo number nahi hai wo NaN ban jayega)
df['number_part'] = pd.to_numeric(df['mixed_col'], errors='coerce')

# 2. Categorical alag karna (jo number hai wo NaN ban jayega)
df['category_part'] = df['mixed_col'].apply(lambda x: x if isinstance(x, str) else np.nan)

___
