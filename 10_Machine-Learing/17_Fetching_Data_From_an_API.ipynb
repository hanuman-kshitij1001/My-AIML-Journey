# API — Real-World Example: IRCTC & Third-Party Apps

> **API = Application Programming Interface**
>
> API basically **software ko aapas mein communicate / baat karne ka way provide karti hai.**

---

## Real-World Example — IRCTC

Maan lo:

* **IRCTC / Railway System** → Railway ticketing ka main system
* **MakeMyTrip** → Travel booking platform
* **ixigo** → Travel booking platform
* **Other authorized platforms** → Third-party applications

Ye sab **alag-alag software/platforms** hain.

Lekin jab user MakeMyTrip ya ixigo par railway ticket search karta hai, toh application ko train ka data kahin se toh lena padega.

### Question

**Train ka data Railway/IRCTC system ka hai, phir MakeMyTrip ya ixigo ko ye data milta kaise hai?**

### Answer — API

IRCTC/Railway system **authorized APIs/integrations** ke through required railway information third-party platforms ko provide kar sakta hai.

Third-party application directly database mein jaakar data nahi uthati.

Instead, API ke through request aur response exchange hota hai:

```text
                    API REQUEST
┌───────────────┐  ──────────────────►  ┌──────────────────────┐
│  MakeMyTrip   │                        │ Railway / IRCTC      │
│     ixigo     │                        │       System         │
└───────────────┘  ◄──────────────────  └──────────────────────┘
                    API RESPONSE
```

---

## Complete Flow

Suppose user search karta hai:

**"Delhi → Mumbai, 25 August"**

### 1. User Request

```text
User
  ↓
MakeMyTrip
```

User MakeMyTrip par train search karta hai.

### 2. API Request

```text
MakeMyTrip
     ↓
API Request
     ↓
Railway / IRCTC System
```

MakeMyTrip API ke through railway system se required information request karta hai.

### 3. Railway System Processes Request

Railway system internally apne available data/services ko check karta hai:

```text
Train
Train Number
Route
Timing
Seat Availability
Fare
etc.
```

### 4. API Response

Railway system response API ke through wapas bhejta hai:

```text
Railway / IRCTC
      ↓
  API Response
      ↓
  MakeMyTrip
```

### 5. User ko Data Show

```text
MakeMyTrip
     ↓
   User
```

User ko app ke UI par train details dikhayi jaati hain.

---

# Data Third-Party App ke paas kaise aata hai?

Ye important concept hai.

**Railway/IRCTC ka data ≠ MakeMyTrip ka own data.**

Third-party application ko **authorized API access/integration** mil sakta hai.

API ke through application ko **specific requested information** milti hai.

For example:

```text
User
"Delhi → Mumbai trains dikhao"
        ↓
   MakeMyTrip
        ↓
   API Request
        ↓
 Railway System
        ↓
   API Response
        ↓
 Train Information
        ↓
   MakeMyTrip
        ↓
      User
```

### Important

Third-party application normally **IRCTC ka entire database download karke apne paas nahi rakh leti**.

Instead, API integration ke through **required data/services access** kiye ja sakte hain, subject to the provider's authorization, rules and technical setup.

---

# Simple Analogy

```text
Railway System
      │
      │  DATA / SERVICES
      ▼
     API
      │
      ▼
MakeMyTrip / ixigo
      │
      ▼
    User
```

### Remember

> **API = Bridge / Medium of Communication**

> **Backend / Database = Data & Business Logic**

> **Frontend = User ko information dikhane wala part**

### Overall Flow

```text
User
  ↓
Application
  ↓
API
  ↓
Backend / System
  ↓
API Response
  ↓
Application
  ↓
User
```

**Yahi basic pattern tumhe almost har modern software architecture mein baar-baar dikhega.**


# TMDB API — Practical Example

Ab hum **TMDB (The Movie Database)** ki API ka use karke movie data apni website par fetch karenge.

## TMDB kya hai?

**TMDB (The Movie Database)** ek platform hai jahan movies aur TV shows ka data available hai.

Is data ko hum **TMDB API** ke through apni application mein use kar sakte hain.

---

## TMDB API kaise kaam karegi?

Hum Google par search karenge:

```text
TMDB API
```

TMDB API documentation mein humein different **API Endpoints** milenge.

Har endpoint ka kaam alag hota hai.

```text
Popular Movies      → Popular movies ka data
Top Rated Movies   → Top rated movies ka data
Upcoming Movies    → Upcoming movies ka data
Movie Search       → Kisi movie ko search karna
Movie Details      → Kisi specific movie ki details
```

Iska matlab:

> **Different API endpoints ko call karne par humein different type ka data milta hai.**

---

## Example — Get Top Rated Movies

Maan lo humein **Top Rated Movies** ka data chahiye.

TMDB documentation mein humein **Get Top Rated** naam ka endpoint milega.

Us endpoint ke saath ek **URL** diya hota hai.

Hum us URL ko request bhejenge:

```text
Our Website
     ↓
API Request
     ↓
TMDB API
     ↓
TMDB Server
     ↓
API Response
     ↓
Our Website
```

TMDB server request ko process karega aur **Top Rated Movies ka data** response mein dega.

Response mein humein movie se related information mil sakti hai, jaise:

```text
Movie Name
Movie ID
Rating
Release Date
Poster
Overview
Language
```

Phir hum is received data ko apni website par **Movie Cards** ke form mein display kar sakte hain.

---

## Main Concept

Hum **TMDB ke database ko directly access nahi kar rahe hain.**

Hum:

```text
API Endpoint
     ↓
Request
     ↓
TMDB Server
     ↓
Response
     ↓
Movie Data
```

ke through data receive kar rahe hain.

**API basically hamari website aur TMDB ke server ke beech communication ka medium hai.**
