### What is an API?

> [!NOTE]
> * **API Actually DO** $\rightarrow$ Software ko aapas mein baat karata hai.
> * **Pipeline** $\rightarrow$ Jo information ko **Point A** se **Point B** tak pahunchata hai.


### Real-World Example — IRCTC & Third-Party Apps

> [!NOTE]
> **API (Application Programming Interface)** basically **software ko aapas mein communicate / baat karne ka way provide karti hai.**

---

#### 1. IRCTC & Third-Party Platforms

Maan lo:

* **IRCTC / Railway System** $\rightarrow$ Railway ticketing ka main system
* **MakeMyTrip** $\rightarrow$ Travel booking platform
* **ixigo** $\rightarrow$ Travel booking platform
* **Other authorized platforms** $\rightarrow$ Third-party applications

Ye sab **alag-alag software/platforms** hain.

Lekin jab user **MakeMyTrip** ya **ixigo** par railway ticket search karta hai, toh application ko train ka data kahin se toh lena padega.

**Question:**
> *Train ka data Railway/IRCTC system ka hai, phir MakeMyTrip ya ixigo ko ye data milta kaise hai?*

**Answer — API:**
* IRCTC/Railway system **authorized APIs/integrations** ke through required railway information third-party platforms ko provide karta hai.
* Third-party application **directly database mein jaakar data nahi uthati**.
* Instead, API ke through **request** aur **response** exchange hota hai.

---

#### 2. Request & Response Flow

Suppose user search karta hai: **"Delhi → Mumbai, 25 August"**

```mermaid
sequenceDiagram
    autonumber
    actor User as User
    participant App as MakeMyTrip / ixigo (Frontend)
    participant API as API Gateway
    participant IRCTC as Railway / IRCTC System (Backend)

    User->>App: 1. Search "Delhi → Mumbai, 25 August"
    App->>API: 2. API Request (Query train details)
    API->>IRCTC: 3. Process Request (Check Route, Timing, Seats, Fare)
    IRCTC-->>API: 4. Raw Data (Train Number, Availability, Fare)
    API-->>App: 5. API Response (Structured Data)
    App-->>User: 6. Display train details on UI
```

---

#### 3. Data Access & Security

Ye important concept hai:

> **Railway/IRCTC ka data $\neq$ MakeMyTrip ka own data.**

* Third-party application ko **authorized API access/integration** mil sakta hai.
* API ke through application ko **specific requested information** milti hai.

> [!IMPORTANT]
> Third-party application normally **IRCTC ka entire database download karke apne paas nahi rakh leti**.  
> Instead, API integration ke through **required data/services access** kiye ja sakte hain, subject to the provider's authorization, rules and technical setup.

---

#### 4. Universal System Flow

```mermaid
flowchart TD
    User["User (Client)"] -->|"1. User Request"| App["Application (Frontend)"]
    App -->|"2. API Request"| API["API (Communication Bridge)"]
    API -->|"3. Data & Business Logic"| Backend["Backend / System (Database)"]
    Backend -->|"4. Return Data"| API
    API -->|"5. API Response"| App
    App -->|"6. Show Data"| User
```

**Key Takeaways:**

> • **API** = Bridge / Medium of Communication  
> • **Backend / Database** = Data & Business Logic  
> • **Frontend** = User ko information dikhane wala part  

> [!TIP]
> **Yahi basic pattern tumhe almost har modern software architecture mein baar-baar dikhega.**

#### 3. Example — Get Top Rated Movies

```mermaid
sequenceDiagram
```

#### 4. Main Concept

Hum **TMDB ke database ko directly access nahi kar rahe hain.**

```mermaid
flowchart LR
```

## 📥 JSON Response ko Read Karna

API se aane wale JSON Response ko samajhna kaafi easy hai, bas neeche diye steps follow karo:

1. **API URL** ko Browser me Run karo.
2. Aapko **JSON Response** milega.
3. JSON ko achi tarah samajhne ke liye **JSON Formatter / JSON Viewer** tool use karo.
4. API ka **JSON Response copy** karo.
5. Usse **JSON Formatter** me **paste** karo.
6. **Format / Beautify** button par click karo.
7. Ab JSON properly **structured form** me dikhega.
8. `results` key ke andar aapko **Movies ka actual data** milega.

---
✅ **Note:** Isse aapka JSON data readable aur clean structure me convert ho jaayega, jisse further processing (jaise data extract karna) easy ho jaata hai.

# Loop se DataFrame Banana

Ab hum **loop** ka use karenge.

Har page se movies ka data lenge, required columns select karenge aur ek **DataFrame me append** karte jayenge.

Loop khatam hone ke baad hamare paas ek **single final DataFrame** hoga, jisme approximately **11,154 movies ka data** hoga. 🎬


````python
import requests

url = "https://api.themoviedb.org/3/movie/top_rated"

params = {
    "api_key": "9c3510b4224a5c362c701d95238b950d"
}

response = requests.get(url, params=params)

print(response.status_code)
print(response.text)
````

````python
df = pd.DataFrame()
for i in range(1,559):
    response = requests.get("https://api.themoviedb.org/3/movie/top_rated?api_key=9c3510b4224a5c362c701d95238b950d&language=en-US&page={}".format(i))
    temp_df = pd.DataFrame(response.json()['results']).head(10)[["id", "title", "release_date", "overview", "popularity", "vote_average", "vote_count"]]
    # df.append(temp_df, ignore_index=True)
    df = pd.concat([df, temp_df], ignore_index=True) 
````
