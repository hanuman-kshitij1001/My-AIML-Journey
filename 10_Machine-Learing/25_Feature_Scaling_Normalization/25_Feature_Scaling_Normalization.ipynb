### ***Definition of Normalization-:***

> **Normalization** Machine Learning me data preparation ka ek important step hota hai. Iska main goal dataset ke numerical columns ki values ko ek **common scale (usually 0 to 1)** me transform karna hota hai, bina unke beech ke relative differences (proportions) ko distort kiye ya data ki information loose kiye.

## ***Simpler Terms me Normalization ka Real-world Intuition***

Maan lete hain hum ek aise dataset par kaam kar rahe hain jisme hamare paas do input features hain:
* <span style="color: #2b6cb0; font-weight: bold;">Height</span> (e.g., in centimeters / feet)
* <span style="color: #2b6cb0; font-weight: bold;">Weight</span> (e.g., in kilograms / grams)

Aur hamara task predict karna hai ki <span style="color: #d69e2e; font-weight: bold;">kya wo person specific sports equipment khareedega ya nahi?</span>

---

#### 1. Numerical Quantities ka Core Concept: Magnitude vs Units
Kisi bhi numerical measurement ke andar **do cheezein** hoti hain:
1. **Magnitude (Numeric Value):** E.g., `70`, `70,000`
2. **Unit (Measurement Scale):** E.g., `kg`, `grams`, `tonnes`

Agar ek dataset me weight **$70 \text{ kg}$** likha hai aur dusre me wahi weight **$70,000 \text{ grams}$** likha hai, toh numeric magnitude zameen-aasman ka fark create kar deta hai, jabki physical value bilkul same hai.

---

#### 2. ML Models ke sath Problem (Kyun Units lafda karti hain?)
Machine Learning algorithms ko **units ka matlab samajh nahi aata**; algorithms sirf raw numbers (magnitudes) ko dekhte hain:
* Agar unit **Grams ($70,000$)** hui, toh algorithm use bohot badi aur high-priority value maan lega.
* Agar unit **Kg ($70$)** hui, toh wo comparison me bohot chhota lagega.

---

#### 3. The Core Idea Behind Normalization
> **The Golden Rule in ML:** <span style="color: #c53030; font-weight: bold;">***Units ko eliminate karo aur sirf pure magnitude ko ek common scale par lekar aao.***</span>

* Normalization ka main purpose dataset se **units ke difference ko hata kar** sabhi features ko ek common bounded range (jaise **$0$ to $1$**) me lana hota hai.
* Jab sabhi numerical features unit-free aur common scale par aa jate hain, tab Machine Learning models **biased hue bina consistent aur accurate results** dete hain.

## <span style="color: #2b6cb0;">***Types of Normalization Techniques***</span>

Machine Learning me Normalization ki kai techniques hoti hain. Agar hum **Scikit-Learn (sklearn)** ke official documentation ko dekhein, toh wahan kai tarah ke transformations available hain. 

Lekin hum mainly in **4 most important techniques** par focus karenge:

---

### 1. <span style="color: #d69e2e;">***Min-Max Scaling***</span> (Most Popular)
* Feature Scaling me casually log **Min-Max Scaling ko hi Normalization** bol dete hain kyunki ye sabse widely used method hai.
* **Industry Reality:** Real-world projects me jab bhi aap Normalization lagayenge, lagbhag **$90\%$ times aap Min-Max Scaling ka hi use kar rahe honge**.

### 2. <span style="color: #319795;">***Mean Normalization***</span>
* Isme data ko scale karne ke liye **Mean ($\mu$)** ka use karke center kiya jata hai, jisse values $[-1, 1]$ ya $[-\text{range}, \text{range}]$ me transform hoti hain.

### 3. <span style="color: #805ad5;">***MaxAbs Scaling (Maximum Absolute Scaling)***</span>
* Ye feature ke **maximum absolute value** se divide karke data ko strictly **$[-1, 1]$** ke beech scale karta hai.
* Ye sparse data (jahan bohot saare zeros hote hain) ke liye best maana jata hai.

### 4. <span style="color: #dd6b20;">***Robust Scaling***</span>
* Ye **Median** aur **Interquartile Range (IQR)** par based hota hai.
* **Super Power:** Agar dataset ke andar **Outliers** maujood hain, toh Robust Scaling sabse effective kaam karta hai.
