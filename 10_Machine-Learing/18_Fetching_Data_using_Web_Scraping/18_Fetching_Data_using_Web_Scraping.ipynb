#### 🚫 HTTP 403 – Forbidden

Jab humne AmbitionBox ko request bheji:

```python
response = requests.get('https://www.ambitionbox.com/list-of-companies?page=1')
```

Humein mila:

```text
<Response [403]>
```

### 🤔 Why?

**403 = Forbidden**

Matlab server ne hamari request ko **allow nahi kiya** aur access **deny/block** kar diya.

```text
Request → Server → Access Denied → 403 Forbidden
```


### 🛠️ Solution for 403

403 ko handle karne ke liye hum request ke saath **Browser ka User-Agent** bhej sakte hain.

Isse server ko request ek normal browser request jaisi identify ho sakti hai.

```python
headers = {
    "User-Agent": "Mozilla/5.0"
}

response = requests.get(
    "https://www.ambitionbox.com/list-of-companies?page=1",
    headers=headers
)
```

#### 🔄 Concept

```text
Without User-Agent
Python Request → Server → 403 ❌

With User-Agent
Browser-like Request → Server → Response ✅
```

> **Note:** User-Agent dene se 403 hamesha solve nahi hota. Website ke additional anti-bot protections bhi ho sakte hain.


````python
header = {'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/58.0.3029.110 Safari/537.3'}
response = requests.get('https://www.ambitionbox.com/list-of-companies?page=1',headers=header)
print(response.status_code)
````

## 🥣 BeautifulSoup

Ab hamare paas website ka **HTML content** aa gaya hai, lekin raw HTML mein se required data nikalna difficult hota hai.

Yahan **BeautifulSoup** ka use hota hai.

### 🤔 BeautifulSoup Kya Hai?

**BeautifulSoup** Python ki ek library hai jo **HTML/XML documents ko parse** karke unke andar se required data easily find aur extract karne mein help karti hai.

Simple words mein:

> **BeautifulSoup → HTML ko samajhne aur usme se required data nikalne ka tool hai.**

### 🔄 Basic Flow

```text
Website → requests.get() → response.text → BeautifulSoup → HTML Parse → Required Data Extract
```

### 📦 Import

```python
from bs4 import BeautifulSoup
```

Ab hum HTML ko BeautifulSoup ke through parse kar sakte hain:

```python
soup = BeautifulSoup(response.text, "html.parser")
```

Yahan:

* `response.text` → Website ka HTML content
* `"html.parser"` → HTML ko parse karne ke liye parser
* `soup` → Parsed HTML document

> **In short:** `requests` se HTML **laate hain**, aur `BeautifulSoup` se HTML ke andar se **data nikalte hain**.


````python
soup = BeautifulSoup(webpage, 'lxml')
````

```python
⬆️ soup = BeautifulSoup(webpage, 'lxml')
```

Yahan:

* `webpage` → Website ka HTML content
* `BeautifulSoup()` → HTML ko parse karta hai
* `'lxml'` → HTML ko parse karne ke liye **parser**
* `soup` → Parsed HTML ka object, jisse hum data extract karenge

> **In short:** `BeautifulSoup` HTML ko parse karta hai aur `lxml` us parsing ke liye parser ka kaam karta hai.


````python
print(soup.prettify())
````

