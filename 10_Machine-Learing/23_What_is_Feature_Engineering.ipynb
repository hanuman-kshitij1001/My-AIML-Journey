# Day 23 — What is Feature Engineering?

Feature Engineering means creating, transforming, or selecting features to improve Machine Learning performance.

A feature is an input variable used by a model.


## Examples

Original:

Date = 2026-08-13

Possible features:

- year
- month
- day
- weekday

Original:

height + weight

Possible feature:

BMI


## Common Feature Engineering Techniques

1. Feature Creation
2. Feature Transformation
3. Feature Extraction
4. Feature Selection
5. Binning
6. Encoding
7. Scaling


```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "height": [170, 180, 165, 175],
    "weight": [70, 80, 60, 75]
})

df["BMI"] = df["weight"] / (df["height"] / 100) ** 2

print(df)
```

## Feature Selection

Not every column is useful.

Remove:

- irrelevant features
- duplicate information
- highly problematic features
- data leakage

Good features can improve both model performance and interpretability.
