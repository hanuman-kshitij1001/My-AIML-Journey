# *****FunctionTransformer in Scikit-Learn*****

*Aaj ke lecture me hum Scikit-Learn ka ek aur useful feature transformer padhenge*: **`FunctionTransformer`**.

---

### FunctionTransformer Kya Hai?

* **Core Concept:** Agar aapko apne dataset ke columns par koi **custom mathematical function** ya **Python function** apply karna ho (jaise: $\log$, square root, reciprocal, custom lambda function), toh Scikit-Learn uske liye `FunctionTransformer` provide karta hai.
* **Why Use It in Pipelines?** Normal `df.apply()` ya `np.log()` direct pipeline ke andar compatible nahi hote. `FunctionTransformer` aapke custom mathematical logic ko ek standard Scikit-Learn transformer object (`fit`, `transform`) me convert kar deta hai taaki wo `Pipeline` aur `ColumnTransformer` me bina kisi error ke fit ho sake.

---

### Common Use Cases (Mathematical Transformations)

* **Log Transformation:** Right-skewed data ko normal/Gaussian distribution me convert karne ke liye ($\log(x)$ ya $\log(1 + x)$).
* **Reciprocal Transformation:** Values ka inverse lene ke liye ($1 / x$).
* **Power / Square Root Transformation:** Variance stabilize karne ke liye ($\sqrt{x}$ ya $x^2$).
* **Custom User-Defined Functions:** Kisi bhi custom logic (jaise feature ratios ya custom cleaning) ko Scikit-Learn transformer format me wrap karne ke liye.

# ****Mathematical & Power Transformations****

Aaj ke lecture me hum skewed aur non-normal numerical features ko Gaussian/Normal distribution me convert karne ke liye 5 sabse important transformations padhenge:

---

### ***1. Log Transform (`np.log` / `np.log1p`)***
* **Use Case:** Jab data heavily **right-skewed** ho (yaani tail right side me lambi ho, jaise Salary, House Prices).
* **Formula:** $y = \ln(x)$ ya zero-safe version ke liye $y = \ln(x + 1)$.
* **Condition:** Data strictly **positive ($x > 0$)** hona chahiye.

---

### ***2. Reciprocal Transform (`1 / x`)***
* **Use Case:** Extreme right-skewed data ko compress karne ke liye aur values ke scale ko completely invert karne ke liye.
* **Formula:** $y = \frac{1}{x}$
* **Condition:** Data me zero ($0$) nahi hona chahiye (division by zero error).

---

### ***3. Power Transform (Square / Square Root)***
* **Square Root ($\sqrt{x}$):** Mild right-skewed data ke liye (log transform se thoda halka compression deta hai).
* **Square ($x^2$):** Left-skewed data (negative skewness) ko handle karne ke liye useful hota hai.

---

### ***4. Box-Cox Transform***
* **Use Case:** Ye ek statistical power transformation technique hai jo automatically best parameter ($\lambda$) find karke data ko best possible normal distribution me convert karti hai.
* **Condition:** Ye strictly **strictly positive numbers ($x > 0$)** par hi kaam karta hai. Negative ya zero values par fail ho jata hai.

---

### ***5. Yeo-Johnson Transform***
* **Use Case:** Box-Cox ka modern aur improved version hai.
* **Advantage:** Ye transformation **positive, zero aur negative values ($x \le 0$)** sabhi ke upar perfectly kaam karti hai bina kisi data loss ke.

### ***Question:***
Ye Mathematical / Power Transformations apply karne ke baad actual me data ke saath hota kya hai?

---

### ***Answer:***
Aapke data ka jo original skewed distribution ya **Probability Density Function (PDF)** hota hai, wo ek **Normal Distribution (Gaussian Distribution / Bell Curve)** ke format me convert ya approximate ho jata hai.

*Kyu Ki Jada Tar apko jo data milega wo noral distrubuted nahi honga esliye hamee esse tranform karna padta hai jab ja ke ye hamra data normal distribution me convert ho jata hai*

---

### ****Kyun Zaroori Hai Normal Distribution Me Convert Karna?****

* **Parametric Algorithms Assumption:** Linear Regression, Logistic Regression, Linear Discriminant Analysis (LDA), aur Naive Bayes jaise models assume karte hain ki input features normally distributed hain.
* **Stable Weights & Better Performance:** Normal distribution se outliers ka effect compress ho jata hai, gradient descent fast converge hota hai, aur model ki accuracy improve hoti hai.

# ******Kaise Pata Chale Ki Data Normally Distributed Hai Ya Nahi?******

### *Question*:
Agar mere paas koi data hai, toh kaise pata chalega ki uska distribution normal hai ya nahi? Matlab mujhe use normal distribution me convert karne ki zaroorat hai ya nahi, ye kaise check karein?

---

Iske **2-3 tarike** hote hain:

* **1. Seaborn ka `distplot` / `histplot` (Visualization):**
  * Seaborn ka distribution plot call kar lo. Ye aapko data ka visual curve plot karke de dega, jisse dekh kar andaza lag jata hai ki data kitna normally distributed hai.

* **2. Pandas ka `skew()` Function (Mathematical Metric):**
  * Pandas ke `df['col'].skew()` function ko call kar lo:
    * **$\approx 0$:** Data normally distributed (symmetric) hai.
    * **Positive value ($> 0$):** Right-skewed data hai.
    * **Negative value ($< 0$):** Left-skewed data hai.

* **3. QQ Plot (Quantile-Quantile Plot):**
  * Ye sabse accha aur sabse reliable statistical tarika hota hai normality check karne ka. Agar points red straight line ke upar lie kar rahe hain, toh data normal hai.

## ***1- Log tranform***



* **Direct Concept:** Agar aapko kisi feature/column par Log Transform lagana hai, toh aapko bas us column ki har ek value ka logarithm ($\log$) lena hota hai.
* **Row-wise Application:** Jaise agar kisi column me 100 values hain, toh transformation ke baad wo 100 ki 100 values apne respective $\log(\text{value})$ me convert ho jayengi:

$$x_{\text{new}} = \log(x_{\text{old}})$$

---

### Important Point to Remember (Zero Value Handling)
* $\log(0)$ mathematically **undefined ($-\infty$)** hota hai.
* Isiliye agar dataset me $0$ present ho, toh standard $\log(x)$ ki jagah $\log(1 + x)$ use kiya jata hai (`np.log1p`).

# ***Log Transform: Kab Use Karna Chahiye aur Kab Nahi?***

---

### 1. Negative & Zero Values Par Mat Lagao ❌
* **Reason:** Negative values ($x < 0$) ya zero ($x = 0$) ke upar Log Transform direct apply nahi kiya ja sakta, kyunki $\log(0)$ undefined ($-\infty$) hota hai aur negative numbers ka logarithm real numbers me exist nahi karta.
* **Condition:** Data strictly **positive ($x > 0$)** hona chahiye. Agar dataset me zeros hain, toh $\log(x + 1)$ yaani `np.log1p` use kiya jata hai.

---

### 2. Right-Skewed (Positively Skewed) Data Par Lagao ✅
* **Reason:** Jahan bhi data heavily right-skewed ho (yaani lambi tail right side me ho aur maximum values left side me concentrated hon, jaise *Salary*, *House Prices*, *Page Views*), wahan Log Transform lagane se large values compress ho kar normal distribution (Bell Curve) ke paas aa jati hain.

---

### 3. Kab Avoid Karein? (Left-Skewed Data) ⚠️
* **Negative / Left Skewed Data:** Agar data left-skewed hai (tail left side me hai), toh Log Transform wahan kaam nahi karega, balki skewness ko aur zyada kharab kar dega. Left-skewed data ke liye Power transformations (jaise $x^2$, $x^3$) ya **Yeo-Johnson Transform** ka use karna chahiye.

# ****Quick Notes: Reciprocal, Square & Square Root Transforms****
---

### 1. **Reciprocal Transform ($1 / x$)**
* **Concept:** Ye simple values ka ulta (inverse) kar deta hai: $y = \frac{1}{x}$.
* **Use Case:** Extreme right-skewed data ko compress karne ke liye.
* **Caution:** Data ke andar zero ($0$) nahi hona chahiye (division by zero error).

---

### 2. **Square Transform ($x^2$)**
* **Concept:** Ye har value ka square (power 2) kar deta hai: $y = x^2$.
* **Use Case:** Ye **Left-Skewed (Negatively Skewed)** data ke upar lagaya jata hai taaki data normal distribution me convert ho sake.

---

### 3. **Square Root Transform ($\sqrt{x}$)**
* **Concept:** Ye values ka square root nikalta hai: $y = \sqrt{x}$.
* **Use Case:** Mild right-skewed data ke liye use hota hai (log transform se thoda halka compression deta hai).
* **Note:** Practical machine learning me iska use kaafi kam hota hai (mostly log transform prefer kiya jata hai).
