# Day 17 — Fetching Data from API

API = Application Programming Interface.

APIs allow applications to communicate with each other.

Typical workflow:

Python → API → JSON → DataFrame


```python
import requests

url = "https://jsonplaceholder.typicode.com/users"

response = requests.get(url)

print(response.status_code)

data = response.json()

print(data[:2])
```

## Status Codes

200 → Success

400 → Bad Request

401 → Unauthorized

403 → Forbidden

404 → Not Found

500 → Server Error


```python
import pandas as pd

df = pd.DataFrame(data)

print(df.head())
print(df.columns)
```

## Important

Always check:

response.status_code

before processing API data.
