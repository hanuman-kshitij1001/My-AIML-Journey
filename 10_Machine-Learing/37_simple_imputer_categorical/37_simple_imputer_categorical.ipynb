# ***Handling Missing Values: Categorical Data***

Ab hum yahan pe **Categorical Data ke Missing Values** ko handle karne wale hain.

Categorical columns me missing values ko fill (impute) karne ke liye humare paas main **2 techniques** hoti hain:

1. **Most Frequent Value Imputation (Mode):**  
   Jo category sabse zyada baar repeat ho rahi hoti hai (Mode), usi se saari missing values ko replace kar dete hain.

2. **Missing Category Imputation (Constant Category):**  
   Hum koi existing category use nahi karte, balki missing values ke liye ek **naya category label** create kar dete hain (jaise `"Missing"` ya `"Unknown"`).

---

### ***Technique 1: Most Frequent Value Imputation (Mode)***

Is method me hum categorical column ka **Mode** nikalte hain (matlab wo category jo dataset me sabse zyada frequently aayi hai) aur jahan bhi `NaN` ya missing values hoti hain, wahan us Mode value ko replace kar dete hain.

## ***1- Most Frequent Value Imputation (Mode)***

Agar aapko yaad ho, numerical data me jab missing values 5% se kam hoti hain toh hum **Mean ya Median** se replace kar dete hain. 

Lekin categorical data me numbers nahi hote, isliye yahan Mean aur Median nikalna possible nahi hai. That's why hum yahan missing values ko **MODE** (jo value sabse zyada baar aayi ho) se replace karte hain.

* **Example:**  
  Maan lo ek `City` ka column hai, aur usme sabse zyada baar `Mumbai` aaya hai (Mode = `Mumbai`). Toh jahan bhi missing values hongi, aap unhe seedha `Mumbai` se replace kar doge.

---

### ***Kab Apply Karte Hain?***

1. **MCAR (Missing Completely at Random):**  
   Jab data randomly missing ho.
2. **Mode Dominant Hona Chahiye:**  
   Column ke andar jo mode value hai, wo baaki categories ke mukable bohot zyada baar repeat honi chahiye (clear dominant category honi chahiye).

---

### ***Advantage:***

* **Easy to Implement:**  
  Isko implement karna bohot aasan hai, bas mode calculate karo aur `fillna()` laga do.

---

### ***Disadvantage:***

* **Data Ka Distribution Change Kar Deta Hai:**  
  Ek hi category (Mode) ka count artificially bohot badh jata hai, jisse category ka original distribution/proportion distort ho jata hai.

---

### ****Code Implementation:****

```python
# 1. Pandas ke through:
most_frequent = df['City'].mode()[0]
df['City_imputed'] = df['City'].fillna(most_frequent)

# 2. Scikit-Learn SimpleImputer ke through:
from sklearn.impute import SimpleImputer
imputer = SimpleImputer(strategy='most_frequent')
df['City_imputed'] = imputer.fit_transform(df[['City']])

____

# ***2. Missing Category Imputation***

Ye second type hai categorical missing data ko handle karne ka.

Maan lo aapke paas ek `City` ka column hai jisme alag-alag cities hain jaise **Mumbai (M), Delhi (D), Kolkata (K)** etc., par usme **bohot zyada values missing hain (10% se bhi kaafi zyada)**. 

Toh is case me hum Mode se replacement **nahi** karte hain, kyunki agar itna bada missing data kisi ek category se bhar diya toh distribution poori tarah barbaad ho jayega. Instead, hum kya karte hain:

Hum ek **nayi category bana dete hain called "Missing"** (ya "Unknown").

---

### ***Iska Main Fayda / Goal:***
Bas isse hamare machine learning algorithm ko explicitly pata chal jata hai ki **ye data missing tha**. Poora game bas yahi hai ki algo ko clear signal mil sake ki is observation me city provide nahi ki gayi thi, aur wo is missingness ko bhi as an important pattern treat kare.

---

### ***Code Implementation:***

```python
# 1. Pandas ke through:
df['City_imputed'] = df['City'].fillna('Missing')

# 2. Scikit-Learn SimpleImputer ke through:
from sklearn.impute import SimpleImputer
imputer = SimpleImputer(strategy='constant', fill_value='Missing')
df['City_imputed'] = imputer.fit_transform(df[['City']])

___


<div style="background:#18181b;padding:20px;border-radius:12px;font-family:sans-serif;color:#f4f4f5;border:1px solid #27272a;max-width:850px;margin:auto;">
<div style="text-align:center;margin-bottom:16px;">
<span style="font-size:16px;font-weight:bold;color:#38bdf8;display:block;">Categorical Imputation: 25-Row Dataset Walkthrough</span>
<span style="font-size:12px;color:#a1a1aa;">Missing data > 10% (Yahan 8/25 = 32% NaN hai). Hence, creating "Missing" label.</span>
</div>
<div style="max-height:420px;overflow-y:auto;border:1px solid #3f3f46;border-radius:8px;margin-bottom:16px;">
<table style="width:100%;font-size:11.5px;color:#e4e4e7;border-collapse:collapse;text-align:center;">
<thead style="position:sticky;top:0;background:#27272a;box-shadow:0 1px 2px rgba(0,0,0,0.5);">
<tr style="color:#94a3b8;border-bottom:1px solid #3f3f46;">
<th style="padding:6px;width:10%;">ID</th>
<th style="padding:6px;width:30%;">Original (City)</th>
<th style="padding:6px;width:20%;">Status</th>
<th style="padding:6px;width:40%;border-left:1px solid #3f3f46;">Imputed (City)</th>
</tr>
</thead>
<tbody>
<tr style="border-bottom:1px solid #27272a;"><td style="padding:4px;">1</td><td style="color:#60a5fa;">Mumbai</td><td style="color:#10b981;">Valid</td><td style="border-left:1px solid #3f3f46;color:#60a5fa;">Mumbai</td></tr>
<tr style="background:rgba(239,68,68,0.12);border-bottom:1px solid #27272a;"><td style="padding:4px;">2</td><td style="color:#f87171;font-weight:bold;">NaN</td><td style="color:#f87171;">Missing</td><td style="border-left:1px solid #3f3f46;color:#c084fc;font-weight:bold;">"Missing"</td></tr>
<tr style="border-bottom:1px solid #27272a;"><td style="padding:4px;">3</td><td style="color:#34d399;">Delhi</td><td style="color:#10b981;">Valid</td><td style="border-left:1px solid #3f3f46;color:#34d399;">Delhi</td></tr>
<tr style="border-bottom:1px solid #27272a;"><td style="padding:4px;">4</td><td style="color:#fbbf24;">Kolkata</td><td style="color:#10b981;">Valid</td><td style="border-left:1px solid #3f3f46;color:#fbbf24;">Kolkata</td></tr>
<tr style="background:rgba(239,68,68,0.12);border-bottom:1px solid #27272a;"><td style="padding:4px;">5</td><td style="color:#f87171;font-weight:bold;">NaN</td><td style="color:#f87171;">Missing</td><td style="border-left:1px solid #3f3f46;color:#c084fc;font-weight:bold;">"Missing"</td></tr>
<tr style="border-bottom:1px solid #27272a;"><td style="padding:4px;">6</td><td style="color:#60a5fa;">Mumbai</td><td style="color:#10b981;">Valid</td><td style="border-left:1px solid #3f3f46;color:#60a5fa;">Mumbai</td></tr>
<tr style="border-bottom:1px solid #27272a;"><td style="padding:4px;">7</td><td style="color:#34d399;">Delhi</td><td style="color:#10b981;">Valid</td><td style="border-left:1px solid #3f3f46;color:#34d399;">Delhi</td></tr>
<tr style="background:rgba(239,68,68,0.12);border-bottom:1px solid #27272a;"><td style="padding:4px;">8</td><td style="color:#f87171;font-weight:bold;">NaN</td><td style="color:#f87171;">Missing</td><td style="border-left:1px solid #3f3f46;color:#c084fc;font-weight:bold;">"Missing"</td></tr>
<tr style="border-bottom:1px solid #27272a;"><td style="padding:4px;">9</td><td style="color:#60a5fa;">Mumbai</td><td style="color:#10b981;">Valid</td><td style="border-left:1px solid #3f3f46;color:#60a5fa;">Mumbai</td></tr>
<tr style="border-bottom:1px solid #27272a;"><td style="padding:4px;">10</td><td style="color:#fbbf24;">Kolkata</td><td style="color:#10b981;">Valid</td><td style="border-left:1px solid #3f3f46;color:#fbbf24;">Kolkata</td></tr>
<tr style="background:rgba(239,68,68,0.12);border-bottom:1px solid #27272a;"><td style="padding:4px;">11</td><td style="color:#f87171;font-weight:bold;">NaN</td><td style="color:#f87171;">Missing</td><td style="border-left:1px solid #3f3f46;color:#c084fc;font-weight:bold;">"Missing"</td></tr>
<tr style="border-bottom:1px solid #27272a;"><td style="padding:4px;">12</td><td style="color:#34d399;">Delhi</td><td style="color:#10b981;">Valid</td><td style="border-left:1px solid #3f3f46;color:#34d399;">Delhi</td></tr>
<tr style="border-bottom:1px solid #27272a;"><td style="padding:4px;">13</td><td style="color:#60a5fa;">Mumbai</td><td style="color:#10b981;">Valid</td><td style="border-left:1px solid #3f3f46;color:#60a5fa;">Mumbai</td></tr>
<tr style="background:rgba(239,68,68,0.12);border-bottom:1px solid #27272a;"><td style="padding:4px;">14</td><td style="color:#f87171;font-weight:bold;">NaN</td><td style="color:#f87171;">Missing</td><td style="border-left:1px solid #3f3f46;color:#c084fc;font-weight:bold;">"Missing"</td></tr>
<tr style="border-bottom:1px solid #27272a;"><td style="padding:4px;">15</td><td style="color:#60a5fa;">Mumbai</td><td style="color:#10b981;">Valid</td><td style="border-left:1px solid #3f3f46;color:#60a5fa;">Mumbai</td></tr>
<tr style="border-bottom:1px solid #27272a;"><td style="padding:4px;">16</td><td style="color:#34d399;">Delhi</td><td style="color:#10b981;">Valid</td><td style="border-left:1px solid #3f3f46;color:#34d399;">Delhi</td></tr>
<tr style="background:rgba(239,68,68,0.12);border-bottom:1px solid #27272a;"><td style="padding:4px;">17</td><td style="color:#f87171;font-weight:bold;">NaN</td><td style="color:#f87171;">Missing</td><td style="border-left:1px solid #3f3f46;color:#c084fc;font-weight:bold;">"Missing"</td></tr>
<tr style="border-bottom:1px solid #27272a;"><td style="padding:4px;">18</td><td style="color:#60a5fa;">Mumbai</td><td style="color:#10b981;">Valid</td><td style="border-left:1px solid #3f3f46;color:#60a5fa;">Mumbai</td></tr>
<tr style="border-bottom:1px solid #27272a;"><td style="padding:4px;">19</td><td style="color:#fbbf24;">Kolkata</td><td style="color:#10b981;">Valid</td><td style="border-left:1px solid #3f3f46;color:#fbbf24;">Kolkata</td></tr>
<tr style="background:rgba(239,68,68,0.12);border-bottom:1px solid #27272a;"><td style="padding:4px;">20</td><td style="color:#f87171;font-weight:bold;">NaN</td><td style="color:#f87171;">Missing</td><td style="border-left:1px solid #3f3f46;color:#c084fc;font-weight:bold;">"Missing"</td></tr>
<tr style="border-bottom:1px solid #27272a;"><td style="padding:4px;">21</td><td style="color:#34d399;">Delhi</td><td style="color:#10b981;">Valid</td><td style="border-left:1px solid #3f3f46;color:#34d399;">Delhi</td></tr>
<tr style="border-bottom:1px solid #27272a;"><td style="padding:4px;">22</td><td style="color:#60a5fa;">Mumbai</td><td style="color:#10b981;">Valid</td><td style="border-left:1px solid #3f3f46;color:#60a5fa;">Mumbai</td></tr>
<tr style="background:rgba(239,68,68,0.12);border-bottom:1px solid #27272a;"><td style="padding:4px;">23</td><td style="color:#f87171;font-weight:bold;">NaN</td><td style="color:#f87171;">Missing</td><td style="border-left:1px solid #3f3f46;color:#c084fc;font-weight:bold;">"Missing"</td></tr>
<tr style="border-bottom:1px solid #27272a;"><td style="padding:4px;">24</td><td style="color:#fbbf24;">Kolkata</td><td style="color:#10b981;">Valid</td><td style="border-left:1px solid #3f3f46;color:#fbbf24;">Kolkata</td></tr>
<tr><td style="padding:4px;">25</td><td style="color:#60a5fa;">Mumbai</td><td style="color:#10b981;">Valid</td><td style="border-left:1px solid #3f3f46;color:#60a5fa;">Mumbai</td></tr>
</tbody>
</table>
</div>
<div style="background:#27272a;padding:14px;border-radius:8px;border:1px solid #3f3f46;">
<div style="font-size:12px;font-weight:bold;color:#e4e4e7;margin-bottom:10px;">Aggregated Distribution (25 Rows Total):</div>
<table style="width:100%;font-size:12px;color:#e4e4e7;border-collapse:collapse;">
<tr style="height:26px;"><td style="width:70px;color:#94a3b8;">Mumbai</td><td><div style="background:#18181b;height:14px;border-radius:3px;overflow:hidden;"><div style="background:#38bdf8;width:32%;height:100%;"></div></div></td><td style="width:60px;text-align:right;color:#94a3b8;">8 (32%)</td></tr>
<tr style="height:26px;"><td style="width:70px;color:#94a3b8;">Delhi</td><td><div style="background:#18181b;height:14px;border-radius:3px;overflow:hidden;"><div style="background:#34d399;width:20%;height:100%;"></div></div></td><td style="width:60px;text-align:right;color:#94a3b8;">5 (20%)</td></tr>
<tr style="height:26px;"><td style="width:70px;color:#94a3b8;">Kolkata</td><td><div style="background:#18181b;height:14px;border-radius:3px;overflow:hidden;"><div style="background:#f59e0b;width:16%;height:100%;"></div></div></td><td style="width:60px;text-align:right;color:#94a3b8;">4 (16%)</td></tr>
<tr style="height:26px;"><td style="width:70px;color:#c084fc;font-weight:bold;">Missing</td><td><div style="background:#18181b;height:14px;border-radius:3px;overflow:hidden;"><div style="background:#a855f7;width:32%;height:100%;"></div></div></td><td style="width:60px;text-align:right;color:#c084fc;font-weight:bold;">8 (32%)</td></tr>
</table>
</div>
<div style="font-size:11px;color:#94a3b8;text-align:center;margin-top:10px;">Agar Mode use karte toh 8 missing values jud kar Mumbai 64% ho jata (over-skewed). Naya label banane se Mumbai apna original 32% ratio maintain rakhta hai.</div>
</div>

<div style="background:#18181b;padding:20px;border-radius:12px;font-family:sans-serif;color:#f4f4f5;border:1px solid #27272a;max-width:850px;margin:auto;">
<div style="text-align:center;margin-bottom:16px;">
<span style="font-size:16px;font-weight:bold;color:#38bdf8;display:block;">Missing Indicator Technique: Creating a New Boolean / Flag Column</span>
<span style="font-size:12px;color:#a1a1aa;">Yahan hum original column ko fill bhi karte hain aur ek naya column bana kar track rakhte hain ki konsi value missing thi (1 = Missing, 0 = Not Missing).</span>
</div>
<div style="max-height:420px;overflow-y:auto;border:1px solid #3f3f46;border-radius:8px;margin-bottom:16px;">
<table style="width:100%;font-size:11.5px;color:#e4e4e7;border-collapse:collapse;text-align:center;">
<thead style="position:sticky;top:0;background:#27272a;box-shadow:0 1px 2px rgba(0,0,0,0.5);">
<tr style="color:#94a3b8;border-bottom:1px solid #3f3f46;">
<th style="padding:6px;width:10%;">ID</th>
<th style="padding:6px;width:25%;">Before: City</th>
<th style="padding:6px;width:20%;">Status</th>
<th style="padding:6px;width:25%;border-left:1px solid #3f3f46;">After: City (Imputed)</th>
<th style="padding:6px;width:20%;border-left:1px solid #3f3f46;">New Col: City_NA</th>
</tr>
</thead>
<tbody>
<tr style="border-bottom:1px solid #27272a;"><td style="padding:4px;">1</td><td style="color:#60a5fa;">Mumbai</td><td style="color:#10b981;">Present</td><td style="border-left:1px solid #3f3f46;color:#60a5fa;">Mumbai</td><td style="border-left:1px solid #3f3f46;color:#94a3b8;">0</td></tr>
<tr style="background:rgba(239,68,68,0.12);border-bottom:1px solid #27272a;"><td style="padding:4px;">2</td><td style="color:#f87171;font-weight:bold;">NaN</td><td style="color:#f87171;">Missing</td><td style="border-left:1px solid #3f3f46;color:#60a5fa;">Mumbai (Mode)</td><td style="border-left:1px solid #3f3f46;color:#f87171;font-weight:bold;background:rgba(239,68,68,0.2);">1</td></tr>
<tr style="border-bottom:1px solid #27272a;"><td style="padding:4px;">3</td><td style="color:#34d399;">Delhi</td><td style="color:#10b981;">Present</td><td style="border-left:1px solid #3f3f46;color:#34d399;">Delhi</td><td style="border-left:1px solid #3f3f46;color:#94a3b8;">0</td></tr>
<tr style="border-bottom:1px solid #27272a;"><td style="padding:4px;">4</td><td style="color:#fbbf24;">Kolkata</td><td style="color:#10b981;">Present</td><td style="border-left:1px solid #3f3f46;color:#fbbf24;">Kolkata</td><td style="border-left:1px solid #3f3f46;color:#94a3b8;">0</td></tr>
<tr style="background:rgba(239,68,68,0.12);border-bottom:1px solid #27272a;"><td style="padding:4px;">5</td><td style="color:#f87171;font-weight:bold;">NaN</td><td style="color:#f87171;">Missing</td><td style="border-left:1px solid #3f3f46;color:#60a5fa;">Mumbai (Mode)</td><td style="border-left:1px solid #3f3f46;color:#f87171;font-weight:bold;background:rgba(239,68,68,0.2);">1</td></tr>
<tr style="border-bottom:1px solid #27272a;"><td style="padding:4px;">6</td><td style="color:#60a5fa;">Mumbai</td><td style="color:#10b981;">Present</td><td style="border-left:1px solid #3f3f46;color:#60a5fa;">Mumbai</td><td style="border-left:1px solid #3f3f46;color:#94a3b8;">0</td></tr>
<tr style="border-bottom:1px solid #27272a;"><td style="padding:4px;">7</td><td style="color:#34d399;">Delhi</td><td style="color:#10b981;">Present</td><td style="border-left:1px solid #3f3f46;color:#34d399;">Delhi</td><td style="border-left:1px solid #3f3f46;color:#94a3b8;">0</td></tr>
<tr style="background:rgba(239,68,68,0.12);border-bottom:1px solid #27272a;"><td style="padding:4px;">8</td><td style="color:#f87171;font-weight:bold;">NaN</td><td style="color:#f87171;">Missing</td><td style="border-left:1px solid #3f3f46;color:#60a5fa;">Mumbai (Mode)</td><td style="border-left:1px solid #3f3f46;color:#f87171;font-weight:bold;background:rgba(239,68,68,0.2);">1</td></tr>
<tr style="border-bottom:1px solid #27272a;"><td style="padding:4px;">9</td><td style="color:#60a5fa;">Mumbai</td><td style="color:#10b981;">Present</td><td style="border-left:1px solid #3f3f46;color:#60a5fa;">Mumbai</td><td style="border-left:1px solid #3f3f46;color:#94a3b8;">0</td></tr>
<tr style="border-bottom:1px solid #27272a;"><td style="padding:4px;">10</td><td style="color:#fbbf24;">Kolkata</td><td style="color:#10b981;">Present</td><td style="border-left:1px solid #3f3f46;color:#fbbf24;">Kolkata</td><td style="border-left:1px solid #3f3f46;color:#94a3b8;">0</td></tr>
<tr style="background:rgba(239,68,68,0.12);border-bottom:1px solid #27272a;"><td style="padding:4px;">11</td><td style="color:#f87171;font-weight:bold;">NaN</td><td style="color:#f87171;">Missing</td><td style="border-left:1px solid #3f3f46;color:#60a5fa;">Mumbai (Mode)</td><td style="border-left:1px solid #3f3f46;color:#f87171;font-weight:bold;background:rgba(239,68,68,0.2);">1</td></tr>
<tr style="border-bottom:1px solid #27272a;"><td style="padding:4px;">12</td><td style="color:#34d399;">Delhi</td><td style="color:#10b981;">Present</td><td style="border-left:1px solid #3f3f46;color:#34d399;">Delhi</td><td style="border-left:1px solid #3f3f46;color:#94a3b8;">0</td></tr>
<tr style="border-bottom:1px solid #27272a;"><td style="padding:4px;">13</td><td style="color:#60a5fa;">Mumbai</td><td style="color:#10b981;">Present</td><td style="border-left:1px solid #3f3f46;color:#60a5fa;">Mumbai</td><td style="border-left:1px solid #3f3f46;color:#94a3b8;">0</td></tr>
<tr style="background:rgba(239,68,68,0.12);border-bottom:1px solid #27272a;"><td style="padding:4px;">14</td><td style="color:#f87171;font-weight:bold;">NaN</td><td style="color:#f87171;">Missing</td><td style="border-left:1px solid #3f3f46;color:#60a5fa;">Mumbai (Mode)</td><td style="border-left:1px solid #3f3f46;color:#f87171;font-weight:bold;background:rgba(239,68,68,0.2);">1</td></tr>
<tr style="border-bottom:1px solid #27272a;"><td style="padding:4px;">15</td><td style="color:#60a5fa;">Mumbai</td><td style="color:#10b981;">Present</td><td style="border-left:1px solid #3f3f46;color:#60a5fa;">Mumbai</td><td style="border-left:1px solid #3f3f46;color:#94a3b8;">0</td></tr>
<tr style="border-bottom:1px solid #27272a;"><td style="padding:4px;">16</td><td style="color:#34d399;">Delhi</td><td style="color:#10b981;">Present</td><td style="border-left:1px solid #3f3f46;color:#34d399;">Delhi</td><td style="border-left:1px solid #3f3f46;color:#94a3b8;">0</td></tr>
<tr style="background:rgba(239,68,68,0.12);border-bottom:1px solid #27272a;"><td style="padding:4px;">17</td><td style="color:#f87171;font-weight:bold;">NaN</td><td style="color:#f87171;">Missing</td><td style="border-left:1px solid #3f3f46;color:#60a5fa;">Mumbai (Mode)</td><td style="border-left:1px solid #3f3f46;color:#f87171;font-weight:bold;background:rgba(239,68,68,0.2);">1</td></tr>
<tr style="border-bottom:1px solid #27272a;"><td style="padding:4px;">18</td><td style="color:#60a5fa;">Mumbai</td><td style="color:#10b981;">Present</td><td style="border-left:1px solid #3f3f46;color:#60a5fa;">Mumbai</td><td style="border-left:1px solid #3f3f46;color:#94a3b8;">0</td></tr>
<tr style="border-bottom:1px solid #27272a;"><td style="padding:4px;">19</td><td style="color:#fbbf24;">Kolkata</td><td style="color:#10b981;">Present</td><td style="border-left:1px solid #3f3f46;color:#fbbf24;">Kolkata</td><td style="border-left:1px solid #3f3f46;color:#94a3b8;">0</td></tr>
<tr style="background:rgba(239,68,68,0.12);border-bottom:1px solid #27272a;"><td style="padding:4px;">20</td><td style="color:#f87171;font-weight:bold;">NaN</td><td style="color:#f87171;">Missing</td><td style="border-left:1px solid #3f3f46;color:#60a5fa;">Mumbai (Mode)</td><td style="border-left:1px solid #3f3f46;color:#f87171;font-weight:bold;background:rgba(239,68,68,0.2);">1</td></tr>
<tr style="border-bottom:1px solid #27272a;"><td style="padding:4px;">21</td><td style="color:#34d399;">Delhi</td><td style="color:#10b981;">Present</td><td style="border-left:1px solid #3f3f46;color:#34d399;">Delhi</td><td style="border-left:1px solid #3f3f46;color:#94a3b8;">0</td></tr>
<tr style="border-bottom:1px solid #27272a;"><td style="padding:4px;">22</td><td style="color:#60a5fa;">Mumbai</td><td style="color:#10b981;">Present</td><td style="border-left:1px solid #3f3f46;color:#60a5fa;">Mumbai</td><td style="border-left:1px solid #3f3f46;color:#94a3b8;">0</td></tr>
<tr style="background:rgba(239,68,68,0.12);border-bottom:1px solid #27272a;"><td style="padding:4px;">23</td><td style="color:#f87171;font-weight:bold;">NaN</td><td style="color:#f87171;">Missing</td><td style="border-left:1px solid #3f3f46;color:#60a5fa;">Mumbai (Mode)</td><td style="border-left:1px solid #3f3f46;color:#f87171;font-weight:bold;background:rgba(239,68,68,0.2);">1</td></tr>
<tr style="border-bottom:1px solid #27272a;"><td style="padding:4px;">24</td><td style="color:#fbbf24;">Kolkata</td><td style="color:#10b981;">Present</td><td style="border-left:1px solid #3f3f46;color:#fbbf24;">Kolkata</td><td style="border-left:1px solid #3f3f46;color:#94a3b8;">0</td></tr>
<tr><td style="padding:4px;">25</td><td style="color:#60a5fa;">Mumbai</td><td style="color:#10b981;">Present</td><td style="border-left:1px solid #3f3f46;color:#60a5fa;">Mumbai</td><td style="border-left:1px solid #3f3f46;color:#94a3b8;">0</td></tr>
</tbody>
</table>
</div>
<div style="background:#27272a;padding:14px;border-radius:8px;border:1px solid #3f3f46;">
<div style="font-size:12px;font-weight:bold;color:#e4e4e7;margin-bottom:10px;">New Binary Indicator Column (City_NA Breakdown):</div>
<table style="width:100%;font-size:12px;color:#e4e4e7;border-collapse:collapse;">
<tr style="height:26px;"><td style="width:120px;color:#94a3b8;">0 (Original Present)</td><td><div style="background:#18181b;height:14px;border-radius:3px;overflow:hidden;"><div style="background:#10b981;width:68%;height:100%;"></div></div></td><td style="width:60px;text-align:right;color:#10b981;">17 (68%)</td></tr>
<tr style="height:26px;"><td style="width:120px;color:#f87171;font-weight:bold;">1 (Was Missing)</td><td><div style="background:#18181b;height:14px;border-radius:3px;overflow:hidden;"><div style="background:#ef4444;width:32%;height:100%;"></div></div></td><td style="width:60px;text-align:right;color:#f87171;font-weight:bold;">8 (32%)</td></tr>
</table>
</div>
<div style="font-size:11px;color:#94a3b8;text-align:center;margin-top:10px;">Is technique ka fayda ye hai ki original column me Mode bharne ke baad bhi model ko City_NA column se pata chal jata hai ki ye value artificially fill ki gayi thi.</div>
</div>
