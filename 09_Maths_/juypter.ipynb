```python
import matplotlib.pyplot as plt
import numpy as np
from numpy.random import normal

sample = normal(loc=50, scale=5, size=1000)
```

```python
sample
```

```python
sample.mean()
```

```python
# Ab Haam Ek histo polt karenge jisse hame pata chaelga ki data kaisa hai like normal ya kuch aur
plt.hist(sample, bins=10)
```

```python
sample_mean = sample.mean()
sample_std = sample.std()
```

```python
from scipy.stats import norm
dist = norm(sample_mean, sample_std)
```

```python
values = np.linspace(sample.min(), sample.max(), 100)
```

```python
sample.min()
```

```python
sample.max()
```

```python
probabili_den = [dist.pdf(value) for value in values]
```

```python
#Haam plot karunga 
plt.hist(sample, bins=10, density=True)
plt.plot(values, probabili_den)
```

```python
# Ab Ye Kaam Asasni se kar pate hai agr seborn use kiya hota toh 
import seaborn as sns
sns.distplot(sample)
```
