```python
# 3️⃣ @classmethod — ye asli "alternate constructor" ka tarika hai ⭐
```

```python
class Car:
    def __init__(self, brand, speed):
        self.brand = brand
        self.speed = speed

    @classmethod
    def from_string(cls, data_string):        # ek "alternate constructor"
        brand, speed = data_string.split("-")
        return cls(brand, int(speed))          # cls() = normal __init__ ko hi call karta hai

    @classmethod
    def default_car(cls):                      # ek aur "alternate constructor"
        return cls("Unknown", 0)

c1 = Car("BMW", 220)                    # normal constructor
c2 = Car.from_string("Audi-180")        # string se bana rahe
c3 = Car.default_car()                  # default wala
```

```python
# Inheritance  == = == == = ==
# Inheritance is a mechanism where a new class passes 
# down and copies all the properties and methods from an existing class automatically.
```

```python
# Parent class
class parent_class:
    def __init__(self, surname, propertys, golds, Money):
        self.surname = surname
        self.propertys = propertys
        self.golds = golds
        self.Money = Money
    def bacche_ke_liye(self):
        print("God Bless You")

class Father_Sons(parent_class):
    def __init__(self, surname, propertys, golds, Money, My_jobs, My_Car, My_GF):
        super().__init__(surname, propertys, golds, Money)
        self.My_jobs = My_jobs
        self.My_Car  = My_Car
        self.My_GF  = My_GF

    def son_method(self):
        print(f"The son with surname '{self.surname}' drives a {self.My_Car}!")
result = Father_Sons("Tiwari", "Na", "500g", "10000 Crore", "AI-Engineer", "Defender", "NA")
```

```python
# Running the methods
result.son_method()        # Child's method
result.bacche_ke_liye()    # Inherited Parent's method
```

```python
import pandas as pd
import plotly.express as px

# Sample Data
data = {
    "Student": ["Aman", "Rohit", "Priya", "Neha", "Karan"],
    "Marks": [85, 92, 78, 88, 95]
}

df = pd.DataFrame(data)

# Bar Chart
fig = px.bar(
    df,
    x="Student",
    y="Marks",
    title="Student Marks",
    text="Marks",              # Display value on bars
    color="Marks",             # Color based on marks
    color_continuous_scale="Viridis",
    hover_data={
        "Student": True,
        "Marks": True
    }
)

# Layout
fig.update_layout(
    xaxis_title="Students",
    yaxis_title="Marks",
    template="plotly_white"
)

# Show Graph
fig.show()
```

```python
import pandas as pd
import plotly.express as px

# Sample Data
df = pd.DataFrame({
    "state": [
        "Uttar Pradesh",
        "Maharashtra",
        "Gujarat",
        "Rajasthan",
        "Madhya Pradesh",
        "Tamil Nadu",
        "Karnataka",
        "West Bengal",
        "Bihar",
        "Punjab"
    ],
    "value": [80, 120, 90, 60, 75, 95, 110, 70, 65, 55]
})

# India State Map
fig = px.choropleth(
    df,
    geojson="https://raw.githubusercontent.com/geohacker/india/master/state/india_telengana.geojson",
    featureidkey="properties.ST_NM",
    locations="state",
    color="value",
    color_continuous_scale="Viridis",
    title="India State-wise Data"
)

fig.update_geos(
    fitbounds="locations",
    visible=False
)

fig.update_layout(
    width=800,
    height=800
)

fig.show()
```

```python
"""
India ka Real Interactive Map - Plotly Express se
--------------------------------------------------
Isme:
  - Real geographic map (coastlines, borders, states)
  - Zoom / Pan / Hover jaise interactive features
  - Major cities ke markers bhi plotted hain

Install karne ke liye (agar plotly install nahi hai):
    pip install plotly
"""

import plotly.express as px
import pandas as pd

# ---------------------------------------
# 1. India ke major cities ka sample data
# ---------------------------------------
data = {
    "City": ["New Delhi", "Mumbai", "Bengaluru", "Kolkata", "Chennai",
             "Hyderabad", "Ahmedabad", "Jaipur", "Lucknow", "Bhopal"],
    "Latitude": [28.6139, 19.0760, 12.9716, 22.5726, 13.0827,
                 17.3850, 23.0225, 26.9124, 26.8467, 23.2599],
    "Longitude": [77.2090, 72.8777, 77.5946, 88.3639, 80.2707,
                  78.4867, 72.5714, 75.7873, 80.9462, 77.4126],
    "Population_Lakh": [330, 200, 130, 150, 110, 100, 80, 40, 35, 20]
}

df = pd.DataFrame(data)

# ---------------------------------------
# 2. Real map banayein (scatter_geo)
# ---------------------------------------
fig = px.scatter_geo(
    df,
    lat="Latitude",
    lon="Longitude",
    text="City",
    size="Population_Lakh",       # City size ke hisaab se marker size
    color="Population_Lakh",
    color_continuous_scale="Viridis",
    hover_name="City",
    projection="natural earth",
    title="India ka Real Interactive Map (Plotly Express)"
)

# ---------------------------------------
# 3. Map ko sirf India ke aas-paas focus/zoom karein
# ---------------------------------------
fig.update_geos(
    scope="asia",
    showcountries=True,          # country borders dikhaye
    countrycolor="black",
    showcoastlines=True,         # real coastlines
    coastlinecolor="darkblue",
    showland=True,
    landcolor="rgb(229, 236, 246)",
    showocean=True,
    oceancolor="lightblue",
    showlakes=True,
    lakecolor="lightblue",
    center=dict(lat=22.0, lon=79.0),   # India ke center par focus
    projection_scale=4                # yeh zoom level set karta hai
)

fig.update_layout(
    title_x=0.5,
    height=700,
    margin=dict(l=0, r=0, t=50, b=0)
)

# ---------------------------------------
# 4. Map dikhaye (browser me open hoga)
#    Isme mouse se zoom-in / zoom-out / drag / pan sab
#    interactively kar sakte hain
# ---------------------------------------
fig.show()

# Agar HTML file ke roop me save karna ho to yeh line uncomment karein:
# fig.write_html("india_map.html")
```
